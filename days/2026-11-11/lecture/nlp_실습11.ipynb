# SentenceTransformer를 이용한 문장 유사도 측정

```python
# 필요한 라이브러리 설치 -> pip install sentence-transformers -> colab 설치 완료
```

```python
# 필요한 모델 클래스 임포트
from sentence_transformers import SentenceTransformer
```

## 영어 문장 유사도 측정

```python
# 유사도 측정을 할 텍스트 데이터 생성
en_sentences = ["What should I do to be a great scientist?",
                "How can I be a good scientist?"]
```

```python
# 임베딩 모델 생성
"""
# SentenceTransformer 클래스
  1) pipeline() 함수와 기능 비슷
  2) 사전 학습된 모델 이름 입력 -> 해당 모델 다운로드
  3) 임베딩 할 텍스트에 대한 전처리(토큰화 -> 임베딩) 자동 처리
"""
model_name='all-MiniLM-L12-v2'
model_eng = SentenceTransformer(model_name)
```

```python
# 텍스트 -> 임베딩 백터(입력 2개 -> 행렬)

## 기능 ㄴ-> 모델 클래스의 전용 함수 -> model.encode([text, ...])
en_embeddings = model_eng.encode(en_sentences)

## 결과 확인
print(f'영어 텍스트에 대한 임베딩의 결과: \n{en_embeddings}')
print('-'*80)
print(f'영어 텍스트에 대한 임베딩의 결과(모양): {en_embeddings.shape}')
```

```python
# 코사인 유사도 측정 -> 전용 함수 similarity() 함수 호출
en_sim = model_eng.similarity(en_embeddings, en_embeddings)
print(f'영어 문장 간의 코사인 유사도: \n{en_sim}')
```

## 한글 문장 유사도 측정

```python
# 유사도 측정을 할 텍스트 데이터 생성

## 매우 유사한 두 문장
ko_sentences1 = ["직원이 무단 퇴사를 했는데 손해 배상 청구할 수 있나요?",
                "무단 퇴사한 직원에 대한 손해 배상 청구가 가능한가요?"]

## 서로 무관한 두 문장
ko_sentences2 = ["직원이 무단 퇴사를 했는데 손해 배상 청구할 수 있나요?",
                 "오늘도 날씨가 매우 덥고 소나기가 내리는 등 동남아 날씨같아요."]
```

```python
# 사전 학습된 다국어 임베딩 모델 생성
model_name='paraphrase-multilingual-MiniLM-L12-v2'
model_multi = SentenceTransformer(model_name)
```

```python
# 문장 임베딩 행렬 생성 -> 전용 함수 -> model.encode() 함수

## 매우 유사한 두 문장에 대한 임베딩 행렬 생성
ko_embeddings1 = model_multi.encode(ko_sentences1)

## 서로 무관한 두 문장에 대한 임베딩 행렬 생성
ko_embeddings2 = model_multi.encode(ko_sentences2)

## 결과 확인
print(f'매우 유사한 두 문장에 대한 임베딩의 결과(모양): {ko_embeddings1.shape}')
print('-'*80)
print(f'서로 무관한 두 문장에 대한 임베딩의 결과(모양): {ko_embeddings2.shape}')
```

```python
# 임배딩 행렬 -> 두 문장 간의 코사인 유사도 측정 -> 전용 함수 -> model.similarity() 함수 호츨

## 매우 유사한 두 문장간의 코사인 유사도
ko_sim1 = model_multi.similarity(ko_embeddings1, ko_embeddings1)
print(f'매우 유사한 두 문장간의 코사인 유사도: \n{ko_sim1}')

print('-'*80)

## 서로 무관한 두 문장간의 코사인 유사도
ko_sim2 = model_multi.similarity(ko_embeddings2, ko_embeddings2)
print(f'서로 무관한 두 문장간의 코사인 유사도: \n{ko_sim2}')
```

```python
# 한국어로 사전 학습된 임베딩 모델 생성
model_name='ddobokki/klue-roberta-base-nli-sts'
model_kor = SentenceTransformer(model_name)
```

```python
# 한글로 학습된 모델 -> 문장 임베딩 -> 전용 함수 -> model.encode()

## 매우 유사한 두 문장에 대한 임베딩
ko_embeddings3 = model_kor.encode(ko_sentences1)

## 서로 무관한 두 문장에 대한 임베딩
ko_embeddings4 = model_kor.encode(ko_sentences2)

## 결과 확인
print(f'매우 유사한 두 문장에 대한 임베딩의 결과(모양): {ko_embeddings3.shape}')
print('-'*80)
print(f'서로 무관한 두 문장에 대한 임베딩의 결과(모양): {ko_embeddings4.shape}')
```

```python
# 한글로 학습된 모델 -> 코사인 유사도 측정 -> 전용 함수 -> model.similarity() 함수

## 매우 유사한 두 문장간의 코사인 유사도
ko_sim3 = model_kor.similarity(ko_embeddings3, ko_embeddings3)
print(f'매우 유사한 두 문장간의 코사인 유사도: \n{ko_sim3}')

print('-'*80)

## 서로 무관한 두 문장간의 코사인 유사도
ko_sim4 = model_kor.similarity(ko_embeddings4, ko_embeddings4)
print(f'서로 무관한 두 문장간의 코사인 유사도: \n{ko_sim4}')
```

## 영어/한글/중국어/스페인어 문장 유사도 측정

```python
# 유사도 측정을 위한 텍스트 데이터 생성
sentences = ['What should I do to be a great scientist?',
             '훌륭한 과학자가 되려면 어떻게 해야 할까요?',
             '怎样才能成为一名优秀的科学家？',
             '¿Qué debo hacer para convertirme en un gran científico?']
```

```python
# 사전 학습된 다국어 모델 생성
model_name= 'intfloat/multilingual-e5-large-instruct'
model_multi1 = SentenceTransformer(model_name)
```

```python
# 4개 언어 텍스트 -> 임베딩 -> 전용 함수 -> model.encode()
multi_embeddings = model_multi1.encode(sentences)
print(f'4개 언어 텍스트에 대한 임베딩의 결과(모양): {multi_embeddings.shape}')
```

```python
# 4개 언어 텍스트 -> 코사인 유사도 -> 전용 함수 -> model.similarity()
multi_sim = model_multi1.similarity(multi_embeddings, multi_embeddings)
print(f'4개 언어 텍스트에 대한 코사인 유사도: \n{multi_sim}')
```
