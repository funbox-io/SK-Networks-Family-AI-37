### 0-5. 데이터 분할과 모델 평가 (p.10~11)

학습에 데이터를 **100% 다 쓰면 안 된다.** 학습에 쓴 데이터로 시험을 보면 점수가 부풀려진다.

| 데이터 | 비율 | 용도 |
|--------|------|------|
| **Train** (학습용) | **60~80%** | 모델 학습 |
| **Test** (평가용) | **40~20%** | 학습하지 않은 **미지의 데이터**로 평가 |

평가 척도 — 분류: **accuracy**(정확도) · **recall**(재현율) · **precision**(정밀도) · **f1 score** / 회귀: **MSE · RMSE**

````python
import pandas as pd

# 실습 데이터 : 붓꽃(iris) 150송이 (노트북과 같은 폴더)
path = "./iris_dataset.csv"
df = pd.read_csv(path)

print("모양 :", df.shape)
print(df.head())

print("-" * 60)

# 레이블(정답) 컬럼의 범주별 개수
labels = df['label'].value_counts()
print(labels)
````

````python
# 특성(X)과 레이블(y) 나누기 → 학습용 80% / 평가용 20% 로 분할 (교재 p.10)
from sklearn.model_selection import train_test_split

X = df.drop(columns='label')     # 특성 4개
y = df['label']                  # 정답

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,       # 평가용 20%
    random_state=42,     # 섞는 순서 고정 → 실행할 때마다 같은 결과
    stratify=y           # 품종 비율(1:1:1)을 train/test 에 똑같이 유지
)

print("학습용 :", X_train.shape, "/ 평가용 :", X_test.shape)
print(y_test.value_counts())
````

> **📝 고친 부분** (`notbook.py` 개념 메모)
> - "지도 학습(**딥러닝**)" → 지도 학습은 딥러닝이 아니다. **정답이 있는 학습 방식**이고, 머신러닝·딥러닝 모두 지도 학습을 할 수 있다.
> - "분류(**문자열**) / 회귀(**정수**)" → 분류는 **범주**(0/1 처럼 숫자로 바꿔도 범주), 회귀는 **연속적인 숫자**(소수 포함)를 예측한다.
> - "학습할 때 100% 전부 사용하면 안 된다. 60~80%" → 교재 p.10 표와 `train_test_split()` 예제로 정리

## 1. 특성 공학(Feature Engineering) 개요 (교재 p.12)

**분석가의 전문 지식과 통찰력을 데이터에 직접 주입해, 모델이 더 똑똑한 예측을 하도록 만드는 과정.** 모델 성능을 높이기 위한 데이터 가공이다.

| 작업 | 대상 | 방법 | 단원 |
|------|------|------|------|
| ① **파생 변수 생성** | 기존 컬럼들 | 조합해 더 의미 있는 새 컬럼 만들기 | 1 |
| ② **범주형 변수 변환** (Encoding) | 성별 · 채널 · 연령대 같은 **글자** | 원-핫 인코딩 · 레이블 인코딩 | 2, 3 |
| ③ **연속형 변수 변환** (Scaling) | 구매액 · 방문횟수 같은 **숫자** | 표준화 · 정규화 | 4 |
| (+) **구간화** (Binning) | 연속형 숫자 → 범주 | `pd.cut()` | 5 |

### 파생 변수 예시 — ROAS (광고비 대비 수익률)

`Cost`(광고비)와 `Revenue`(매출)는 따로 보면 "많이 썼다 / 많이 벌었다"뿐이지만,

## 2. 원-핫 인코딩(One-Hot Encoding) (교재 p.13~15)

**범주형 컬럼의 고유한 값 n개 → 새 컬럼 n개**를 만들고, 해당하는 컬럼에만 **1**, 나머지에는 **0** 을 넣는 방법.
예) `Channel` 의 `Google` · `Facebook` · `Naver` → `[1 0 0]` · `[0 1 0]` · `[0 0 1]`

```python
pd.get_dummies(data=df, dtype=int)                            # 문자열 컬럼 전부
pd.get_dummies(data=df, columns=['col1', 'col2'], dtype=int)  # 지정한 컬럼만
```

| 매개변수 | 뜻 |
|---------|-----|
| `columns` | 인코딩할 컬럼 목록. 생략하면 **문자열(범주형) 컬럼 전부** |
| `dtype=int` | 결과를 `True/False`(bool) 대신 **1/0**(int) 으로 |

| 장점 | 단점 |
|------|------|
| 범주끼리 **크기 순서가 생기지 않는다** → 구분이 명확 | 범주 수만큼 **컬럼이 늘어난다** |

````python
# 교재 p.13 예시 : 요일 컬럼
df_day = pd.DataFrame({'id': [0, 1, 2, 3, 4],
                       'day': ['월', '월', '화', '수', '화']})

# day 컬럼만 원-핫 인코딩 → day_수, day_월, day_화 컬럼 생성
print(pd.get_dummies(data=df_day, columns=['day'], dtype=int))
````

````python
# 교재 p.14 예시 : 컬럼 2개 + 결측치(NaN)
import numpy as np

fruit = pd.DataFrame({'name': ['apple', 'banana', 'cherry', 'durian', np.nan],
                      'color': ['red', 'yellow', 'red', 'green', np.nan]})

# 전체 인코딩 : name, color 둘 다 → NaN 행은 모든 컬럼이 0
print(pd.get_dummies(fruit, dtype=int))

print("-" * 60)

# 특정 컬럼만 인코딩 : color 는 그대로 남는다
print(pd.get_dummies(fruit, columns=['name'], dtype=int))
````

## 3. 레이블 인코딩(Label Encoding) (교재 p.16~17)

**범주형 문자열을 0 부터 n-1 까지의 연속된 숫자**로 바꾸는 방법. 컬럼 수는 그대로다.

```python
df.replace({'문자열1': 숫자1, '문자열2': 숫자2, ...})
# 예) df = df.replace({'no': 0, 'yes': 1})
```

| 장점 | 단점 |
|------|------|
| **간결**하다 — 컬럼이 늘지 않는다 | 바꿀 값을 **직접(수동으로)** 적어야 한다 |
| | 숫자에 **크기 순서**가 생긴다 (0 < 1 < 2) — 순서가 없는 범주에는 주의 |

````python
# 교재 p.16 예시 : 은행 고객 데이터의 yes/no 컬럼
df_bank = pd.DataFrame({
    'age': [58, 36, 25, 53, 24],
    'job': ['management', 'technician', 'blue-collar', 'technician', 'technician'],
    'default': ['no', 'no', 'no', 'no', 'no'],
    'housing': ['yes', 'yes', 'yes', 'no', 'yes'],
    'loan': ['no', 'yes', 'no', 'no', 'yes'],
})

# 바꿀 컬럼만 골라 no → 0, yes → 1
yes_no = ['default', 'housing', 'loan']
df_bank[yes_no] = df_bank[yes_no].replace({'no': 0, 'yes': 1}).astype(int)
print(df_bank)
````

````python
# iris 실습 : 품종 3개 → 0, 1, 2
label_map = {'setosa': 0, 'versicolor': 1, 'virginica': 2}

df_label = df.replace(label_map)
print("replace 후 label 자료형 :", df_label['label'].dtype)   # 숫자처럼 보여도 object 일 수 있다

# 컬럼 하나만 바꿀 때는 map() 이 깔끔하다 → 처음부터 정수형
df_label['label'] = df['label'].map(label_map)
print("map 후 label 자료형     :", df_label['label'].dtype)

df_label.head()
````

> **📝 고친 부분**
> - `df.replace({...})` 를 **DataFrame 전체**에 쓰면 다른 컬럼에 같은 글자가 있을 때 함께 바뀐다 → 바꿀 컬럼을 지정하거나 `map()` 사용
> - `replace` 로 바꾼 컬럼은 값이 숫자여도 **자료형이 `object` 로 남을 수 있다** (pandas 2.x 에서는 FutureWarning, 3.x 에서는 object 그대로).
>   모델에 넣기 전 `.astype(int)` 로 바꾸거나 `map()` 을 쓴다.
> - `display(df_label)` → `head()`
> - 필기 요약 "원-핫(장점: 구분이 쉽다 / 단점: 컬럼이 늘어난다), 레이블(장점: 간결하다 / 단점: 수동으로 수정한다)" 은 표로 정리하고, 레이블 인코딩의 **크기 순서 문제**를 추가

## 4. 스케일링(Scaling) (교재 p.18~20)

**특성들의 분포(평균, 표준편차)나 범위를 똑같이 맞춰 주는 작업.**

**왜 필요한가?** — 나이는 20~45, 연봉은 1,000~6,000 처럼 특성마다 범위가 다르면,
모델은 **숫자가 큰 특성**을 더 중요하게 여겨 제대로 학습되지 않을 수 있다.

| 종류 | 변환 결과 | 공식 | scikit-learn |
|------|----------|------|--------------|
| **표준화** (Standardization) | 평균 **0**, 표준편차 **1** | $\dfrac{x - \mu}{\sigma}$ ($\mu$ 평균, $\sigma$ 표준편차) | `StandardScaler` |
| **정규화** (Normalization) | 범위 **0 ~ 1** | $\dfrac{x - x_{min}}{x_{max} - x_{min}}$ | `MinMaxScaler` |

scikit-learn 스케일러 사용법은 항상 같다.

```python
scaler = MinMaxScaler()                    # ① 도구 만들기
result = scaler.fit_transform(df)          # ② 학습(fit: 최소·최대 / 평균·표준편차 계산) + 변환(transform)
                                           #    결과는 numpy 2차원 배열 → 필요하면 DataFrame 으로 다시 감싼다
```

````python
# 스케일링 실습 데이터 : 나이와 연봉(만원) — 범위가 크게 다르다
data = {
    "나이": [22, 24, 45, 38, 29],
    "연봉": [1000, 3500, 6000, 4300, 5200],
}
df_dic = pd.DataFrame(data)
print(df_dic)

print("-" * 60)
print(df_dic.describe().loc[['mean', 'std', 'min', 'max']])
````

````python
# 정규화 : 최솟값 0, 최댓값 1
from sklearn.preprocessing import MinMaxScaler

minmax_scal = MinMaxScaler()                      # 정규화 도구 생성
data_minmax = minmax_scal.fit_transform(df_dic)   # 결과 : numpy 2차원 배열
print(type(data_minmax))

# DataFrame 으로 다시 감싸기 (인덱스 유지, 컬럼 이름 지정)
df_minmax = pd.DataFrame(data_minmax, index=df_dic.index, columns=['나이_정규화', '연봉_정규화'])
print("정규화\n", df_minmax.round(2))

print("-" * 60)

# 공식으로 직접 계산해서 비교 : (x - min) / (max - min)
by_hand = (df_dic - df_dic.min()) / (df_dic.max() - df_dic.min())
print("공식과 같은가? :", np.allclose(by_hand, data_minmax))
````

````python
# 표준화 : 평균 0, 표준편차 1
from sklearn.preprocessing import StandardScaler

std_scal = StandardScaler()                       # 표준화 도구 생성
data_std = std_scal.fit_transform(df_dic)

df_std = pd.DataFrame(data_std, index=df_dic.index, columns=['나이_표준화', '연봉_표준화'])
print("표준화\n", df_std.round(2))

print("-" * 60)

# 확인 : 평균 ≈ 0, 표준편차 = 1
print("평균 :", df_std.mean().round(2).tolist())
print("표준편차(ddof=0) :", df_std.std(ddof=0).round(2).tolist())

# 공식으로 직접 계산 : (x - 평균) / 표준편차  ※ StandardScaler 는 모집단 표준편차(ddof=0)를 쓴다
by_hand = (df_dic - df_dic.mean()) / df_dic.std(ddof=0)
print("공식과 같은가? :", np.allclose(by_hand, data_std))
````

> - 컬럼 이름 `연령(인원)` 인데 결과 컬럼은 `연봉_정규화` → 원본 컬럼도 **`연봉`** 으로 통일

> - 결과 변수 `data_minmax` 를 DataFrame 으로 덮어쓰던 것 → 배열 `data_minmax` / 표 `df_minmax` 로 이름 분리
> - 공식으로 직접 계산해 scikit-learn 결과와 같은지 확인하는 코드 추가 (표준화는 **ddof=0** 이라는 점 포함)
> - 오타 `# Scling` → Scaling, 메모 "큰수 쪽으로 연산하려고 할 때" → 범위가 큰 특성이 모델을 **지배하지 않도록** 맞추는 것

## 5. 구간화(Binning) (교재 p.21~22)

**연속형 수치 데이터를 특정 구간으로 나눠 범주형 데이터로 바꾸는 작업.** 예) 나이 → 어린이 · 청소년 · 청년 …

```python
pd.cut(x, bins, labels)
```

| 매개변수 | 뜻 |
|---------|-----|
| `x` | 나눌 데이터 (Series) |
| `bins` | 구간의 **경계값** 리스트 |
| `labels` | 각 구간의 이름 (**구간 개수 = 경계값 개수 − 1**) |

> 구간은 기본으로 **왼쪽 열림, 오른쪽 닫힘** `(a, b]` 이다.
> `bins=[0, 12, 19, 49, 70, 100]` → `(0,12]` 어린이 · `(12,19]` 청소년 · `(19,49]` 청년 · `(49,70]` 중년 · `(70,100]` 노년

````python
data = {
    'age': [1, 10, 15, 13, 21, 23, 37, 31, 43, 80, 61, 20, 41, 32, 100],
    'gender': ['male', 'male', 'female', 'female', 'female', 'male', 'male', 'female',
               'female', 'female', 'male', 'male', 'female', 'female', 'female'],
}
df_age = pd.DataFrame(data)

# bins : 구간 경계값 (6개 → 구간 5개)
age_bins = [0, 12, 19, 49, 70, 100]

# labels : 구간 이름 (5개)
age_label = ['어린이', '청소년', '청년', '중년', '노년']

# 새로운 컬럼에 추가 → df['새 컬럼'] = 데이터
df_age['age_step'] = pd.cut(x=df_age['age'], bins=age_bins, labels=age_label)

print(df_age)

print("-" * 60)
print(df_age['age_step'].value_counts(sort=False))   # 구간 순서대로 개수
````

> **📝 고친 부분**
> - 오타 `매걔변수` · `매깨변수` → 매개변수, 변수 이름 `df_dicrm` → `df_age`
> - 메모 "bins(구간 **크기** 리스트), labels … **균등하게**" → `bins` 는 구간의 **경계값**이고, 직접 주면 **균등하지 않아도** 된다.
>   (`bins=5` 처럼 **정수**를 주면 그때 최솟값~최댓값을 균등하게 5등분한다)
> - 경계 주의 : 나이 **0** 은 `(0, 12]` 에 들지 않아 **NaN** 이 된다 → 0 을 포함하려면 `include_lowest=True`
> - 결과를 구간별 개수로 요약하는 코드 추가 (청년 8명이 가장 많다)

## 6. `df.query()` — 조건으로 행 추출 (교재 p.23~25)

**데이터프레임에서 특정 조건을 만족하는 행만 추출(필터링)하는 함수.**
조건문을 **문자열**로 쓰기 때문에, 여러 조건을 조합해도 코드가 복잡해지지 않고 SQL 의 `WHERE` 처럼 직관적이다.

```python
df.query('조건문 문자열')
```

| 비교 연산자 | 뜻 | | 논리 연산자 | 문법 | 뜻 |
|-----------|-----|---|-----------|------|-----|
| `>` · `<` | 왼쪽 값이 크다 · 작다 | | `and` / `&` | `a and b` | 논리곱 — **둘 다** 참일 때 참 |
| `>=` · `<=` | 크거나 같다 · 작거나 같다 | | `or` / `\|` | `a or b` | 논리합 — **하나만** 참이어도 참 |
| `==` · `!=` | 같다 · 같지 않다 | | `not` / `~` | `not x` | 논리부정 — 참·거짓을 뒤집음 |

> **문자열 값은 따옴표 안에 따옴표** — 바깥을 `"…"` 로 감쌌으면 안쪽 값은 `'IT'` 처럼 작은따옴표로.
> **컬럼 이름은 대소문자를 구분**한다 — `age >= 28` 은 `Age` 컬럼을 찾지 못해 에러.

````python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 22, 28],
    'Salary': [5000, 7000, 4500, 8000, 6000],
    'Dept': ['HR', 'IT', 'HR', 'IT', 'Marketing'],
}
df_emp = pd.DataFrame(data)     # 원본은 그대로 두고, 결과는 새 변수에

# 비교 연산 : 나이 28세 이상
print(df_emp.query("Age >= 28"))
print("=" * 40)

# 문자열 비교 : 부서가 IT (값은 작은따옴표로)
print(df_emp.query("Dept == 'IT'"))
print("=" * 40)

# 논리곱 & : 나이 25세 이상 그리고 급여 6000 이상
print(df_emp.query("Age >= 25 & Salary >= 6000"))
print("=" * 40)

# 논리합 | : 부서가 HR 이거나 급여 7000 이상
print(df_emp.query("Dept == 'HR' | Salary >= 7000"))
````

````python
# 교재 p.24 의 and / or / not 키워드도 query 안에서 그대로 쓸 수 있다
print(df_emp.query("Age >= 25 and Salary >= 6000"))       # & 와 같다
print("=" * 40)

# not : 부서가 IT 가 아닌 사람
print(df_emp.query("not Dept == 'IT'"))                   # Dept != 'IT' 와 같다
print("=" * 40)

# 파이썬 변수를 조건에 쓰려면 @ 를 붙인다
min_salary = 6000
print(df_emp.query("Salary >= @min_salary"))
````

````python
# 광고 데이터에 적용
df_ad = pd.read_csv("./ad_performance.csv")

# 채널이 Facebook
df_fb = df_ad.query("Channel == 'Facebook'")

# 매출이 1,500,000 초과 → 매출 오름차순 정렬
df_rev = df_ad.query("Revenue > 1500000").sort_values(by="Revenue")

# 다중 조건 : 채널이 Google 이고 매출이 1,500,000 초과
df_google = df_ad.query("Channel == 'Google' & Revenue > 1500000").sort_values(by="Revenue")

print("Facebook :", len(df_fb), "행")
print("매출 150만 초과 :", len(df_rev), "행")
print("Google & 매출 150만 초과 :", len(df_google), "행")
df_google.head()
````

> **📝 고친 부분**
> - `df1 = df1.query(...)` 처럼 **원본 변수를 결과로 덮어쓰고**, 다음 조건마다 `pd.DataFrame(data=data)` 를 다시 만들던 것 → 원본 `df_emp` 를 한 번만 만들고 결과만 출력
> - 주석 "대소 문자(에러 원인!)" → 무엇의 대소문자인지 설명 추가 (**컬럼 이름**)
> - 교재 p.24 `and` · `or` · `not` 키워드, 변수 참조 `@` 예제 추가
> - 광고 데이터 — 주석은 "매출이 150000" 인데 바로 위 조건은 1,500,000 이었다. 매출 최솟값이 약 10만 원이라 **15만 원 초과는 거의 전부**가 걸려 의미가 없다 → **1,500,000** 으로 통일
> - 변수 이름 `df1` · `df2` · `df3` → `df_fb` · `df_rev` · `df_google`

## 7. `pd.concat()` — 데이터프레임 병합 (교재 p.26~28)

**여러 데이터프레임을 하나로 이어 붙이는 함수.**

```python
pd.concat([df1, df2, df3], ignore_index=True)    # axis=0 (기본값)
```

| 매개변수 | 뜻 |
|---------|-----|
| `[df1, df2, …]` | 붙일 DataFrame 들을 **리스트**로 |
| `axis=0` (기본) | **공통된 컬럼을 기준으로 y축(아래) 방향**으로 병합 — 행이 늘어난다 |
| `axis=1` | x축(옆) 방향으로 병합 — 컬럼이 늘어난다 |
| `ignore_index=True` | 원래 인덱스를 버리고 **0, 1, 2 … 새로 매김** |

````python
# 교재 p.26 예시 : 같은 컬럼(A~D)을 가진 DataFrame 3개
df1 = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'],
                    'B': ['B0', 'B1', 'B2', 'B3'],
                    'C': ['C0', 'C1', 'C2', 'C3'],
                    'D': ['D0', 'D1', 'D2', 'D3']},
                   index=[0, 1, 2, 3])

df2 = pd.DataFrame({'A': ['A4', 'A5', 'A6', 'A7'],
                    'B': ['B4', 'B5', 'B6', 'B7'],
                    'C': ['C4', 'C5', 'C6', 'C7'],
                    'D': ['D4', 'D5', 'D6', 'D7']},
                   index=[4, 5, 6, 7])

df3 = pd.DataFrame({'A': ['A8', 'A9', 'A10', 'A11'],
                    'B': ['B8', 'B9', 'B10', 'B11'],
                    'C': ['C8', 'C9', 'C10', 'C11'],
                    'D': ['D8', 'D9', 'D10', 'D11']},
                   index=[8, 9, 10, 11])

# axis=0 방향으로 병합 (기본값) → 4행 + 4행 + 4행 = 12행
df_concat = pd.concat([df1, df2, df3], ignore_index=True)
df_concat
````

````python
# 광고 데이터 : 6단원의 조건 결과 3개를 하나로 합치기
df_all = pd.concat([df_fb, df_rev, df_google], ignore_index=True).sort_values(by="Revenue")
print("합친 행 수 :", len(df_all))

# 주의! 두 조건에 동시에 걸린 행은 '두 번' 들어간다
#   (예: Facebook 이면서 매출 150만 초과인 행 → df_fb 와 df_rev 양쪽에 있음)
print("중복된 행 수 :", df_all.duplicated().sum())

# 중복 제거
df_all_unique = df_all.drop_duplicates()
print("중복 제거 후 :", len(df_all_unique))
````

> **📝 고친 부분**
> - 변수 이름 `df_cunt` → `df_all`
> - 조건 결과를 그대로 이어 붙이면 **여러 조건에 동시에 걸린 행이 중복**된다 → `duplicated()` 로 개수를 확인하고 `drop_duplicates()` 로 제거
>   (원본 필기 출력에서 매출 1,998,748 · 1,999,100 행이 **두 번씩** 보이던 이유)

## 8. 오늘 정리

| 작업 | 목적 | 코드 |
|------|------|------|
| 데이터 분할 | 학습용 60~80% / 평가용 40~20% | `train_test_split(X, y, test_size=0.2)` |
| 파생 변수 | 컬럼 조합으로 새 의미 | `df['ROAS'] = df['Revenue'] / df['Cost']` |
| 원-핫 인코딩 | 범주 → 0/1 컬럼 n개 | `pd.get_dummies(df, columns=[...], dtype=int)` |
| 레이블 인코딩 | 범주 → 0 ~ n-1 | `df.replace({...})` · `df['col'].map({...})` |
| 정규화 | 범위 0 ~ 1 | `MinMaxScaler().fit_transform(df)` |
| 표준화 | 평균 0, 표준편차 1 | `StandardScaler().fit_transform(df)` |
| 구간화 | 숫자 → 범주 | `pd.cut(x, bins, labels)` |
| 조건 추출 | 조건을 만족하는 행만 | `df.query("A >= 1 & B == 'x'")` |
| 병합 | 여러 DataFrame 이어 붙이기 | `pd.concat([df1, df2], ignore_index=True)` |

- 머신러닝은 **정답이 있으면 지도 학습**(범주 → 분류, 숫자 → 회귀), **없으면 비지도 학습**(클러스터링).
- 모델은 숫자만 이해한다 → 글자는 **인코딩**, 범위가 다른 숫자는 **스케일링**.
- 스케일러 결과는 **numpy 배열** — DataFrame 으로 다시 감쌀 때 `index` · `columns` 를 지정한다.
- `replace` 로 인코딩한 컬럼은 자료형이 `object` 로 남을 수 있으니 확인한다.
- `concat` 으로 조건 결과를 합치면 **중복**이 생길 수 있다.
