## 2. 직선 그래프 — `px.line()` (교재 p.46)

**두 개 이상의 데이터 포인트를 직선으로 연결**해 연속적인 변화와 추세를 보여 준다.

```python
px.line(data_frame, x, y, color)
```

| 하이퍼파라미터 | 뜻 |
|---------------|-----|
| `data_frame` | 데이터프레임 |
| `x`, `y` | 시각화할 컬럼 |
| `color` | 새로운 변수를 추가해 **색상으로 구분** |

**목표** — 요금의 총합(`total_bill`)이 커질수록 팁(`tip`)도 커지는가?

> `px.line()` 은 **행 순서대로** 점을 잇는다. x 축 컬럼으로 **먼저 오름차순 정렬**해야 선이 엉키지 않는다.

````python
# x 축 : 요금의 총합(total_bill) → 크기 순(오름차순)으로 정렬
df_sorted = df.sort_values(by='total_bill')
print(df_sorted.head())

print('-' * 80)

# time 컬럼의 항목별 빈도수 → value_counts()
counts_time = df['time'].value_counts()
print(f'식사 시간대별 빈도수: \n{counts_time}')
````

````python
# 식사 시간대(time)별로 색을 나눠 그래프 생성 및 출력
fig1 = px.line(data_frame=df_sorted, x='total_bill', y='tip', color='time',
               title='요금의 총합과 팁 — 점심 / 저녁')
fig1.show()

# 그래프를 HTML 파일로 저장 → 브라우저에서 그대로 열리고, 확대·툴팁도 동작한다
# (plotly.js 가 통째로 들어가 파일이 약 4MB 가 된다)
fig1.write_html("line.html")
````

> **📝 고친 부분**
> - `display(df_sorted)` → `head()`
> - 빈도수 확인과 그래프를 한 셀에서 섞어 찍던 것 → 데이터 확인 셀 / 그래프 셀로 분리
> - `print('-'*80)` 뒤 빈 줄 여러 개 정리, 그래프에 `title` 추가
> - `write_html("line.html")` 이 만드는 파일이 이 폴더의 [`line.html`](./line.html) 이다.
>   용량을 줄이려면 `fig1.write_html("line.html", include_plotlyjs="cdn")` — plotly.js 를 인터넷에서 불러오게 해 수 KB 로 줄어든다.

## 3. 막대 그래프 — `px.bar()` (교재 p.47)

**범주별 수치 크기를 막대로 표현**하는 그래프. 막대의 길이로 크기를 비교한다.

```python
px.bar(data_frame, x, y)                       # 수직 막대 그래프
px.bar(data_frame, x, y, orientation='h')      # 수평 막대 그래프
```

| 하이퍼파라미터 | 수직형 | 수평형 |
|---------------|--------|--------|
| `x` | **범주형** 컬럼 | **수치형** 컬럼 |
| `y` | **수치형** 컬럼 | **범주형** 컬럼 |
| `orientation` | `'v'` (기본값) | `'h'` |

### 3-1. 그래프용 표 만들기 — `groupby` → `reset_index` → `rename`

**목표** — 요일별 요금의 총합(`total_bill`) **평균** 비교

`px.bar()` 에는 **"요일 | 평균 요금"** 두 컬럼짜리 DataFrame 을 넘기는 게 편하다. 그래서 3단계로 표를 만든다.

| 단계 | 코드 | 결과 |
|------|------|------|
| ① 그룹별 평균 | `df.groupby(by='day')['total_bill'].mean()` | **Series** (요일이 행 인덱스) |
| ② 인덱스 → 컬럼 | `.reset_index()` | **DataFrame** (`day` 컬럼 + 0,1,2… 번호 인덱스) |
| ③ 컬럼 이름 바꾸기 | `.rename(columns={'total_bill': 'avg_total_bill'})` | 이름만 봐도 **평균**인 걸 알 수 있게 |

````python
# ① 통계 분석 : 요일별 그룹핑(groupby) + total_bill 컬럼 지정 + 평균(mean)
day_mean = df.groupby(by='day')['total_bill'].mean()
print(f'요일 별 평균 요금: \n{day_mean}')

print('-' * 80)

# ② Series → DataFrame 변환 (행 인덱스 → 컬럼 + 번호 행 인덱스 추가)
df_day_mean = day_mean.reset_index()
print(df_day_mean)

print('-' * 80)

# ③ 컬럼 이름 변경 → rename()
df_day_mean1 = df_day_mean.rename(columns={'total_bill': 'avg_total_bill'})
df_day_mean1
````

### 3-2. 수직 막대 그래프

`groupby` 결과는 요일이 **알파벳 순**(Fri · Sat · Sun · Thur)이다.
요일 순서대로 보려면 `category_orders={'컬럼': [순서 리스트]}` 로 x 축 항목 순서를 직접 정한다.

````python
# 기본 수직 막대 그래프 (x : 범주형, y : 수치형)
fig2 = px.bar(data_frame=df_day_mean1, x='day', y='avg_total_bill')
fig2.show()
````

````python
# x 축 항목을 요일 순(목, 금, 토, 일)으로 정렬 → category_orders
fig3 = px.bar(
    data_frame=df_day_mean1,
    x='day',
    y='avg_total_bill',
    category_orders={'day': ['Thur', 'Fri', 'Sat', 'Sun']}
)
fig3.show()
````

````python
# 요일 순 정렬 + 요일별로 색상 구분 → color
fig4 = px.bar(
    data_frame=df_day_mean1,
    x='day',
    y='avg_total_bill',
    category_orders={'day': ['Thur', 'Fri', 'Sat', 'Sun']},
    color='day',
    title='요일별 평균 요금'
)
fig4.show()
````

### 3-3. 수평 막대 그래프 — `orientation='h'`

수평형은 **x 와 y 의 역할이 바뀐다** — x 에 수치형, y 에 범주형.
항목 이름이 길거나 항목 수가 많을 때 읽기 편하다.

`text_auto` 는 막대 위에 **값을 글자로** 표시한다. `True` 면 소수점이 길게 나오니 `'.2f'` 처럼 형식을 주면 깔끔하다.

````python
# 수평 막대 그래프 : x 에 수치형, y 에 범주형 + orientation='h'
fig5 = px.bar(
    data_frame=df_day_mean1,
    x='avg_total_bill',
    y='day',
    category_orders={'day': ['Thur', 'Fri', 'Sat', 'Sun']},
    color='day',
    orientation='h',
    text_auto='.2f',      # 막대에 값 표시 (소수 둘째 자리까지)
    title='요일별 평균 요금 (수평)'
)
fig5.show()
````

> **📝 고친 부분**
> - `display(...)` → `print()` / 셀 마지막 줄에 변수 이름만 두기
> - `fig2` · `fig3` · `fig4` 를 한 셀에서 연달아 그리던 것 → **한 셀에 그래프 하나**로 나눠 차이가 바로 보이게
> - `text_auto=True` → `text_auto='.2f'` (`17.682741935483875` 처럼 길게 찍히던 값을 `17.68` 로)
> - 결과 해석 : 평균 요금은 **일(21.41) > 토(20.44) > 목(17.68) > 금(17.15)** — 주말 손님이 더 많이 쓴다

## 4. 원 그래프 — `px.pie()` (교재 p.48)

**범주형 데이터의 비율을 부채꼴 조각으로 표현**하는 그래프. 조각 크기로 **전체 중 비율**을 비교한다.

```python
px.pie(data_frame, names, values, hole)
```

| 하이퍼파라미터 | 뜻 |
|---------------|-----|
| `names` | **범주형** 컬럼 — 조각의 이름 |
| `values` | **수치형** 컬럼 — 조각의 크기 |
| `hole` | 가운데 구멍 크기 **0 ~ 1** (0 이면 원, 0.4 처럼 주면 **도넛** 차트) |

**목표** — 고객의 성별 빈도수를 **비율**로 비교

`value_counts(normalize=True)` 는 개수 대신 **비율(합계 1)** 을 돌려준다.

````python
# 통계 분석 : sex 컬럼 인덱싱 + value_counts(normalize=True) → 비율
sex_ratio = df['sex'].value_counts(normalize=True)
print(f'고객의 성별 비율: \n{sex_ratio}')

print('-' * 80)

# Series → DataFrame 변환 (행 인덱스 → 컬럼 + 번호 행 인덱스 추가)
df_sex_ratio = sex_ratio.reset_index()
df_sex_ratio
````

````python
# 도넛 차트 : names(범주형) = 조각 이름, values(수치형) = 조각 크기
fig6 = px.pie(
    data_frame=df_sex_ratio,
    names='sex',
    values='proportion',
    hole=0.4,
    title='고객 성별 비율'
)
fig6.show()
````

> **📝 고친 부분**
> - `reset_index()` 결과의 비율 컬럼 이름 `proportion` 은 **pandas 2.0 부터**다. 그 전 버전은 이름이 달라 `values='proportion'` 에서 에러가 난다.
>   버전과 상관없이 쓰려면 `sex_ratio.rename('ratio').reset_index()` 처럼 **이름을 직접 정해** 두면 된다.
> - 사실 `px.pie()` 는 **비율이 아니라 개수를 넣어도** 알아서 퍼센트로 바꿔 그린다 → `df['sex'].value_counts().reset_index()` 의 `count` 를 `values` 로 줘도 같은 그림
> - 결과 해석 : 남성 **64.3%** · 여성 **35.7%**

## 5. 히스토그램 — `px.histogram()` (교재 p.49)

**수치형 데이터의 분포를 구간(bin)별 빈도 막대로 표현**하는 그래프.
값을 일정한 구간으로 나누고, 각 구간에 데이터가 **몇 개 쌓였는지**를 막대 높이로 보여 준다.

```python
px.histogram(data_frame, x, y=None, nbins, color)     # 수직 히스토그램
px.histogram(data_frame, x=None, y, color)            # 수평 히스토그램
```

| 하이퍼파라미터 | 뜻 |
|---------------|-----|
| `x` (또는 `y`) | 분포를 볼 **수치형** 컬럼 — `y` 에 주면 수평 히스토그램 |
| `nbins` | 구간(bin) 개수. **지정하지 않으면 자동**으로 정해진다 |
| `color` | 범주형 컬럼으로 **그룹을 색으로 구분** |

> **막대 그래프 vs 히스토그램** — 막대 그래프의 x 축은 **범주**(요일)이고, 히스토그램의 x 축은 **수치 구간**(10~15달러)이다. 그래서 히스토그램 막대는 서로 **붙어** 있다.

### 5-1. 수직 히스토그램 — 구간 개수 `nbins` · 경계선 `update_traces`

**목표** — `total_bill` 금액 구간별 빈도수 비교

````python
# 구간 개수 자동 (nbins 미지정)
fig7 = px.histogram(data_frame=df, x='total_bill')
fig7.show()
````

````python
# 구간을 50개로 잘게 나누기 → 막대가 가늘어지고 들쭉날쭉해진다
fig7_50 = px.histogram(data_frame=df, x='total_bill', nbins=50)
fig7_50.show()
````

````python
# 막대가 붙어 있어 경계가 안 보인다 → 모든 막대에 일괄적으로 검은색 경계선 넣기
# update_traces() 는 이미 만든 그래프(fig7)를 '그 자리에서' 고친다
fig7.update_traces(marker_line_color='black', marker_line_width=1)
fig7.show()
````

### 5-2. 그룹별 히스토그램 — `color` + `barmode`

`color='time'` 을 주면 점심/저녁 두 그룹의 분포를 한 그래프에 그린다. 두 그룹을 **어떻게 배치할지**는 `barmode` 로 정한다.

| `barmode` | 배치 | 언제 |
|-----------|------|------|
| `'relative'` (기본값) | 위로 **쌓기** | 전체 분포 + 그룹 구성 |
| `'overlay'` | **겹치기** (+ `opacity` 로 반투명) | 두 분포의 **모양** 비교 |
| `'group'` | 구간마다 **옆으로 나란히** | 구간별 개수 직접 비교 |

````python
# color='time' 추가 → 식사 시간대별, 금액 구간별 빈도수 (기본 : 위로 쌓기)
fig8 = px.histogram(data_frame=df, x='total_bill', color='time')
fig8.update_traces(marker_line_color='black', marker_line_width=1)
fig8.show()
````

````python
# 겹쳐서 표시 → barmode='overlay' + opacity(투명도 0~1) 로 뒤쪽 막대도 보이게
fig9 = px.histogram(
    data_frame=df,
    x='total_bill',
    color='time',
    barmode='overlay',
    opacity=0.7
)
fig9.update_traces(marker_line_color='black', marker_line_width=1)
fig9.show()
````

````python
# 분리해서 표시 → barmode='group' (구간마다 점심/저녁 막대를 나란히)
fig10 = px.histogram(
    data_frame=df,
    x='total_bill',
    color='time',
    barmode='group'
)
fig10.update_traces(marker_line_color='black', marker_line_width=1)
fig10.show()
````

### 5-3. 수평 히스토그램 — `y` 에 컬럼 지정 (교재 p.49)

## 6. 박스 플롯 — `px.box()` (교재 p.50) ✚ 필기에 없어 추가

**데이터의 분포를 최솟값 · Q1 · 중앙값 · Q3 · 최댓값으로 요약**하고, **이상치(outlier)를 식별**하는 그래프.

```python
px.box(data_frame, x, y, color)
```

| 하이퍼파라미터 | 수직형 | 수평형 |
|---------------|--------|--------|
| `x` | **범주형** 컬럼 | **수치형** 컬럼 |
| `y` | **수치형** 컬럼 | **범주형** 컬럼 |
| `color` | 범주형 컬럼으로 그룹 색 구분 | |

### 박스 읽는 법

```
   ●            ← 이상치 : 수염 밖에 따로 찍힌 점
   ┬            ← 최댓값 (이상치를 뺀 범위 안에서)
   │            ← 수염
 ┌───┐          ← Q3 (75%)
 │   │
 ├───┤          ← Q2 = 중앙값 (50%)
 │   │
 └───┘          ← Q1 (25%)
   │
   ┴            ← 최솟값
```

- **박스 길이 = IQR(Q3 − Q1)** → 가운데 50% 데이터가 퍼진 정도
- 수염은 박스에서 **IQR × 1.5** 까지만 뻗는다. 그 밖의 값은 **점(이상치)** 으로 따로 찍힌다
- Day 018 의 `describe()` 에 나온 `25%` · `50%` · `75%` 를 **그림으로** 그린 것이다

````python
# 수직 박스 플롯 : x 범주형(day), y 수치형(total_bill)
fig12 = px.box(
    data_frame=df,
    x='day',
    y='total_bill',
    category_orders={'day': ['Thur', 'Fri', 'Sat', 'Sun']},
    title='요일별 요금 분포'
)
fig12.show()
````

````python
# color='time' 추가 → 요일마다 점심/저녁 박스를 나란히
fig13 = px.box(
    data_frame=df,
    x='day',
    y='total_bill',
    color='time',
    category_orders={'day': ['Thur', 'Fri', 'Sat', 'Sun']},
    title='요일 · 식사 시간대별 요금 분포'
)
fig13.show()
````

````python
# 수평 박스 플롯 : x 수치형(tip), y 범주형(smoker) — 역할이 바뀐다
fig14 = px.box(data_frame=df, x='tip', y='smoker', color='smoker',
               title='흡연석 여부별 팁 분포')
fig14.show()
````

### 이상치를 숫자로 확인하기

그래프에 찍힌 점이 정말 이상치 기준을 넘는지 pandas 로 직접 계산해 본다.

| 기준 | 식 |
|------|-----|
| 위쪽 경계 | `Q3 + 1.5 × IQR` |
| 아래쪽 경계 | `Q1 − 1.5 × IQR` |

````python
# total_bill 의 사분위수와 IQR
q1 = df['total_bill'].quantile(0.25)
q3 = df['total_bill'].quantile(0.75)
iqr = q3 - q1

upper = q3 + 1.5 * iqr      # 위쪽 경계
lower = q1 - 1.5 * iqr      # 아래쪽 경계
print(f'Q1 = {q1:.2f}, Q3 = {q3:.2f}, IQR = {iqr:.2f}')
print(f'정상 범위 : {lower:.2f} ~ {upper:.2f}')

print('-' * 80)

# 경계 밖의 행 = 이상치
outliers = df[(df['total_bill'] > upper) | (df['total_bill'] < lower)]
print('이상치 개수 :', len(outliers))
outliers.sort_values(by='total_bill')
````

> 결과 해석
> - 요금의 정상 범위는 약 **-2.82 ~ 40.30달러**, 40.30달러를 넘는 **9건**이 이상치로 찍힌다 (아래쪽은 없음)
> - 요일별 중앙값은 **일 19.63 > 토 18.24 > 목 16.20 > 금 15.38** — 막대 그래프(평균)와 순서가 같다
> - 평균보다 중앙값이 조금씩 작다 → 고액 손님(이상치) 몇 명이 **평균을 위로 끌어올렸다**. 이럴 때는 중앙값이 더 믿을 만하다

## 7. 산점도 — `px.scatter()` (교재 p.51) ✚ 필기에 없어 추가

**두 변수 간의 상관관계를 점으로 표현**하는 그래프. 행 하나가 점 하나다.

```python
px.scatter(data_frame, x, y, color)
```

| 하이퍼파라미터 | 뜻 |
|---------------|-----|
| `x`, `y` | 관계를 볼 두 **수치형** 컬럼 |

|------------|-----|
| 오른쪽 **위로** 모인다 | 양의 상관 |
| 오른쪽 **아래로** 모인다 | 음의 상관 |
| 아무 방향 없이 흩어진다 | 상관 거의 없음 |

````python
# 산점도 : 요금의 총합(x)과 팁(y)의 관계, 식사 시간대별 색 구분
fig15 = px.scatter(
    data_frame=df,
    x='total_bill',
    y='tip',
    color='time',
    title='요금의 총합과 팁의 관계'
)
fig15.show()

print('상관계수 total_bill - tip :', round(df['total_bill'].corr(df['tip']), 2))
````

점들이 오른쪽 위로 모여 있다 → **양의 상관(0.68)**. 계산 금액이 클수록 팁도 커진다.

## 8. 히트맵 — `px.imshow()` (교재 p.52) ✚ 필기에 없어 추가

**두 변수 간의 상관계수 크기를 색의 농도로 표현**하는 그래프.
변수가 많아도 **어떤 쌍이 강하게 연결돼 있는지** 한눈에 보인다.

```python
px.imshow(data_frame, text_auto, color_continuous_scale)
```

| 하이퍼파라미터 | 뜻 |
|---------------|-----|
| `data_frame` | **상관 행렬** — `df.corr()` 의 결과 |
| `text_auto` | 칸 안에 값 표시. `'.2f'` → 소수 둘째 자리까지 |
| `color_continuous_scale` | 색 테마. 예) `'RdBu_r'` (음수 파랑 ↔ 양수 빨강), `'Viridis'` |

먼저 Day 018 에서 배운 `corr()` 로 **상관 행렬**을 만든다. 문자열 컬럼은 상관계수를 구할 수 없으니 `numeric_only=True`.

````python
# 상관 행렬 만들기 (수치형 컬럼끼리)
corr_matrix = df.corr(numeric_only=True)
corr_matrix
````

````python
# 히트맵 : 상관 행렬을 색 농도로
fig16 = px.imshow(
    corr_matrix,
    text_auto='.2f',                    # 칸 안에 상관계수 (소수 둘째 자리)
    color_continuous_scale='RdBu_r',    # 음수 파랑 ↔ 양수 빨강
    zmin=-1, zmax=1,                    # 색 범위를 상관계수 범위(-1 ~ 1)에 고정
    title='tips 수치형 컬럼 상관계수'
)
fig16.show()
````

> `zmin=-1, zmax=1` 은 교재에 없지만 꼭 같이 쓰는 걸 추천한다.
> 안 주면 **데이터에 있는 값 범위**(여기서는 0.49 ~ 1.0)에 맞춰 색이 늘어나서, 0.49 가 진한 파랑(=강한 음의 상관처럼)으로 보인다.

결과 해석
- `total_bill` – `tip` **0.68** : 가장 강한 양의 상관
- `total_bill` – `size` **0.60** : 일행이 많을수록 계산 금액도 크다
- `tip` – `size` **0.49** : 일행 수와 팁도 어느 정도 같이 움직인다
- 대각선은 자기 자신과의 상관이라 항상 **1.00**

| 그래프 | 함수 | 핵심 하이퍼파라미터 | 데이터 준비 |
