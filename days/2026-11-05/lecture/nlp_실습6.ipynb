```python
#필요한 라이브러리 임포
import pandas as pd

# 파일 경로 지정
file_path = '/content/movies_metadata.csv'

# 데이터 불러오기
df = pd.read_csv(file_path)

# 데이터 첫 5행 출력
display(df.head())
```

#데이터프레임 전처리

```python
# overview, title 컬럼 팬시 인덱싱 (이중 대괄호 사용)
df1 = df[['overview', 'title']]
display(df1)
```

#누락 데이터 처리

```python
# 각 컬럼별 누락 데이터의 수 확인 (전처리 전)
num_nulls1 = df1.isnull().sum()
print("결측치 제거 전:\n", num_nulls1)

# 결측치 제거 후 원본 df1에 반영
df2 = df1.dropna(ignore_index=True)

# 각 컬럼별 누락 데이터의 수 확인 (전처리 후)
#num_nulls2 = df1.isnull().sum()
#print("\n결측치 제거 후:\n", num_nulls2)

display(df2)
```

```python
#필요한 클래스 임퐅트
from sklearn.feature_extraction.text import CountVectorizer

#모델 CountVectorizer 호출 임배딩 생성
cv = CountVectorizer(stop_words='english')

#
```

```python
print(df2.head())
```

```python
#영어 텍스트 데이터 생성단어 사전 및 임베딩 행렬 생성
#from sklearn.feature_extraction.text import CountVectorizer

# CountVectorizer 객체 생성 (정의 누락 방지)
#cv = CountVectorizer()

#데이터 수 -> 1만개 축소(행: 0~9999)
df3 = df2[0:9999]
#임베딩 리스트
text_list = df3['overview'].tolist()

# 전체를 출력하면 출력이 너무 길어지므로 앞의 3개만 확인합니다.
print("=== 추출된 텍스트 샘플(3개) ===")
print(text_list[:3])

cv.fit(text_list)
movie_vecab = cv.vocabulary_

# 생성된 단어 사전 중 앞의 10개 아이템만 확인합니다.
print("\n=== 생성된 단어 사전 일부(10개) ===")
print(dict(list(movie_vecab.items())[:10]))
print("=="*50)

movie_vecab2 = cv.transform(text_list).toarray()
print(f"overview 컬럼:{movie_vecab2.shape}")
```

```python
# 1만개 영화 줄거리(overview) -> 코사인 유사도 측정

from sklearn.metrics.pairwise import cosine_similarity

# 코사인 유사도 측정
cos_sim = cosine_similarity(movie_vecab2, movie_vecab2)

# 결과 확인
print("코사인 유사도 측정:", cos_sim)
```

#영화 추천 서비스 구현

```python
# 코사인 유사도 켤과 값 -> 데이터프레임 생성
#print(df3['title'].tolist())

# 컬럼과 행 인덱스 설정
index = df3['title'].values
coml = df3['title'].values

df_sim = pd.DataFrame(cos_sim, index=index, columns=coml)
display(df_sim)
```

```python
#특정 영화 지정-> 특정 커럼 지정

#컬럼의 값 -> 크기 순으로 정렬(내림차순)
n = 10
#
top_10 = df_sim['Toy Story'].sort_values(ascending=False)[1:n+1]
print(top_10)
```

```python
#특정 영화 기준 -> 줄거리가 유사한 영화 추천
def top_n_func(k,title):

  if title in df_sim.columns:
    top_n = df_sim[title].sort_values(ascending=False)[1:k+1]
    return top_n
  else:
      print("존재하지 않는 영화 제목이오니, 확인 후 다시 입력을 해주세요")
```

```python
num = 10
name = 'GoldenEye' #제목 설정

#함수 호출
top_10 = top_n_func(num,name)
print(top_10)
```
