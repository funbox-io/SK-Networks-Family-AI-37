# 수치 미분

```python
import numpy as np

# 수치 미분 함수 정의
# f: 미분 대상
# x: 목표 지정
def number_diff(f,x):
    h = 1/10000
    output= (f(x+h)-f(x-h))/(2*h)
    return output
```

```python
def func(x):
    return x**2

# 목표 지점
target_x = 3.0

# 해석적 미분 df/dx = 2x)
output1 = 2*target_x
print("미분을 계산 target_x 값:",target_x)
print("해석한 값:",output1)

output2 = number_diff(func,target_x)
print("수치 미분:",output2)
```
