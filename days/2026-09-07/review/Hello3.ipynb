## 1. 리스트(list)

### 1-1. 중첩 리스트와 인덱싱 / 슬라이싱

- **인덱싱**: 값 하나를 꺼낸다. `리스트[번호]`
- **슬라이싱**: 범위를 잘라 **리스트**로 꺼낸다. `리스트[시작:끝]` (끝 번호는 포함하지 않음)
- 리스트 안에 리스트를 넣으면 **중첩 리스트**가 되고, `[i][j]` 처럼 두 번 인덱싱한다.

```python
# 학생 이름만 담은 리스트 (문자열)
student_names = ["김진호", "이지현", "박서준"]

# 학생 점수만 담은 리스트 (정수, 실수가 섞여도 됨)
student_scores = [88, 95.5, 72]

# 이름 리스트와 점수 리스트를 한 번에 묶은 중첩 리스트
score_table = [student_names, student_scores]

print("성적표 전체 :", score_table)
print("인덱싱      :", score_table[0][0])   # 0번 리스트(이름)의 0번 값 -> "김진호"
print("슬라이싱    :", score_table[:-1])    # 마지막 리스트(점수)를 뺀 나머지
```

### 1-2. 리스트 값 수정

리스트는 **변경 가능(mutable)** 한 자료형이라서 인덱스로 값을 바꿀 수 있다.

```python
# 월~목 판매량
weekly_sales = [120, 250, 310, 400]
print("수정 전 :", weekly_sales)

weekly_sales[3] = 500      # 3번 인덱스(목요일) 판매량을 500으로 정정
print("수정 후 :", weekly_sales)
```

## 2. 딕셔너리(dict)

### 2-1. 딕셔너리란?

- **매핑(mapping)** 자료형: `key`와 `value`가 **1:1로 대응**한다.
- 형태: `{key: value, key: value, ...}`
- `key`는 **중복 불가**, `value`는 **중복 가능**하다.
- 리스트가 *순서(번호)* 로 값을 찾는다면, 딕셔너리는 *이름(key)* 으로 값을 찾는다.

### 2-2. 딕셔너리 생성

```python
# 1) 빈 딕셔너리 (앞으로 회원 정보를 채워 넣을 주소록)
address_book = {}
print("빈 주소록 :", address_book)

# 2) key = 문자열, value = 문자열 / 숫자
member = {'name': '이지현', 'age': 20}
print("회원 한 명 :", member)

# 3) key = 숫자도 사용 가능 (회원번호로 이름 찾기)
member_by_id = {1: '김진호', 2: '이지현'}
print("번호별 회원 :", member_by_id)

# 4) value 자리에 리스트를 넣어 여러 명을 한 번에 저장
member_group = {'name': ['김진호', '이지현'], 'age': [30, 20]}
print("회원 그룹 :", member_group)

# 5) 이후 예제에서 계속 사용할 고객 정보
customer = {'name': '김진호', 'age': 30, 'phone': '010-1234-5678'}
print("고객 정보 :", customer)
```

### 2-3. 값(value) 가져오기

| 방법 | 형태 | 없는 key를 찾으면 |
|------|------|------------------|
| 대괄호 | `딕셔너리[key]` | **에러(KeyError)** 발생 |
| get() | `딕셔너리.get(key)` | `None` 반환 (에러 없음) |

```python
# 대괄호 방식 : key 값으로 value 호출
customer_name = customer['name']
customer_age = customer['age']

# get() 방식 : key가 없어도 에러가 나지 않는다
customer_phone = customer.get('phone')

print("고객 이름 :", customer_name)
print("고객 나이 :", customer_age)
print("고객 번호 :", customer_phone)

# 없는 key를 get()으로 찾으면 None이 나온다
print("고객 주소 :", customer.get('address'))
```

### 2-4. 데이터 추가 / 수정

- 형태: `딕셔너리[key] = value`
- **없는 key** 이면 → 새로 추가
- **있는 key** 이면 → 기존 값을 덮어쓴다

```python
# 'birth'는 없던 key라서 새로 추가된다
customer['birth'] = '08/09'
print("생일 추가 :", customer)

# 'age'는 이미 있는 key라서 값이 덮어써진다
customer['age'] = 31
print("나이 수정 :", customer)
```

```python
# 노래 한 곡의 정보를 하나씩 채워 나가기
song_info = {'title': 'NoNoNo'}
print("최초 상태 :", song_info)

song_info['singer'] = '지코'
print("가수 추가 :", song_info)

song_info['genre'] = '힙합'
print("장르 추가 :", song_info)

song_info['release_date'] = '2020.01.13'
print("발매일 추가 :", song_info)
```

### 2-5. 데이터 삭제

- `del 딕셔너리[key]` : key 하나를 삭제
- `딕셔너리.clear()` : 전체 삭제(빈 딕셔너리가 됨)

```python
# key 하나만 삭제
del song_info['release_date']
print("발매일 삭제 :", song_info)

# 전체 삭제
song_info.clear()
print("전체 삭제   :", song_info)
```

### 2-6. 딕셔너리 연습 문제

```python
# 1번 - 딕셔너리를 만들고 전체 출력하기
my_profile = {'name': '김진호', 'job': '프로그래머'}
print("1번 문제 :", my_profile)

# 2번 - key로 value를 꺼내 문장에 넣기
my_town = {'city': '서울', 'population': 100}
city_name = my_town['city']
population = my_town['population']
print(f"2번 문제 : 제가 사는 곳은 {city_name}이고, 인구는 {population}만 명입니다.")

# 3번 - 새로운 key와 value 추가하기
product = {'name': '커피', 'price': 5000}
product['stock'] = 100          # 없는 key -> 추가
print("3번 문제 :", product)

# 4번 - 기존 값 수정하기
product['price'] = 5500         # 있는 key -> 덮어쓰기
print("4번 문제 :", product)

# 5번 - value에 리스트를 넣기 (영화별 출연 배우)
movie_cast = {'존윅': ['키아누', '윌렘'], '존윅3': ['키아누', '할리']}
print("5번 문제 :", movie_cast)

# 6번 - key 하나 삭제하기
del movie_cast['존윅']
print("6번 문제 :", movie_cast)
```

## 3. 튜플(tuple)

### 3-1. 리스트와 튜플의 차이

|  | 리스트 | 튜플 |
|---|--------|------|
| 표기 | `['우유', '빵']` | `(1999, 1, 1)` |
| 인덱싱·슬라이싱 | 가능 | 가능 |
| 값 변경 | **가능**(mutable) | **불가능**(immutable) |

둘 다 여러 값을 담는 **묶음 자료형**이다.
바뀔 수 있는 값은 리스트(장바구니), 바뀌면 안 되는 값은 튜플(생년월일)에 담는다고 생각하면 쉽다.

```python
# 리스트 : 언제든 바뀔 수 있는 장바구니
shopping_list = ['우유', '빵', '계란']
print(f"리스트 자료형 : {shopping_list}")

# 튜플 : 절대 바뀌면 안 되는 생년월일
birth_date = (1999, 1, 1)
print(f"튜플 자료형   : {birth_date}")

# 공통점 : 둘 다 인덱싱이 된다
first_item = shopping_list[0]
birth_year = birth_date[0]
print(f"첫 장보기 품목 : {first_item} / 태어난 해 : {birth_year}")
```

```python
# 차이점 : 값 변경 시도

# 리스트는 변경된다 (우유 대신 두유로 교체)
shopping_list[0] = '두유'
print("리스트 변경 성공 :", shopping_list)

# 튜플은 변경하려고 하면 TypeError가 발생한다
try:
    birth_date[0] = 2000
except TypeError as error:
    print("튜플 변경 실패 :", error)
```

## 4. 숫자형과 연산자

### 4-1. 숫자형 자료형

- **정수형(int)** : 소수점이 없는 수 → 사람 수, 개수
- **실수형(float)** : 소수점이 있는 수 → 평균, 키, 몸무게
- `type()` 함수로 자료형을 확인할 수 있다.

```python
student_count = 30              # 학생 수 -> 개수는 소수점이 없다
print("정수형 데이터 :", student_count, type(student_count))

average_score = 87.5            # 평균 점수 -> 소수점이 필요하다
print("실수형 데이터 :", average_score, type(average_score))
```

### 4-2. 사칙연산과 제곱

| 연산자 | 의미 |
|--------|------|
| `+` `-` `*` `/` | 더하기 · 빼기 · 곱하기 · 나누기 |
| `**` | 제곱 |
| `//` | 몫(정수 나눗셈) |
| `%` | 나머지 |

```python
bigger_number = 10
smaller_number = 3

print("더하기 :", bigger_number + smaller_number)
print("빼기   :", bigger_number - smaller_number)
print("곱하기 :", bigger_number * smaller_number)
print("나누기 :", bigger_number / smaller_number)   # 결과는 항상 실수(float)
```

```python
# 제곱 : 밑(base) ** 지수(exponent)
base = 3
exponent = 4
power_result = base ** exponent          # 3의 4제곱 = 81
print("3의 4제곱 :", power_result)
```

### 4-3. 나머지(%)와 몫(//) — 짝수 / 홀수 판별

- `%` : 나눗셈의 **나머지**
- `//` : 나눗셈의 **몫**
- 2로 나눈 나머지가 `0`이면 **짝수**, `1`이면 **홀수**

```python
# 하루에 몇 쪽씩 읽으면 며칠이 걸리고 몇 쪽이 남을까?
total_pages = 7
pages_per_day = 3

print("꽉 채워 읽는 날 :", total_pages // pages_per_day, "일")   # 몫 -> 2
print("마지막 날 남는 쪽 :", total_pages % pages_per_day, "쪽")   # 나머지 -> 1
```

```python
# 짝수 / 홀수 판별
locker_number = 2
print(f"{locker_number}번 사물함 나머지 :", locker_number % 2)   # 0 -> 짝수

locker_number = 1
print(f"{locker_number}번 사물함 나머지 :", locker_number % 2)   # 1 -> 홀수
```

## 5. 문자열(string)

### 5-1. 여러 줄 문자열

따옴표 3개(`'''` 또는 `"""`)로 감싸면 줄바꿈이 포함된 **여러 줄 문자열**을 만들 수 있다.

```python
# 주소처럼 줄을 나눠서 보여줘야 하는 데이터
address = '''서울특별시 강남구
테헤란로 123
파이썬빌딩 5층'''

print(address)
print("글자 수 :", len(address))   # 줄바꿈(\n)도 한 글자로 센다
```

### 5-2. 인덱싱과 슬라이싱

- 인덱스는 **0번부터** 시작한다.
- 뒤에서부터 셀 때는 **-1번부터** 시작한다.
- `문자열[시작:끝]` → 끝 번호 **직전까지** 잘라낸다.

```python
lesson_text = "문자열은 인덱싱과 슬라이싱이 가능하다."
print("원본     :", lesson_text)

print("첫 글자  :", lesson_text[0])      # 0번째 글자
print("끝 글자  :", lesson_text[-1])     # 마지막 글자
print("슬라이싱 :", lesson_text[0:5])    # 0번 ~ 4번 글자
```

### 5-3. 문자열 연산

- `+` : 문자열 **연결**
- `*` : 문자열 **반복**

```python
# 인사말을 두 단어로 나눠서 저장
greeting_word = 'Good'
time_word = 'Morning'

# + 연산자로 연결하기 (사이에 공백 문자열을 넣음)
greeting = greeting_word + " " + time_word
print(greeting)
```

```python
# * 연산자로 반복하기
print(greeting * 3)          # 인사말을 3번 반복
print("=" * 20)              # 구분선을 만들 때 자주 쓰는 방법
```

## 6. 불(bool) 자료형

### 6-1. 참(True)과 거짓(False)

- 비교 연산(`>`, `<`, `==`, `!=`)의 결과는 항상 `True` 또는 `False` 이다.

```python
# 가진 돈으로 물건을 살 수 있는지 비교하기
my_money = 8000
item_price = 10000

can_buy = my_money > item_price       # 8000 > 10000 -> False
print("살 수 있는가? :", can_buy)

# 행운의 숫자가 홀수인지 확인하기 (2로 나눈 나머지가 1인가?)
lucky_number = 7
is_odd = (lucky_number % 2 == 1)      # True
print("홀수인가?     :", is_odd)
```

### 6-2. bool의 특징

`True`는 숫자 `1`, `False`는 숫자 `0`처럼 계산된다. → **개수를 셀 때** 유용하다.

```python
true_count = can_buy + is_odd         # False(0) + True(1) = 1
print("조건을 만족한 개수 :", true_count)

# 출석 여부를 더하면 출석 일수가 된다
attended_monday = True
attended_tuesday = True
attended_wednesday = False
print("출석 일수 :", attended_monday + attended_tuesday + attended_wednesday)
```
