`print()`는 값을 화면에 출력하는 내장 함수입니다.
- 인자를 여러 개 넘기면 기본적으로 **공백**으로 구분해 이어 붙입니다.
- `sep` : 값 사이 구분자 (기본값 `' '`)
- `end` : 출력 끝에 붙는 문자 (기본값 `'\n'` = 줄바꿈)

````python
var = 100          # 정수 변수
fsval = "Fstring"  # 문자열 변수

print("Hello World")

# 인자를 여러 개 넘기면 공백으로 구분되어 출력된다
print("var =", var)

# sep: 값 사이 구분자 지정
print(2026, 9, 5, sep="-")

# end: 줄바꿈 대신 다른 문자로 끝내기 -> 두 print가 한 줄에 이어짐
print("줄바꿈", end=" 없이 ")
print("이어서 출력")
````

<h2 style="font-size:16px; font-weight:700; color:#2f6feb; margin:1.6em 0 0.5em; padding-bottom:5px; border-bottom:1px solid rgba(128,128,128,0.32);"><span style="display:inline-block; width:1.55em; height:1.55em; line-height:1.55em; text-align:center; border-radius:50%; background:#2f6feb; color:#fff; font-size:0.82em; margin-right:0.5em; vertical-align:1px;">2</span>f-문자열 (f-string)</h2>

문자열 앞에 `f`를 붙이면 중괄호 `{}` 안의 **표현식이 값으로 치환**됩니다.

⚠️ 원본 코드의 실수:
```python
print(f"f접두사:{fsval}, type(fsval)")   # type(fsval)이 중괄호 밖 -> 글자 그대로 출력됨
```
중괄호 안에 넣어야 실제로 실행됩니다.

````python
# 잘못된 예: type(fsval)이 중괄호 밖에 있어 그냥 글자로 출력된다
print(f"(X) f접두사:{fsval}, type(fsval)")

# 올바른 예: 중괄호 안에 넣어야 함수가 실행된다
print(f"(O) f접두사:{fsval}, {type(fsval)}")

# 중괄호 안에서는 계산식도 가능하다
print(f"var * 2 = {var * 2}")
````

````python
# f-string 서식(format spec) 자주 쓰는 것들
pi = 3.141592653589793

print(f"소수점 2자리 : {pi:.2f}")      # 반올림해서 2자리
print(f"전체 10칸 오른쪽 정렬 : |{pi:>10.3f}|")
print(f"천 단위 콤마 : {1234567:,}")
print(f"백분율 : {0.8567:.1%}")

# 디버깅용 = 기호 : 변수명과 값을 함께 출력 (Python 3.8+)
print(f"{var = }")
````

<h2 style="font-size:16px; font-weight:700; color:#2f6feb; margin:1.6em 0 0.5em; padding-bottom:5px; border-bottom:1px solid rgba(128,128,128,0.32);"><span style="display:inline-block; width:1.55em; height:1.55em; line-height:1.55em; text-align:center; border-radius:50%; background:#2f6feb; color:#fff; font-size:0.82em; margin-right:0.5em; vertical-align:1px;">3</span>자료형 (Data Type)</h2>

파이썬은 변수를 선언할 때 자료형을 적지 않습니다(동적 타이핑). 대입된 값에 따라 자료형이 자동 결정됩니다.

| 값 예시 | 자료형 | 이름 |
|---|---|---|
| `100` | `int` | 정수 |
| `1.0` | `float` | 실수 |
| `"A"`, `"OUTPUT"` | `str` | 문자열 |
| `True`, `False` | `bool` | 논리값 |

⚠️ C/Java와 달리 파이썬에는 **문자(char) 타입이 없습니다.** `'A'`도 길이가 1인 문자열(`str`)입니다.
그리고 `bool`은 사실 `int`의 하위 타입이라 `True == 1`, `False == 0` 입니다.

````python
ival = var        # int   : 정수
fval = 1.0        # float : 실수
cval = 'A'        # str   : 파이썬에 char 타입은 없다! 길이 1짜리 문자열
sval = "OUTPUT"   # str   : 문자열 (작은따옴표/큰따옴표 모두 가능)
bval = True       # bool  : 참(True) / 거짓(False), 첫 글자 대문자 주의

for name, value in [("ival", ival), ("fval", fval), ("cval", cval),
                    ("sval", sval), ("bval", bval)]:
    print(f"{name:5} = {value!r:10} -> {type(value).__name__}")
````

````python
# 형 변환 (casting)
print(int("123") + 1)      # 문자열 -> 정수
print(float("3.14") * 2)   # 문자열 -> 실수
print(str(100) + "점")      # 정수 -> 문자열 (문자열끼리만 + 가능)
print(int(3.9))            # 실수 -> 정수 : 반올림이 아니라 소수점 버림
print(bool(0), bool(""), bool([]))   # 0, 빈 문자열, 빈 리스트는 모두 False

# bool은 int의 하위 타입이라 계산에 그대로 쓸 수 있다
print(True + True)         # 2
````

<h2 style="font-size:16px; font-weight:700; color:#2f6feb; margin:1.6em 0 0.5em; padding-bottom:5px; border-bottom:1px solid rgba(128,128,128,0.32);"><span style="display:inline-block; width:1.55em; height:1.55em; line-height:1.55em; text-align:center; border-radius:50%; background:#2f6feb; color:#fff; font-size:0.82em; margin-right:0.5em; vertical-align:1px;">4</span>연산자</h2>

<h3 style="font-size:15px; font-weight:700; margin:1.2em 0 0.4em; padding-left:9px; border-left:3px solid #2f6feb;">산술 연산자</h3>

| 기호 | 의미 | 예 | 결과 |
|---|---|---|---|
| `+` | 더하기 | `7 + 2` | `9` |
| `-` | 빼기 | `7 - 2` | `5` |
| `*` | 곱하기 | `7 * 2` | `14` |
| `/` | 나누기 | `7 / 2` | `3.5` |
| `//` | 몫 (소수점 버림) | `7 // 2` | `3` |
| `%` | 나머지 | `7 % 2` | `1` |
| `**` | 거듭제곱 | `2 ** 10` | `1024` |

기억할 점
- `/` 는 나누어떨어져도 항상 실수(`float`)를 돌려줍니다. `4 / 2` 의 결과는 `2.0` 입니다.
- 제곱근은 지수를 `1/2` 로 주면 됩니다. `9 ** 0.5` → `3.0`
- `1e3` 은 지수 표기로 `1000.0` 이며 정수가 아니라 실수입니다.

<h3 style="font-size:15px; font-weight:700; margin:1.2em 0 0.4em; padding-left:9px; border-left:3px solid #2f6feb;">비교 연산자</h3>

| 기호 | 의미 |
|---|---|
| `==` | 같다 |
| `!=` | 같지 않다 |
| `<` | 작다 |
| `>` | 크다 |
| `<=` | 작거나 같다 |
| `>=` | 크거나 같다 |

기억할 점
- 결과는 항상 `True` 또는 `False` 입니다.
- `=` 는 값을 넣는 **대입** 연산자입니다. "같다"는 `==` 이며, 원본 주석의 `= 같다` 는 잘못된 설명입니다.
- `80 <= score < 90` 처럼 수학처럼 이어서 쓸 수 있습니다.

<h3 style="font-size:15px; font-weight:700; margin:1.2em 0 0.4em; padding-left:9px; border-left:3px solid #2f6feb;">논리 연산자</h3>

| 기호 | 참(`True`)이 되는 조건 | 예 |
|---|---|---|
| `and` | A와 B가 **모두** 참일 때 | `True and False` → `False` |
| `or` | A와 B 중 **하나라도** 참일 때 | `True or False` → `True` |
| `not` | 뒤의 값이 거짓일 때 (참↔거짓 뒤집기) | `not True` → `False` |

기억할 점
- 원본 주석에는 `or` 가 "하나가 거짓일 때"라고 적혀 있는데, 반대로 **하나라도 참일 때** 참입니다.
- 숫자 `0`, 빈 문자열 `""`, 빈 리스트 `[]` 는 거짓으로 취급됩니다.

<h3 style="font-size:15px; font-weight:700; margin:1.2em 0 0.4em; padding-left:9px; border-left:3px solid #2f6feb;">진법 변환</h3>

| 함수 | `n = 255` 일 때 결과 | 반환 자료형 |
|---|---|---|
| `bin(n)` | `'0b11111111'` | `str` |
| `oct(n)` | `'0o377'` | `str` |
| `hex(n)` | `'0xff'` | `str` |

기억할 점
- 세 함수 모두 정수가 아니라 접두사(`0b`, `0o`, `0x`)가 붙은 **문자열**을 돌려줍니다.
- 접두사 없이 보려면 f-string 서식을 씁니다. `f"{n:b}"`, `f"{n:o}"`, `f"{n:X}"`
- 반대로 바꿀 때는 `int(문자열, 진법)` 을 사용합니다. `int("ff", 16)` → `255`

````python
a, b = 7, 2

print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}       <- 나누기 결과는 항상 float")
print(f"{a} // {b} = {a // b}      <- 몫")
print(f"{a} % {b} = {a % b}       <- 나머지")
print(f"{a} ** {b} = {a ** b}      <- 거듭제곱")
print(f"9 ** (1/2) = {9 ** (1/2)}  <- 제곱근")
print(f"1e3 = {1e3}  (type: {type(1e3).__name__})")
````

````python
# 비교 연산자 : 결과는 항상 bool
print(3 == 3, 3 != 3, 3 < 5, 3 >= 5)

# 파이썬은 비교를 연달아 쓸 수 있다 (수학 표기와 동일)
score = 85
print(80 <= score < 90)   # (80 <= score) and (score < 90) 와 같은 뜻

# 논리 연산자
T, F = True, False
print("and :", T and F)   # 둘 다 참이어야 True
print("or  :", T or F)    # 하나라도 참이면 True
print("not :", not T)
````

````python
n = 255

# bin/oct/hex는 접두사(0b, 0o, 0x)가 붙은 '문자열'을 반환한다
print(bin(n), type(bin(n)).__name__)
print(oct(n))
print(hex(n))

# 접두사 없이 보고 싶다면 f-string 서식 사용
print(f"2진수: {n:b}, 8진수: {n:o}, 16진수: {n:X}")

# 반대로 문자열 -> 정수 (두 번째 인자가 진법)
print(int("11111111", 2), int("ff", 16))
````

<h2 style="font-size:16px; font-weight:700; color:#2f6feb; margin:1.6em 0 0.5em; padding-bottom:5px; border-bottom:1px solid rgba(128,128,128,0.32);"><span style="display:inline-block; width:1.55em; height:1.55em; line-height:1.55em; text-align:center; border-radius:50%; background:#2f6feb; color:#fff; font-size:0.82em; margin-right:0.5em; vertical-align:1px;">5</span>리스트 · 인덱싱 · 슬라이싱 · 중첩</h2>

변수 하나에는 값을 하나만 담을 수 있습니다. 값이 여러 개일 때 변수를 100개 만들 수는 없으니,
**대괄호 `[ ]`** 안에 값을 쉼표로 나열해 하나로 묶은 것이 리스트입니다.

```python
list_data1 = [1, 2, 3]
```

리스트의 특징
- **순서가 있습니다.** 넣은 순서가 그대로 유지되고, 순서 번호(인덱스)로 값을 꺼냅니다.
- **값을 바꿀 수 있습니다.**(mutable) `lst[0] = 99` 처럼 나중에 수정이 가능합니다.
- **자료형을 섞어도 됩니다.** 숫자와 문자열, 심지어 리스트까지 함께 담을 수 있습니다.
- `+` 는 두 리스트를 이어 붙이고, `*` 는 반복합니다. 숫자를 더하는 것과는 의미가 다릅니다.
- 길이는 `len(lst)` 로 셉니다.

아래 예시에서 `list_data2 = ['1', '2', '3']` 은 따옴표가 있으므로 **숫자가 아니라 문자열** 리스트입니다.

````python
list_data1 = [1, 2, 3]                  # 정수 리스트
list_data2 = ['1', '2', '3']            # 문자열 리스트 ('1'과 1은 다른 값!)
list_data3 = ["ABC", "String"]
list_data4 = list_data1 + list_data3    # + 는 두 리스트를 이어 붙인다
list_data5 = [1, 2, 3, ['a', 'b', 'c']] # 리스트 안에 리스트 = 중첩 리스트

print("연결 :", list_data4)
print("길이 :", len(list_data4))
print("반복 :", list_data1 * 2)         # * 는 반복
````

<h3 style="font-size:15px; font-weight:700; margin:1.2em 0 0.4em; padding-left:9px; border-left:3px solid #2f6feb;">인덱싱 — 값 하나 꺼내기</h3>

리스트의 각 자리에는 **0부터 시작하는 번호(인덱스)** 가 붙어 있습니다. 1이 아니라 0부터라는 점을 꼭 기억하세요.
음수를 쓰면 뒤에서부터 세며, `-1` 이 마지막 값입니다. 값이 5개인데 `nums[5]` 처럼 범위를 넘기면 `IndexError` 가 납니다.

<h3 style="font-size:15px; font-weight:700; margin:1.2em 0 0.4em; padding-left:9px; border-left:3px solid #2f6feb;">슬라이싱 — 여러 값을 잘라내기</h3>

`리스트[시작:끝:간격]` 형태로 원하는 구간을 잘라 **새 리스트**로 얻습니다.

- **끝 인덱스는 포함되지 않습니다.** `nums[1:3]` 은 1번과 2번만 나옵니다.
- 시작을 생략하면 처음부터, 끝을 생략하면 마지막까지입니다. `nums[:3]`, `nums[-2:]`
- 간격에 `-1` 을 주면 거꾸로 뒤집힙니다. `nums[::-1]`

슬라이싱은 문자열에도 똑같이 쓸 수 있습니다. `"파이썬"[0:2]` → `'파이'`

````python
nums = [10, 20, 30, 40, 50]
#        0   1   2   3   4     <- 인덱스
#       -5  -4  -3  -2  -1     <- 음수 인덱스

print("첫 번째   :", nums[0])
print("마지막    :", nums[-1])
print("1~2번째   :", nums[1:3])    # 끝 인덱스 3은 포함되지 않음
print("앞에서 3개:", nums[:3])
print("뒤 2개    :", nums[-2:])
print("2칸씩     :", nums[::2])
print("거꾸로    :", nums[::-1])

nums[0] = 99                        # 리스트는 값을 바꿀 수 있다
print("수정 후   :", nums)
````

<h3 style="font-size:15px; font-weight:700; margin:1.2em 0 0.4em; padding-left:9px; border-left:3px solid #2f6feb;">중첩 리스트 — 리스트 안의 리스트</h3>

리스트의 원소로 또 다른 리스트를 넣을 수 있습니다. 표(행과 열)나 좌표처럼 묶음이 여러 개인 자료를 다룰 때 씁니다.

```python
list_data5 = [1, 2, 3, ['a', 'b', 'c']]
```

이때 `list_data5[3]` 은 값 하나가 아니라 **리스트 전체**(`['a', 'b', 'c']`)입니다.
안쪽 값을 꺼내려면 대괄호를 한 번 더 붙여 `list_data5[3][0]` 처럼 씁니다. 바깥부터 안쪽 순서로 읽으면 됩니다.

주의: `"글자" + 숫자` 는 `TypeError` 가 납니다. 문자열과 숫자를 함께 출력할 때는 `str()` 로 바꾸거나 f-string을 쓰세요.

````python
# 중첩 리스트는 대괄호를 두 번 써서 접근한다
print("중첩 리스트:", list_data5[3])          # ['a', 'b', 'c']
print("중첩 리스트:" + list_data5[3][0])      # 'a'  (문자열 + 문자열)
print(f"중첩 리스트: {list_data5[3][2]}")     # f-string이 더 안전하다

# 주의: 문자열 + 숫자는 TypeError -> f-string이나 str()을 사용
# print("값:" + list_data5[0])   # TypeError!
print("값:" + str(list_data5[0]))
````

<h3 style="font-size:15px; font-weight:700; margin:1.2em 0 0.4em; padding-left:9px; border-left:3px solid #2f6feb;">자주 쓰는 리스트 메서드</h3>

`리스트.기능()` 형태로 호출하는 함수를 메서드라고 합니다. 아래 메서드는 대부분 **원본 리스트를 직접 바꾸고 `None`을 반환**하므로,
`lst = lst.sort()` 처럼 대입하면 리스트가 사라집니다. 그냥 `lst.sort()` 로 호출하세요.

| 메서드 | 하는 일 |
|---|---|
| `append(값)` | 맨 뒤에 값 추가 |
| `insert(위치, 값)` | 지정한 위치에 값 삽입 |
| `remove(값)` | 그 값을 앞에서부터 찾아 하나 삭제 |
| `pop()` | 맨 뒤 값을 꺼내면서 삭제 (꺼낸 값을 반환) |
| `sort()` | 오름차순 정렬 (`reverse=True` 면 내림차순) |
| `reverse()` | 순서 뒤집기 |

메서드는 아니지만 함께 자주 쓰는 것들: `len()`, `max()`, `min()`, `sum()`, 그리고 값이 있는지 확인하는 `in`.

````python
# 자주 쓰는 리스트 메서드
lst = [3, 1, 2]

lst.append(4)        # 맨 뒤에 추가
lst.insert(0, 0)     # 0번 위치에 삽입
print("추가 후 :", lst)

lst.remove(1)        # 값 1을 삭제 (첫 번째로 만나는 것)
last = lst.pop()     # 맨 뒤 값을 꺼내면서 삭제
print("삭제 후 :", lst, "| 꺼낸 값:", last)

lst.sort()           # 오름차순 정렬 (원본을 바꿈)
print("정렬    :", lst)
lst.reverse()
print("뒤집기  :", lst)

print("최대/최소/합계 :", max(lst), min(lst), sum(lst))
print("3이 있나? :", 3 in lst)
````
