#필요한 라이브러리 임포트

```python
from konlpy.tag import Okt
from mecab import MeCab as mc
```

```python
#형태소 객체 생성
okt = Okt()
mecap = mc()

#파일 경로
path = r"/content/drive/Othercomputers/내 컴퓨터/Drive/StudyCode/work/constitution.txt"

#파일 입출력 (인코딩 추가 및 readlines 인자 수정)
with open(path, 'r', encoding='utf-8') as fp:
  text = fp.read()
  #print(text[:5])  # 너무 길게 출력되는 것을 방지하기 위해 처음 5줄만 출력

print(len(text))
print("헌법: \n",text)
```

#형태소'Okt' 명사 추출

```python
#형태소 단위로 추출 -> 명사 만 추출
output1 = okt.nouns(text)
print("Okt 명사 추출:\n",output1)
print("="*50)

print("Okt 명사 총 개수:",len(output1))
print("="*50)
#######################################
cout = 0
index = text.find(text)
target = '법률'
while index !=-1:
  cout += 1
  index = text.find(target,index+1)
#명사 부분 단어개수
print("명사 단어 개수:",cout)
#######################################
```

```python
#명사 글자 길이 가 짧은것 만 제거

#결과 저장할 빈 리스트 생성
final_nouns = []
one_nouns = []
for noun in output1:
  if len(noun) > 1:
    final_nouns.append(noun)
  else:
    one_nouns.append(noun)

print("",one_nouns)

```

```python
#길이가 1보다 큰 명사인 '형태소'의 빈도수 분석

#필요한 라이브러리 임포트
from collections import Counter

count = Counter(final_nouns)

#결과 확인
print("다 명사 중복개수:",count)
print("="*50)
print("총개수:",len(count))
```

#형태소'MeCap' 분석기

```python
#형태소 단위로 추출 -> 명사 만 추출
########################################
output1 = mecap.nouns(text)
print("MeCap 명사 추출:\n",output1)
print("="*50)

print("MeCap 명사 총 개수:",len(output1))
print("="*50)
########################################
```

```python
#명사인 형태소 -> 길이가 1보다 큰형태소만 추출

#결과를 저장할 빈 리스트 생성
final_nouns_mecap = []
one_nouns_mecap = []
## 반복문 if문작성
for num in output1:
  if len(num) > 1:
    final_nouns_mecap.append(num)
  else:
    one_nouns_mecap.append(num)
########################################
#결과 확인
print("MeCap 명사 추출:\n",final_nouns_mecap)
print("="*50)
print("총 개수:",len(final_nouns_mecap))
```

```python
#필요한 라이브러리 임포트
from collections import Counter

#MeCap 빈도수
count_mecap = Counter(final_nouns_mecap)

print("MeCap 명사 중복개수:",count_mecap)
print("="*50)
print("총개수:",len(count_mecap))
```
