```python
from enum import Flag
from tokenizers import BertWordPieceTokenizer

# 단어장(Tokenizer) 생성
BartToken = BertWordPieceTokenizer(lowercase=False, strip_accents=False)

# 데이터 (텍스트 파일) 경로 설정
path1 = '/content/drive/Othercomputers/내 컴퓨터/Drive/StudyCode/learn/ratings_test.txt'
path2 = '/content/drive/Othercomputers/내 컴퓨터/Drive/StudyCode/learn/ratings_train.txt'

# 토크나이저 학습
BartToken.train(
    files=[path1, path2],
    vocab_size=30000
)
```

```python
# 내장된 단어사전 추출
vocab = BartToken.get_vocab()

# 일부 단어만 확인해보기 위해 dictionary 형태로 출력
print("전체 단어 수:", len(vocab))
print("단어 자료형:", type(vocab))  # dtype 대신 파이썬 내장 type() 사용
#단어
print("일부 단어 사전 예시:", list(vocab.items())[:20])
```

```python
#파일 경로 저장
#단어사전은 파일로 저장된다.
save_path = "/content/drive/Othercomputers/내 컴퓨터/Drive/StudyCode/learn"

# 저장 함수
# 토큰화 해야 학습이 가능하다
BartToken.save_model(save_path)
```

#한글 문장 토큰화

```python
#허깅 페이스에서 제공하는 트렌스포머 라이브러리 설치(코랩)

#BertTokenizerFast로 토큰화
from transformers import BertTokenizerFast
#vocab_path == "/content/drive/Othercomputers/내 컴퓨터/Drive/StudyCode/learn/vocab.txt"
BertFast = BertTokenizerFast.from_pretrained(save_path,lowercase=False, strip_accents=False)

#테스트 용 한글 샘플
sample_text = "가는말이고아야 오는말이 곱다"

#테스트 용 한글 샘플을 매개변수 로 사용
tokens = BertFast.tokenize(sample_text)
print("한글 샘플:",sample_text)
print("="*50)
print("토큰화 결과:",tokens)
```

```python
#단어를 대좋해 고유 번호를 준다
token_id = BertFast.convert_tokens_to_ids(tokens)
print("토큰화 ID:",token_id)

#단어 생성 ->  생성
```
