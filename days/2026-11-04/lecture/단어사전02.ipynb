```python
#임포트
from konlpy.tag import Okt
from mecab import MeCab as mecab
#'형태소'모델 객체 생성
okt = Okt()
mc = mecab()
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')
```

#'형태소'(Okt) 추출

```python
# text 한글 or list를

text = "한글 형태소 분석기로 테스트를 하였다, 작동은 정상이다."
#문장 -> 형태소 분석 -> 형태소 단위로 토큰화 -> morphs() 함수 호출
morph = okt.morphs(text)
########################################
#결과 확인
print("Okt.morphs() 한글 결과:\n",morph)
print("="*50)
########################################
# text(한글 문장) 비교1
# 공백 단위로 분할 -> split()
print("공백 토큰화:\n",text.split())
print("="*50)
########################################
# word_wokenize() 비교2
# 띄어쓰기 기중으로 토큰화 했을때
word = word_tokenize(text)
print("띄어쓰기 토큰화:\n",word)
########################################
```

#형태소+품사 추출

```python
# 형태소+품사 (taging) -> pos 함수 호출
########################################
pos_okt = okt.pos(text)

# 결과 출력
print("형태소+품사 pos:\n",pos_okt)
########################################
```

# '형태소'(MeCap) 추출

```python
noun = okt.nouns(text)
print("형태소:",noun)
```

```python
#MeCap은 Okt 보다 빠르고 정확하다
text = "오는말이 고아야 가는말이 곱게 전달된다."

#분장으로 부터 '형태소' 추출 후 morphs() 함수 사용
mecap_morph_okt = mc.morphs(text)

#결과 확인
print("MeCap 형태소 분석기:\n", mecap_morph_okt)
print("="*50)
########################################
print("Okt 형태소 분석기:\n", okt.morphs(text))
```

#형태소(MeCap) 품사 추출

```python
#MeCap 품사 추출
pos_mecap = mc.pos(text)

#Mecap 결과 추출
print("MeCap 품사 추출:\n", pos_mecap)
print("="*50)
#Okt 결과 추출
print("Okt 품사 추출:\n", okt.pos(text))
```

#형태소(MeCap) 명사 호출

```python
#명사 추출 nouns() 함수 호출
noun_mc = mc.nouns(text)

#결과 출력
print("MeCap 명사 추출:\n", noun_mc)
print("="*50)
print("Okt 명사 추출:\n", okt.nouns(text))
```
