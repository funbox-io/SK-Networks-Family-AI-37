## 1. Numpy (넘파이)

**수학적 계산을 위한 파이썬 라이브러리.** 자료형은 **다차원 배열(`ndarray`)** 이다. (교재 p.4)

### 1-1. 파이썬 리스트 vs 넘파이 배열 (교재 p.5)

| | python list | numpy 다차원 배열 |
|---|---|---|
| 존재 목적 | 정수·실수·문자열·리스트까지 **여러 자료형을 섞어** 담기 | **숫자 한 종류**를 대규모로 묶어 빠르게 처리 |
| 요소별 수학 연산 | **불가능** | **가능** (최대 장점) |
| 처리 방식 | 하나씩 처리 | **동시에** 처리 |

주요 특징 (교재 p.7)

1. **압도적인 연산 성능** — 대량의 숫자 데이터를 매우 빠르게 처리
2. **풍부한 수학 함수** — 사칙연산, 통계(평균·합계·표준편차), 선형대수, 푸리에 변환 등

### 1-2. 배열 만들기 — `np.array(data)` (교재 p.6, p.8)

```python
import numpy as np          # numpy 를 import 하고 앞으로 np 라는 이름으로 부른다

변수 = np.array(data)        # data = 파이썬 리스트 또는 튜플
```

차원은 **대괄호가 몇 겹인지**로 정해진다.

| 차원 | 모양 | 부르는 말 |
|------|------|-----------|
| 1차원 | 선 | 값이 한 줄로 |
| 2차원 | 행(가로) × 열(세로) | 표 |
| 3차원 | 깊이 × 행 × 열 | 표를 여러 장 쌓은 것 |

````python
import numpy as np

list_data1 = [1, 2, 3]
list_data2 = [4, 5, 6]
list_data3 = [7, 8, 9]

# 2차원 : 대괄호가 두 겹
total = [[1, 2, 3],
         [4, 5, 6],
         [7, 8, 9]]

print("List(total) :", total)
print("*" * 50)

arr = np.array(total)          # 리스트 -> 배열
print("Array :")
print(arr)

print("차원 수 :", arr.ndim, "/ 모양(행, 열) :", arr.shape, "/ 자료형 :", arr.dtype)
````

> **📝 고친 부분**
> - `3차열` → **3차원** (오타)
> - `np.array(list(lis))` → `np.array(lis)`. `lis` 는 이미 리스트라 `list()` 로 다시 감쌀 필요가 없다.
> - 주석으로 막아 둔 반복문은 `num2` 를 만들기 전에 쓰고 있어 **NameError** 가 난다. 아래 1-4 에서 인덱싱으로 다시 정리했다.
> - `ndim` · `shape` · `dtype` 을 찍어 차원과 모양을 눈으로 확인하도록 추가

### 1-3. 요소별 연산 — 리스트와 가장 다른 점

리스트의 `+` 는 **이어 붙이기**, `*` 는 **반복**이다.
배열의 `+` `*` 는 **값 하나하나에 대한 계산**이다.

````python
arr1 = np.array(list_data1)

print("리스트 + 리스트 :", list_data1 + list_data2)   # 이어 붙이기
print("배열  + 배열   :", arr1 + np.array(list_data2))  # 요소별 덧셈

print("-" * 40)

print("리스트 * 3 :", list_data1 * 3)      # 세 번 반복
print("배열  * 3 :", arr1 * 3)             # 각 값에 3을 곱함
````

### 1-4. 통계 함수 — `np.mean()`

파이썬으로 직접 구한 평균과 결과는 같지만, **돌려주는 자료형이 다르다.**

````python
avg1 = sum(list_data1) / len(list_data1)
print(f"파이썬 평균 : {avg1} {type(avg1)}")

print("*" * 20)

avg2 = np.mean(list_data1)
print(f"넘파이 평균 : {avg2} {type(avg2)}")   # numpy 전용 실수형(np.float64)

print("-" * 40)
print("합계 :", np.sum(arr), "/ 최댓값 :", np.max(arr), "/ 표준편차 :", round(float(np.std(arr)), 3))
````

### 1-5. 인덱싱 (교재 p.9~10)

리스트는 대괄호를 **두 번** 써야 하지만, 배열은 `[행, 열]` 로 **한 번에** 쓸 수 있다.

| 자료형 | 쓰는 법 |
|--------|---------|
| 리스트 | `list_data[행][열]` |
| 배열 | `arr[행][열]` 또는 **`arr[행, 열]`** |

````python
list_data = [[1, 2, 3], [4, 5, 6]]

# 리스트 : 값이 아니라 '행 전체'가 먼저 나온다
output1 = list_data[0]
print("0번 행 전체 :", output1)
print("그 중 0번 값 :", output1[0])
print("한 번에 쓰면 :", list_data[1][0])

print("*" * 10)

arr_data = np.array(list_data)
print("2차원 배열 :")
print(arr_data)

print("인덱스 두 번  :", arr_data[0][0])
print("(행, 열) 한 번 :", arr_data[0, 0])     # numpy 방식 (권장)
print("마지막 값     :", arr_data[-1, -1])
````

> **📝 고친 부분**
> - `ouput1` `ouput3` → **`output1`** `output3` (오타)
> - `print("이차원 배열", arr_data)` 처럼 한 줄에 붙여 찍으면 줄이 밀려 보기 어렵다 → 줄을 나눠 출력

### 1-6. 슬라이싱 (교재 p.11~12)

**1차원** — `배열[start : stop]` → start 부터 **stop 바로 앞까지** (`start <= x < stop`)

**2차원** — `배열[행 start : 행 stop, 열 start : 열 stop]`

> ⚠️ 콤마(`,`)가 **행과 열을 나누는 기호**다.
> `배열[0:4:9]` 처럼 콜론을 두 번 쓰면 `[start:stop:step]` 이 되어 **행만** 잘린다.

````python
list_data4 = [range(0, 10), range(10, 20), range(20, 30), range(30, 40)]
arr_data2 = np.array(list_data4)

print("원본 (4행 10열) :")
print(arr_data2)

print("*" * 20)

# 1차원 슬라이싱 : 0번 행에서 2~4번 값
print("arr_data2[0, 2:5] :", arr_data2[0, 2:5])

print("-" * 40)

# 2차원 슬라이싱 : 0~1번 행 x 0~2번 열
print("arr_data2[0:2, 0:3] :")
print(arr_data2[0:2, 0:3])

print("-" * 40)

# 비교 : 콜론을 두 번 쓰면 '행'만 9칸씩 건너뛴다 (원본 필기의 결과)
print("arr_data2[:4:9] (행만 잘림) :")
print(arr_data2[:4:9])
````

> **📝 고친 부분**
> - `arr_data2[:4:9]` 는 `[start:stop:step]` 으로 읽혀 **0번 행 하나만** 나온다.
>   교재 p.12 의 2차원 슬라이싱은 **콤마**를 쓴 `arr_data2[0:4, 0:9]` 형태다. 둘을 나란히 두고 비교하도록 바꿈
> - 결과를 덮어쓰던 변수명 `arr_data` 를 그대로 쓰지 않고 출력만 하도록 정리

## 2. Pandas (판다스)

**행(Row)과 열(Column)로 된 표 형식 데이터(Tabular Data)** 를 다루기 위한 라이브러리. (교재 p.13)

CSV · Excel · DB 테이블 등 다양한 정형 데이터를 파이썬에서 쉽게 불러오고, 정제하고, 분석할 수 있다.

### 자료형 (교재 p.14)

| 자료형 | 구성 |
|--------|------|
| **Series** | 행 + **1개의 열** → 1차원 |
| **DataFrame** | 행 + 열, 각 열이 Series → 2차원 (가장 많이 쓴다) |
| **Index** | 행이나 열을 식별하는 정수 또는 문자열 |

> MySQL 의 Table 과 달리 DataFrame 의 행 인덱스는 **0번부터** 시작한다.

```python
import pandas as pd        # pandas 를 import 하고 앞으로 pd 라는 이름으로 부른다

```

### 2-1. 실습 데이터 소개 (교재 p.17)

가상의 이커머스 회사가 **2025년 1월 1일 ~ 6월 30일** 집행한 디지털 광고의 일별 성과.

| 컬럼 | 뜻 |
|------|-----|
| `Date` | 광고 성과가 기록된 날짜 |
| `Channel` | 광고가 집행된 마케팅 채널 또는 유입 경로 |
| `Age_Group` | 타겟 고객의 연령대 그룹 |
| `Gender` | 타겟 고객의 성별 |
| `Impressions` | 광고가 잠재 고객에게 노출된 총 횟수 |
| `Clicks` | 노출된 광고를 클릭한 총 횟수 |
| `Cost` | 광고 집행에 지출된 총 비용(원) |
| `Conversions` | 광고를 통해 발생한 실제 목표 달성 횟수 (구매·가입·설치 등) |

````python
import pandas as pd

path = "./ad_performance.csv"      # 노트북과 같은 폴더
df = pd.read_csv(path)

print("자료형 :", type(df))
print("모양(행, 열) :", df.shape)
df.head()
````

### 2-2. DataFrame 분해 (교재 p.18)

**DataFrame = 행 인덱스 + 컬럼 인덱스 + 값**

| 속성 | 돌려주는 것 |
|------|------------|
| `df.index` | 행 인덱스 |
| `df.columns` | 컬럼 인덱스 |
| `df.values` | 값 → **넘파이 2차원 배열** |

````python
print("index   :", df.index)
print("columns :", list(df.columns))
print("values  :", type(df.values), df.values.shape)   # numpy 배열로 돌아온다
````

### 2-3. 인덱싱 (교재 p.19)

| 하고 싶은 것 | 쓰는 법 | 결과 |
|--------------|---------|------|
| 컬럼 1개 | `df['컬럼명']` | **Series** |
| 컬럼 여러 개 (fancy 인덱싱) | `df[['컬럼1', '컬럼2']]` | **DataFrame** |
| 특정 행 + 컬럼 | `df.loc[행 인덱스, '컬럼명']` | 값 하나 |
| 여러 행 + 여러 컬럼 | `df.loc[[행1, 행2], ['컬럼1', '컬럼2']]` | DataFrame |

> 대괄호가 **한 겹이면 Series, 두 겹이면 DataFrame** 이다.

````python
# 컬럼 1개 -> Series
info_col1 = df['Date']
print("컬럼 1개 :", type(info_col1))
print(info_col1.head(3))

print("-" * 40)

# 컬럼 여러 개 (fancy 인덱싱) -> DataFrame
info_col2 = df[['Date', 'Gender']]
print("컬럼 2개 :", type(info_col2))
print(info_col2.head(3))

print("-" * 40)

# 특정 행 + 컬럼
print("4번 행의 Clicks :", df.loc[4, 'Clicks'])
print("0~2번 행의 Channel, Cost :")
print(df.loc[[0, 1, 2], ['Channel', 'Cost']])
````

### 2-4. 파생 변수 만들기 (교재 p.20)

없던 컬럼에 값을 대입하면 **새 컬럼이 생긴다.** (`df['새컬럼'] = 계산식`)

| 지표 | 뜻 | 계산식 |
|------|-----|--------|
| **CTR** (Click-Through Rate) | 클릭률 — 노출 대비 얼마나 눌렸나 | `Clicks / Impressions * 100` |
| **ROAS** (Return On Ad Spend) | 광고 수익률 — 쓴 돈 대비 얼마나 벌었나 | `Revenue / Cost * 100` |

지울 때는 `df = df.drop('컬럼명', axis=1)`. **`axis=1` 이 열**, `axis=0` 이 행이다.

````python
# 파생 변수 생성
df['CTR'] = df['Clicks'] / df['Impressions'] * 100
df['ROAS'] = df['Revenue'] / df['Cost'] * 100

print(df[['Impressions', 'Clicks', 'CTR', 'Cost', 'Revenue', 'ROAS']].head())

print("-" * 40)
print("평균 CTR  :", round(df['CTR'].mean(), 2), "%")
print("평균 ROAS :", round(df['ROAS'].mean(), 2), "%")
````

````python
# 컬럼 삭제 (axis=1 -> 열)
df = df.drop('ROAS', axis=1)
print("삭제 후 컬럼 :", list(df.columns))
````

> **📝 고친 부분**
> - `df['New_Data'] = 0 #df['Clicks'] / df['Impressions']*10` — 계산식이 주석으로 막혀 0만 들어가 있었다.
>   교재 p.20 의 CTR 은 `* 100` (원본은 `* 10`).
> - 주석 `# 루아스 계산식` 이 **CTR 식 옆에** 붙어 있었다. 루아스(ROAS)는 `Revenue / Cost`, 클릭률(CTR)은 `Clicks / Impressions` 로 서로 다른 지표라 나눠서 정리
> - 컬럼명 `Ross` → **`ROAS`**
> - `pd.DataFrame` 만 적은 줄은 클래스를 가리키기만 할 뿐 아무 일도 하지 않아 지움
> - `display(df)` 는 주피터 전용이고 5,000행을 다 찍는다 → `df.head()` 로 바꿈

### 2-5. `value_counts()` (교재 p.21)

컬럼이 **범주형 데이터**일 때 **항목별 개수(빈도수)** 를 세어 준다.

1. `df['컬럼']` 로 컬럼을 인덱싱하고
2. `.value_counts()` 를 호출한다

````python
# 채널별 집행 건수
print("[Channel]")
print(df['Channel'].value_counts())

print("-" * 40)

# 연령대별 · 성별 건수
print("[Age_Group]")
print(df['Age_Group'].value_counts())

print("-" * 40)

print("[Gender]")
print(df['Gender'].value_counts())
````

> **📝 고친 부분**
> - `df['Date'].value_counts()` 는 날짜가 180일이라 결과가 길기만 하다.
>   범주형(`Channel` · `Age_Group` · `Gender`)으로 바꿔 **빈도수를 세는 의미**가 드러나게 함

## 3. 오늘 정리

- **numpy** 는 숫자 계산용, **pandas** 는 표 계산용 라이브러리다.
- 배열의 최대 장점은 **요소별 연산** — 리스트의 `+`는 이어 붙이기, 배열의 `+`는 값끼리 더하기.
- 배열 인덱싱은 `arr[행, 열]`, 슬라이싱은 `arr[행start:행stop, 열start:열stop]`. **콤마가 행과 열을 나눈다.**
- `stop` 인덱스는 **포함되지 않는다** (`start <= x < stop`).
- DataFrame = **행 인덱스 + 컬럼 인덱스 + 값**, `df.values` 는 넘파이 배열로 돌아온다.
- 대괄호 **한 겹이면 Series, 두 겹이면 DataFrame**.
- 없던 컬럼에 대입하면 **파생 변수**가 생긴다. 지울 때 `axis=1` 이 열.
- 범주형 컬럼의 빈도수는 `df['컬럼'].value_counts()`.
