# 데이터 불러오기

```python
# 파이썬 경고 무시 설정
import warnings
warnings.filterwarnings('ignore')
```

```python
# 필요한 라이브러리 임포트
import pandas as pd

# 파일 경로 설정
file_path=r'./Telco-Customer-Churn.csv'

# DataFrame 생성
df = pd.read_csv(file_path)

# 결과 확인
display(df)
```

# 데이터 전처리

```python
# DataFrame -> 전용 함수 -> 기본 정보(행과 열) 제공(print) -> info() 함수
df.info()
```

```python
# 문제점: TotalCharges 컬럼 데이터 -> 공백과 같은 숫자로 변환할 수 없는 문자열(" ") 포함

## TotalCharges 컬럼의 값 -> 공백인 문자열(" ")인 행 인덱스 추출
space_index = df.query("TotalCharges == ' '").index   
print(space_index)
```

```python
# 전처리 방법: TotalCharges 컬럼 -> 공백과 같이 숫자로 변환활 수 없는 문자열 포함인 행 제거

## 1단계: TotalCharges 컬럼의 자료형: 문자열 -> 숫자형으로 변환 ->pd.to_numeric(컬럼)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print("빈 문자열 처리 후 info 함수 적용")
df.info()
```

```python
# 전처리 방법: TotalCharges 컬럼 -> 공백과 같이 숫자로 변환활 수 없는 문자열 포함인 행 제거

## 2단계: TotalCharges 컬럼 -> 누락인 행 제거 -> dropna()
df1 = df.dropna(ignore_index=True)
display(df1)
```

One-Hot-Encoding

```python
# 고유 번호 삭제
df2 = df1.drop(columns='customerID')
# 출력!
#display(df2)

#replace(str) == map(int) 
df2['Churn'] = df2['Churn'].map({'No':0,'Yes':1})
display(df2)
```

## 특정 범주형 컬럼 만 (원핫 인코딩 적용)
## 범주형 컴럼 추출 -> df.select_dtype(include=[])->리스트 자료형으로 변환

```python
df_categorical = df2.select_dtypes(include=[str])
#display(df_categorical)

# select_dtypes() ,매개변수 include=[타입 만 묶음]
#df_categorical.info()

#"ALL columns output!""
cols = df_categorical.columns

#출력!
#print("ALL columns:\n",cols)
#df2.info()

#리스트 로 형변환 tolist()
colos_to_list = cols.tolist()
#print("형변환 to list:\n",colos_to_list)

#범주형,원-핫(인코딩)
#teste = df2.select_dtypes(include=[str]).columns.tolist()
#print(teste)

#float64 2개 있다.
#int64 타입으로 구별해야될것 같다. 하지만 모르겠다.
#문자열
df_onehost = pd.get_dummies(df2,columns=colos_to_list,dtype=int)
#df_onehost.info()
display(df_onehost)
#print("one hot:",)
```

# 학습용 데이터와 평가용 생성

```python
X_data = df_onehost.drop(columns=['Churn'])
print("="*50)
y_data = df_onehost['Churn']

#새로운 데이터프레임에 "정규화" 진행!
ratio = y_data.value_counts(normalize=True)
print("정답 데이터 생성!")
```

```python
## 학습용:평가용(8:2 비율)
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
    X_data,
    y_data,
    test_size=0.2,
    random_state=0,
    stratify=y_data
)

# 학습용 데이터 확인 -> 행 인덱스
print(X_train.index)
print(y_train.index)
print("="*50)
#############################
#빈도수 연산!
ratio_train = y_train.value_counts(normalize=True)
print(ratio_train)
print("="*50)
```

## 분류 (로지스틱 회귀)

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
## 모델 생성
lr =  LogisticRegression()

## 모델 학습
lr.fit(X_train,y_train)
print("학습 완료!")

## 모델 예측
pred_test_lr = lr.predict(X_test)

## 모델 평가
acc_lr = accuracy_score(y_test, pred_test_lr)
print(acc_lr)
```

```python
#다중 모델
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

#모델 생성
svc = SVC()

#표준화가 아닌가? 아니다 표준화를 한다!
scaler = StandardScaler()
#X_train_scaled = scaler.fit_transform(X_train)

#모델 학습
svc.fit(X_train,y_train)
print("학습 완료!")

#모델 예측
pred_test = svc.predict(X_test) 
#print(pred_test)

#모델 평가
acc_svc = accuracy_score(y_test,pred_test)
print(acc_svc)
print("="*50)
```

트리 분류 모델

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

#모델 생성
dt = DecisionTreeClassifier(random_state=0)

#모델 학습
dt.fit(X_train,y_train)
print("학습 완료!")

#모델 예측
predict_dt = dt.predict(X_test)

#모델 평가
acc_dt = accuracy_score(y_test,predict_dt)
print(f"모델 성능:{acc_dt}")
```
