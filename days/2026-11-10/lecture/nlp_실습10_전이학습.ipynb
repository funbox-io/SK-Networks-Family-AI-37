# 학습용 데이터 생성

```python
# 네이버 영화 리뷰 감성 분석 데이터 -> 데이터프레임 생성, 누락 데이터 제거

## 필요한 라이브러리 /함수 임폴트
import pandas as pd
from tqdm import tqdm

## 학습용 데이터 --> 데이터프레임 생성
file_path='/content/drive/MyDrive/sk/ratings_train.txt'
df_train = pd.read_csv(file_path, sep='\t')
print('학습용 데이터 확인')
display(df_train)

print('-'*80)

## 학습용 데이터 --> 각 컬럼 별 누락 데이터의 수 확인
num_nulls = df_train.isnull().sum()
print(f'각 컬럼 별 누락 데이터의 수 : \n{num_nulls}')

print('-'*80)

## 누락 데이터 제거 --> dropna()
df_train1 = df_train.dropna(ignore_index=True)

## 결과 확인하기
print('누락 제거 후 학습용 데이터')
display(df_train1)
```

# 모델별 토크나이저 생성

```python
# 필요한 라이브러리 / 함수 임폴트
from transformers import AutoTokenizer
```

```python
# 사전 학습된 다국어 BERT 토크나이저 다운로드

## 사용할 모델 지정
model_name = 'google-bert/bert-base-multilingual-cased'

## 토크나이저 다운로드
multi_tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
# 사전 학습된 한국어 BERT 토크나이저 다운로드

## 사용할 모델 지정
model_name = 'klue/bert-base'

## 토크나이저 다운로드
klue_tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
# 토큰화 성능 확인 함수 정의
def inspect_tokenization(tokenizer, name):
    ## 텍스트 문장 생성
    text = '너와 함께한 시간 모두 눈부셨다. 날이 좋아서 날이 좋지 않아서 날이 적당해서 모든 날이 좋았다.'
    ## tokenize() 함수 사용 --> 토큰화
    result = tokenizer.tokenize(text)
    print(f'[{name}] 토큰화 결과: \n{result}')
    print('-'*80)
```

```python
# 토큰화 성능 확인 함수 실행
inspect_tokenization(multi_tokenizer, "Multilingual BERT")
inspect_tokenization(klue_tokenizer, "KLUE BERT")
```

```python
# multi_tokenizer -> tokenize() 함수 -> 리뷰별 토큰 수 추출

## df_train1 -> document 컬럼 추출
data = df_train1['document']

## 전체 리뷰별 토큰 수 측정
total_length_multi = []
for review in tqdm(data):
    ### 리뷰 1개에 대한 토큰화 --> len() 함수 사용 --> 토큰의 수 측정
    review_length = len(multi_tokenizer.tokenize(review))
    total_length_multi.append(review_length)

## 결과 확인하기
print(f'BERT 다국어 토크나이저를 이용하여 측정한 영화 리뷰별 토크의 수 : \n{total_length_multi}')
```

```python
# klue_tokenizer -> tokenize() 함수 --> 리뷰별 토큰 수 추출

## df_train1 -> document 컬럼 추출
data = df_train1['document']

## 전체 리뷰별 토큰 수 측정
total_length_klue = []
for review in tqdm(data):
    ### 리뷰 1개에 대한 토큰화 --> len() 함수 사용 --> 토큰의 수 측정
    review_length = len(klue_tokenizer.tokenize(review))
    total_length_klue.append(review_length)

## 결과 확인하기
print(f'BERT 한국어 토크나이저를 이용하여 측정한 영화 리뷰별 토크의 수 : \n{total_length_klue}')
```

```python
# 영화 리뷰 별 토큰의 수 -> 통계 추출 -> np.percentilte()

## 필요한 라이브러리 임폴트
import numpy as np

## BERT 다국어 토크나이저로 토큰화 한 결과 --> 90%에 해당하는 토큰의 수 추출
number_multi = np.percentile(a=total_length_multi, q=90)
print(number_multi)

print('-'*80)

## BERT 한국어 토크나이저로 토큰화 한 결과 --> 90%에 해당하는 토큰의 수 추출
number_klue = np.percentile(a=total_length_klue, q=90)
print(number_klue)
```

# 모델 입력 데이터 생성

```python
# BERT 다국어 모델 3가지 입력 데이터: input_ids, attention_mask, token_type_ids 생성
multi_train_input = multi_tokenizer(X_train, truncation=True, padding=True, max_length=50, return_tensors="pt")
multi_val_input = multi_tokenizer(X_val, truncation=True, padding=True, max_length=50, return_tensors="pt")
multi_test_input = multi_tokenizer(X_test, truncation=True, padding=True, max_length=50, return_tensors="pt")
```

```python
# BERT 한국어 모델 3가지 입력 데이터: input_ids, attention_mask, token_type_ids 생성
klue_train_input = klue_tokenizer(X_train, truncation=True, padding=True, max_length=42, return_tensors="pt")
klue_val_input = klue_tokenizer(X_val, truncation=True, padding=True, max_length=42, return_tensors="pt")
klue_test_input = klue_tokenizer(X_test, truncation=True, padding=True, max_length=42, return_tensors="pt")
```

```python
# Custom Dataset 클래스 함수 정의

## 필요한 라이브러리 임폴트
import torch

## 클래스 함수 정의
class ClassificationDataset(torch.utils.data.Dataset):
    ### 입력 데이터(inputs)와 정답 라벨(labels)을 받아 변수로 저
    def __init__(self, inputs, labels):
        self.inputs = inputs
        self.labels = labels
    ### 분류 모델 입력 형식에 맞춘 최종 딕셔너리(input_ids, attention_mask, token_type_ids, labels)를 반환
    def __getitem__(self, idx):
        item = {key: val[idx].clone().detach() for key, val in self.inputs.items()}
        item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item
    ### 전체 데이터의 개수를 반환
    def __len__(self):
        return len(self.inputs["input_ids"])
```

```python
# BERT 다국어 모델 입력 데이터 생성

## Custom Dataset 클래스 함수 호출, 모델 입력 데이터 생성
multi_train_dataset = ClassificationDataset(multi_train_input, y_train)
multi_val_dataset = ClassificationDataset(multi_val_input, y_val)
multi_test_dataset = ClassificationDataset(multi_test_input, y_test)

## 결과 확인(학습용 데이터)
print(f'BERT 다국어 모델 학습용 데이터 : \n{multi_train_dataset}')
print('-'*80)
print(f'BERT 다국어 모델 검증용 데이터 : \n{multi_val_dataset}')
```

```python
# BERT 한국어 모델 입력 데이터 생성

## Custom Dataset 클래스 함수 호출, 입력 데이터 생성
klue_train_dataset = ClassificationDataset(klue_train_input, y_train)
klue_val_dataset = ClassificationDataset(klue_val_input, y_val)
klue_test_dataset = ClassificationDataset(klue_test_input, y_test)

## 결과 확인(학습용 데이터)
print(f'BERT 다국어 모델 학습용 데이터 : \n{klue_train_dataset}')
print('-'*80)
print(f'BERT 다국어 모델 검증용 데이터 : \n{klue_val_dataset}')
```

# 모델 별 감성 분석

```python
# 평가 함수 정의

## 필요한 함수 임폴트
from sklearn.metrics import accuracy_score

## 평가 함수 정의
def compute_metrics(pred):
    labels = pred.label_ids
    preds = pred.predictions.argmax(-1)

    # 정확도 측정
    accuracy = accuracy_score(labels, preds)
    return {'accuracy': accuracy}
```

## BERT 다국어 모델을 이용한 감성 분석

### 모델 학습

```python
# 사전 학습된 BERT 다국어 모델 생성하기

## 필요한 함수 임폴트
from transformers import BertConfig, BertForSequenceClassification

## 사전 학습된 모델 config 설정
model_name='google-bert/bert-base-multilingual-cased'
num_labels=2
config = BertConfig.from_pretrained(model_name, num_labels=num_labels)

## 모델 다운로드
multi_model = BertForSequenceClassification.from_pretrained(model_name, config=config)
```

```python
## Trainer 환경 설정

## 필요한 클래스 임폴트
from transformers import TrainingArguments

## TrainingArguments 클래스 호출, 환경 설정 도구 생성
multi_training_args = TrainingArguments(
    ### 출력 설정
    output_dir="/content/drive/MyDrive/sk/multi",
    ### 학습 규모 및 배치 크기 설정
    num_train_epochs=1,
    per_device_train_batch_size=128,
    per_device_eval_batch_size=64,
    ### 최적화 설정
    learning_rate=0.0001,
    weight_decay=0.01,
    ### 검증 및 저장 설정
    eval_strategy="epoch",
    save_strategy="epoch",
    metric_for_best_model="eval_loss",
    greater_is_better=False,
    load_best_model_at_end=True,
    save_total_limit=1,
    ### 시각화 도구 사용 설정
    report_to="none"
)
```

```python
# Trainer 생성

## 필요한 클래스 임폴트
from transformers import Trainer

## Trainer 클래스 호출, Trainer 생성
multi_trainer = Trainer(
    model=multi_model,
    args=multi_training_args,
    train_dataset=multi_train_dataset,
    eval_dataset=multi_val_dataset,
    processing_class=multi_tokenizer,
    compute_metrics=compute_metrics
    # callbacks = [EarlyStoppingCallback(early_stopping_patience=2)]
)
```

```python
# 학습 실행
multi_trainer.train()
```

### 모델 평가

```python
# trainer.evaluate() 사용
multi_performance = multi_trainer.evaluate(multi_test_dataset)
print(f'Multilingual BERT 평가용 데이터 성능 : {multi_performance}')
```
