# 파이썬 예외 처리 · 파일 처리 정리

예외 처리(`try` / `except` / `else` / `finally`)와 파일 처리(`open` · `with` · 읽기/쓰기 메서드)를 정리한 학습 노트입니다.
각 단원은 **개념 설명 → 예제 코드 → 연습 문제** 순서로 구성되어 있습니다.

## 1. 예외 처리 (Exception Handling)

### 1-1. 왜 필요한가

파이썬은 **위에서 아래로 한 줄씩 순차 실행**된다.
중간에 에러가 나면 그 즉시 프로그램이 멈추고 **아래 코드는 실행되지 않는다.**

예외 처리를 쓰면 에러가 나도 프로그램이 죽지 않고,
대신 준비해 둔 코드로 넘어가서 **계속 실행**된다.

````python
# 예외 처리가 없으면 에러가 난 줄에서 프로그램이 멈춘다
print("1단계 : 시작")

# print(10 / 0)      # <- 주석을 풀면 ZeroDivisionError 가 나면서 아래 줄이 실행되지 않는다

print("2단계 : 여기까지 실행")
````

### 1-2. try ~ except 기본 구조

```python
try:
    # 에러가 날 수 있는 코드
    num = int(input("숫자 입력: "))
    print(10 / num)
except:
    # 에러가 났을 때 대신 실행되는 코드
    print("잘못된 입력입니다.")

print("프로그램은 계속 실행됩니다.")   # 에러가 나도 이 줄은 실행됨
```

흐름은 이렇다.

`try` 안에서 에러 발생 → **남은 `try` 코드는 건너뜀** → `except` 실행 → 그 아래 코드 정상 진행

> 노트북에서 `input()` 을 쓰면 입력창이 뜰 때까지 멈추므로,
> 아래 예제에서는 입력값을 변수에 미리 담아 **같은 상황을 재현**했다.

````python
# input() 대신 여러 입력값을 리스트로 두고 하나씩 넣어 본다
user_inputs = ["5", "0", "가나다"]

for user_input in user_inputs:
    try:
        num = int(user_input)        # "가나다" 면 여기서 에러
        print(f"입력 {user_input!r} -> 결과 {10 / num}")   # num 이 0 이면 여기서 에러
    except:
        print(f"입력 {user_input!r} -> 잘못된 입력입니다.")

print("프로그램은 계속 실행됩니다.")
````

### 1-3. 명시적 예외 처리 — 에러 종류를 지정하기

`except` 뒤에 에러 이름을 적어 두면 **그 에러만** 잡는다.
다른 종류의 에러가 나면 잡지 못하고 프로그램이 멈춘다.
대신 **무슨 에러인지 정확히 알고 대응**할 수 있다는 게 장점이다.

```python
try:
    num = int(input("숫자 입력: "))
    print(10 / num)
except ValueError:          # 숫자가 아닌 걸 입력했을 때
    print("숫자만 입력하세요.")
except ZeroDivisionError:   # 0으로 나눴을 때
    print("0으로 나눌 수 없습니다.")
```

**자주 쓰는 에러 종류**

| 에러 이름 | 언제 발생하나 |
|-----------|---------------|
| `ValueError` | 값의 형태가 잘못됨 (`int("가나다")`) |
| `TypeError` | 자료형이 안 맞음 (`"3" + 5`) |
| `ZeroDivisionError` | 0으로 나눔 |
| `IndexError` | 리스트 범위를 벗어난 인덱스 |
| `KeyError` | 딕셔너리에 없는 키 |
| `FileNotFoundError` | 파일이 존재하지 않음 |

````python
# 에러 종류별로 따로 잡아 보기
user_inputs = ["5", "0", "가나다"]

for user_input in user_inputs:
    try:
        num = int(user_input)
        print(f"입력 {user_input!r} -> 결과 {10 / num}")
    except ValueError:            # 숫자가 아닌 값을 int() 로 바꾸려 할 때
        print(f"입력 {user_input!r} -> 숫자만 입력하세요.")
    except ZeroDivisionError:     # 0으로 나눌 때
        print(f"입력 {user_input!r} -> 0으로 나눌 수 없습니다.")
````

````python
# 표에 정리한 에러를 하나씩 직접 일으켜 보기

try:
    int("가나다")
except ValueError as e:
    print("ValueError        :", e)

try:
    "3" + 5
except TypeError as e:
    print("TypeError         :", e)

try:
    10 / 0
except ZeroDivisionError as e:
    print("ZeroDivisionError :", e)

try:
    [1, 2, 3][5]
except IndexError as e:
    print("IndexError        :", e)

try:
    {'name': '김민수'}['age']
except KeyError as e:
    print("KeyError          :", e)      # 없는 key 이름이 그대로 나온다

try:
    open("없는파일.txt", "r", encoding="utf-8")
except FileNotFoundError as e:
    print("FileNotFoundError :", e)
````

### 1-4. 포괄적 예외 처리 — 모든 에러를 한 번에

에러 이름을 안 적거나 `Exception` 을 쓰면 **모든 에러**를 잡는다.
`as e` 를 붙이면 에러 내용을 변수 `e` 에 담아 확인할 수 있다.

```python
try:
    ...
except Exception as e:   # as e -> 에러 내용을 변수 e에 담음
    print("에러 발생:", e)
```

| 방식 | 장점 | 단점 |
|------|------|------|
| 명시적 (`except ValueError:`) | 무슨 에러인지 정확히 알고 대응 | 예상 못 한 에러는 놓침 |
| 포괄적 (`except Exception as e:`) | 안전망 역할 | 무슨 에러인지 뭉개져서 디버깅이 어려움 |

> 실무에서는 **명시적으로 먼저 쓰고, 맨 마지막에 포괄적으로 마무리**하는 방식을 쓴다.

````python
# 명시적 -> 포괄적 순서로 쓰기 (위에서부터 차례대로 검사한다)
test_values = ["가나다", 0, [1, 2, 3]]

for value in test_values:
    try:
        print("결과 :", 10 / int(value))
    except ValueError:
        print("값 오류")
    except Exception as e:        # 위에서 못 잡은 나머지 전부
        print("기타 오류 :", e)
````

### 1-5. else 와 finally

| 절 | 언제 실행되나 |
|----|---------------|
| `try` | 먼저 실행되는 코드 |
| `except` | `try` 에서 에러가 났을 때 |
| `else` | **에러가 없을 때만** |
| `finally` | **에러가 있든 없든 무조건** (파일 닫기 등 마무리 작업) |

```python
try:
    ...
except ValueError:
    print("값 오류")
except Exception as e:
    print("기타 오류:", e)
else:
    print("에러가 없을 때만 실행")
finally:
    print("에러가 있든 없든 무조건 실행")
```

````python
# 에러가 날 때와 안 날 때, else / finally 가 어떻게 달라지는지 비교
def divide(value):
    print(f"--- 입력값 : {value!r} ---")
    try:
        result = 10 / int(value)
    except ValueError:
        print("값 오류")
    except Exception as e:
        print("기타 오류 :", e)
    else:
        print("에러가 없을 때만 실행 -> 결과", result)
    finally:
        print("에러가 있든 없든 무조건 실행")

divide(5)        # 정상 -> else 실행
divide("가나다")   # ValueError -> else 건너뜀
divide(0)        # ZeroDivisionError -> 포괄적 except 가 잡음
````

### 1-6. 예외 처리 연습 문제 (1 ~ 6번)

````python
# 1번 문제 - 문자열을 정수로 바꾸다가 실패하면 0을 돌려주는 함수
def to_int(text):
    try:
        return int(text)
    except ValueError:
        return 0

print("1번 문제 :", to_int("100"), to_int("백"))


# 2번 문제 - 0으로 나누는 경우를 안전하게 처리하기
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "0으로 나눌 수 없습니다."

print("2번 문제 :", safe_divide(10, 2), "/", safe_divide(10, 0))


# 3번 문제 - 리스트 범위를 벗어난 인덱스 처리 (IndexError)
scores = [90, 85, 77]
try:
    print("3번 문제 :", scores[5])
except IndexError:
    print("3번 문제 : 없는 번호입니다.")


# 4번 문제 - 딕셔너리에 없는 key 처리 (KeyError)
student = {'name': '이영희', 'dept': '데이터공학과'}
try:
    print("4번 문제 :", student['age'])
except KeyError:
    print("4번 문제 : 없는 정보입니다.")
    # get() 을 쓰면 애초에 에러가 나지 않고 None 이 나온다
    print("4번 문제 get() :", student.get('age'))


# 5번 문제 - 에러 내용을 그대로 출력해 보기 (as e)
try:
    result = "3" + 5
except Exception as e:
    print("5번 문제 :", type(e).__name__, "-", e)


# 6번 문제 - 여러 값을 반복하며 에러가 나도 멈추지 않게 하기
raw_scores = ["90", "85", "미응시", "77"]
total = 0

for raw in raw_scores:
    try:
        total += int(raw)
    except ValueError:
        continue          # 숫자가 아니면 건너뛰고 반복을 계속한다

print("6번 문제 : 합계", total)
````

## 2. 파일 처리

### 2-1. open() 과 인코딩

**인코딩(encoding)** 이란 사람이 읽는 텍스트(문자)를 컴퓨터가 저장하는 숫자로 바꾸는 규칙이다.
한글은 `utf-8` 을 지정하지 않으면 깨질 수 있으므로 **항상 `encoding="utf-8"` 을 붙이는 습관**을 들인다.

```python
open("파일이름", "모드", encoding="utf-8")
```

**모드**

| 모드 | 의미 |
|------|------|
| `"w"` | 쓰기 — 파일이 없으면 새로 만들고, **있으면 기존 내용 전부 삭제** |
| `"a"` | 추가 — 기존 내용 뒤에 이어서 씀 |
| `"r"` | 읽기 — 파일이 없으면 `FileNotFoundError` |

### 2-2. 쓰기 메서드

| 메서드 | 하는 일 |
|--------|---------|
| `write("문자열")` | 문자열 하나를 씀 (줄바꿈은 `\n` 을 직접 넣어야 함) |
| `writelines(리스트)` | 리스트에 든 문자열들을 이어서 씀 |

기본 방식은 반드시 **`close()` 로 닫아야** 저장이 완료된다.

```python
f = open("test.txt", "w", encoding="utf-8")
f.write("안녕하세요!")
f.close()   # 안 닫으면 내용이 저장 안 될 수 있음
```

````python
# w 모드 - 파일을 새로 만들어 쓰기 (있던 내용은 사라진다)
f = open("test.txt", "w", encoding="utf-8")
f.write("안녕하세요!\n")          # 줄바꿈은 \n 을 직접 넣어야 한다
f.write("파이썬 파일 처리입니다.\n")
f.close()                         # 반드시 닫기

# a 모드 - 기존 내용 뒤에 이어서 쓰기
f = open("test.txt", "a", encoding="utf-8")
f.writelines(["세 번째 줄\n", "네 번째 줄\n"])   # 리스트를 이어서 쓴다
f.close()

print("test.txt 작성 완료")
````

### 2-3. 읽기 메서드

| 메서드 | 결과 |
|--------|------|
| `read()` | 파일 전체를 **하나의 문자열**로 |
| `readline()` | **한 줄만** 문자열로 |
| `readlines()` | 전체를 **줄 단위 리스트**로 (`['1줄\n', '2줄\n']`) |

````python
# read() - 전체를 하나의 문자열로
f = open("test.txt", "r", encoding="utf-8")
print("read() :")
print(f.read())
f.close()

print("-" * 30)

# readline() - 한 줄만
f = open("test.txt", "r", encoding="utf-8")
print("readline() :", f.readline())      # 줄 끝의 \n 이 함께 딸려 온다
f.close()

print("-" * 30)

# readlines() - 줄 단위 리스트로
f = open("test.txt", "r", encoding="utf-8")
print("readlines() :", f.readlines())
f.close()
````

### 2-4. with 방식 (권장)

블록이 끝나면 **자동으로 닫아 준다.** 중간에 에러가 나도 닫히기 때문에 훨씬 안전하다.

```python
with open("test.txt", "w", encoding="utf-8") as f:
    f.write("반갑습니다.")
# 여기서 자동으로 close()
```

````python
# with 문 - close() 를 쓰지 않아도 블록이 끝나면 자동으로 닫힌다
with open("test.txt", "w", encoding="utf-8") as f:
    f.write("반갑습니다.\n")
    f.write("with 를 쓰면 close() 를 잊어버릴 걱정이 없다.\n")

with open("test.txt", "r", encoding="utf-8") as f:
    print(f.read())

print("파일이 닫혔는가? :", f.closed)     # with 블록을 벗어나면 True
````

### 2-5. 읽기 + 예외 처리를 합친 실전 형태

파일을 **줄 단위로 하나씩** 읽으면 전체를 메모리에 올리지 않아 효율적이다.
`strip()` 으로 줄 끝의 `\n` 을 제거한다.

```python
try:
    with open("test.txt", "r", encoding="utf-8") as f:
        for line in f:          # 줄 단위로 하나씩 읽기 (메모리 효율적)
            print(line.strip()) # strip() -> 줄 끝의 \n 제거
except FileNotFoundError:
    print("파일이 없습니다.")
```

````python
# 있는 파일을 읽을 때
try:
    with open("test.txt", "r", encoding="utf-8") as f:
        for line in f:              # 줄 단위로 하나씩
            print(line.strip())     # 줄 끝의 \n 제거
except FileNotFoundError:
    print("파일이 없습니다.")

print("-" * 30)

# 없는 파일을 읽을 때 - 프로그램이 멈추지 않고 안내 문구만 나온다
try:
    with open("없는파일.txt", "r", encoding="utf-8") as f:
        print(f.read())
except FileNotFoundError:
    print("파일이 없습니다.")

print("프로그램은 계속 실행됩니다.")
````

### 2-6. 파일 처리 연습 문제 (1 ~ 5번)

````python
# 1번 문제 - 리스트를 한 줄에 하나씩 파일로 저장하기
todo_list = ["수업 듣기", "복습하기", "블로그 정리하기"]

with open("todo.txt", "w", encoding="utf-8") as f:
    for todo in todo_list:
        f.write(todo + "\n")        # 줄바꿈을 직접 붙여 준다

print("1번 문제 : todo.txt 저장 완료")


# 2번 문제 - 저장한 파일을 다시 읽어 리스트로 되돌리기
with open("todo.txt", "r", encoding="utf-8") as f:
    loaded = [line.strip() for line in f.readlines()]

print("2번 문제 :", loaded)


# 3번 문제 - a 모드로 한 줄 덧붙이기
with open("todo.txt", "a", encoding="utf-8") as f:
    f.write("일찍 자기\n")

with open("todo.txt", "r", encoding="utf-8") as f:
    print("3번 문제 :", f.read().strip().split("\n"))


# 4번 문제 - 파일이 없을 때 안내 문구를 돌려주는 함수
def read_file(file_name):
    try:
        with open(file_name, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        return f"{file_name} 파일이 없습니다."

print("4번 문제 :", read_file("없는파일.txt"))


# 5번 문제 - 파일의 줄 수와 글자 수 세기
with open("todo.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

line_count = len(lines)
char_count = sum(len(line.strip()) for line in lines)
print(f"5번 문제 : {line_count}줄, 공백 제외 {char_count}글자")
````

## 3. 종합 응용 문제 (1 ~ 4번)

예외 처리와 파일 처리를 함께 쓰는 형태를 연습한다.

````python
# 1번 문제 - 점수 파일을 읽어 평균 내기 (숫자가 아닌 줄은 건너뛰기)
with open("scores.txt", "w", encoding="utf-8") as f:
    f.writelines(["90\n", "85\n", "미응시\n", "77\n"])

total = 0
count = 0

with open("scores.txt", "r", encoding="utf-8") as f:
    for line in f:
        try:
            total += int(line.strip())
            count += 1
        except ValueError:
            print("1번 문제 : 숫자가 아닌 값 건너뜀 ->", line.strip())

print("1번 문제 : 평균", round(total / count, 1))


# 2번 문제 - 로그를 파일에 계속 덧붙여 기록하는 함수
def write_log(message):
    with open("log.txt", "a", encoding="utf-8") as f:
        f.write(message + "\n")

write_log("프로그램 시작")
write_log("에러 발생 : ZeroDivisionError")
write_log("프로그램 종료")

with open("log.txt", "r", encoding="utf-8") as f:
    print("2번 문제 :")
    print(f.read().strip())
````

````python
# 3번 문제 - 에러가 나면 로그 파일에 남기고, 프로그램은 계속 진행하기
def safe_divide_with_log(a, b):
    try:
        result = a / b
    except Exception as e:
        with open("log.txt", "a", encoding="utf-8") as f:
            f.write(f"{type(e).__name__} : {e}\n")
        return None
    else:
        return result
    finally:
        print(f"3번 문제 : {a} / {b} 처리 완료")     # 에러가 있든 없든 실행

print("3번 문제 결과 :", safe_divide_with_log(10, 2))
print("3번 문제 결과 :", safe_divide_with_log(10, 0))


# 4번 문제 - 파일에서 특정 단어가 들어간 줄만 골라내기
def find_lines(file_name, keyword):
    try:
        with open(file_name, "r", encoding="utf-8") as f:
            return [line.strip() for line in f if keyword in line]
    except FileNotFoundError:
        return []

print("4번 문제 :", find_lines("log.txt", "Error"))
````
