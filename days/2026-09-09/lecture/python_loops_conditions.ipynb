## 1. 함수(function)

### 1-1. 함수의 기본 구조

```python
def 함수이름(매개변수1, 매개변수2):
    실행할 코드
    return 돌려줄 값
```

- **매개변수(parameter)** : 함수를 정의할 때 받기로 약속한 값
- **인수(argument)** : 함수를 호출할 때 실제로 넘기는 값
- `return` : 결과를 **밖으로 돌려준다**. `return`이 없으면 `None`을 돌려준다.
- **키워드 인수** : `함수(a=10, b=3)` 처럼 이름을 지정해 넘기면 순서를 바꿔도 된다.

````python
# 1번 문제 - 두 수를 곱해서 돌려주는 함수
def multiply(x, y):
    result = x * y
    return result

print("1번 문제 :", multiply(4, 5))


# 2번 문제 - 두 수를 빼는 함수 (키워드 인수로 호출)
def subtract(a, b):
    result = a - b
    return result

# 키워드 인수법 : 매개변수 이름을 직접 지정해서 전달
print("2번 문제 :", subtract(a=10, b=3))


# 3번 문제 - 제곱을 구하는 함수
def power(base, exponent):
    result = base ** exponent
    return result

print("3번 문제 :", power(3, 3))


# 4번 문제 - 두 단어를 공백으로 이어 붙이는 함수
def concat_words(first_word, second_word):
    return first_word + " " + second_word

print("4번 문제 :", concat_words("Hello", "World"))


# 5번 문제 - return 없이 출력만 하는 함수
def print_greeting():
    print("5번 문제 : 파이썬 함수 단원 복습 시작!")

print_greeting()
````

### 2-1. 문자열 메서드

| 메서드 | 하는 일 |
|--------|---------|
| `lower()` / `upper()` | 전부 소문자 / 대문자로 |
| `capitalize()` | 첫 글자만 대문자 |

> 문자열 메서드는 원본을 바꾸지 않고 **새 문자열을 돌려준다.**

### 2-3. 함수 + 메서드 종합 문제 (17 ~ 22번)

````python
# 17번 문제 - 문장을 대문자로 바꿔 돌려주는 함수
def to_upper_sentence(sentence):
    return sentence.upper()

print("17번 문제 :", to_upper_sentence("good morning"))


# 18번 문제 - 리스트에 값을 추가하고 그 리스트를 돌려주는 함수
def add_and_count(target_list, value):
    target_list.append(value)        # 리스트는 함수 안에서 바꿔도 원본이 바뀐다
    return target_list

my_list = [1, 2, 3]
add_and_count(my_list, 4)
print("18번 문제 :", len(my_list), "개 ->", my_list)


# 19번 문제 - 문장을 단어 단위로 잘라 리스트로 돌려주는 함수
def split_words(sentence):
    return sentence.split(' ')

print("19번 문제 :", split_words("data analysis with python"))


# 20번 문제 - 딕셔너리에서 원하는 key의 값을 찾아 문장으로 만들어 주는 함수
def get_info_msg(info_dict, key_name):
    value = info_dict.get(key_name)
    # f-string 안에서 변수를 넣을 때는 중괄호 {} 를 써야 한다 ([] 는 그냥 글자)
    return f"요청하신 {key_name} 정보는 {value}입니다."

person = {'name': '이영희', 'age': 28}
print("20번 문제 :", get_info_msg(person, 'name'))


# 21번 문제 - 날짜 문자열 정리 (공백 제거 후 - 를 / 로 교체)
def format_date(raw_date):
    clean_date = raw_date.strip()             # 앞뒤 공백·줄바꿈 제거
    formatted_date = clean_date.replace("-", "/")
    return formatted_date

print("21번 문제 :", format_date("\n 2026-07-28 \n"))


# 22번 문제 - 딕셔너리에 담긴 value의 개수 세기
def count_values(target_dict):
    return len(target_dict.values())

menu = {'아메리카노': 3000, '라떼': 3500, '티': 4000}
print("22번 문제 :", count_values(menu))
````

## 3. 반복문 (for / while)

### 3-1. for 문 — 정해진 데이터를 하나씩 꺼내기

```python
for 변수 in 반복할_데이터:
    반복할 코드
```

리스트, 문자열, 딕셔너리 등 **여러 값이 들어 있는 자료형**을 앞에서부터 하나씩 꺼내 준다.

````python
# 리스트 반복 : 원소를 하나씩 꺼낸다
food_list = ['햄버거', '피자', '치킨']
for food in food_list:
    print("오늘의 메뉴 :", food)

print("-" * 30)

# 딕셔너리 반복 : 그냥 반복하면 'key'가 꺼내진다
menu_price = {'햄버거': 6000, '피자': 15000}
for menu_name in menu_price:
    print(menu_name, ":", menu_price[menu_name], "원")

print("-" * 30)

# key와 value를 한 번에 꺼내려면 items() 사용
for menu_name, price in menu_price.items():
    print(f"{menu_name}의 가격은 {price}원입니다.")
````

````python
# 문자열 반복 : 글자를 하나씩 꺼낸다
greeting = "Hello World"

for letter in greeting:
    print(letter)
````

### 3-2. range() — 숫자를 자동으로 만들어 주기

- `range(끝)` → `0` 부터 `끝-1` 까지
- `range(시작, 끝)` → `시작` 부터 `끝-1` 까지
- `range()` 자체를 출력하면 숫자가 아니라 `range(0, 5)` 라고만 나온다. → `list()`로 감싸면 눈으로 확인 가능

````python
# range 객체는 그대로 출력하면 내용이 보이지 않는다
print("range 그대로 :", range(0, 5))

# list()로 감싸면 내용을 확인할 수 있다
print("list로 변환  :", list(range(0, 5)))     # 0, 1, 2, 3, 4

print("-" * 30)

# 0 ~ 4 까지 5번 반복
for number in range(0, 5):
    print(f"{number}번째 반복입니다.")
````

### 3-3. while 문 — 조건이 참인 동안 계속 반복

```python
while 조건식:
    반복할 코드
    변수 변경          # 이걸 빠뜨리면 무한 루프!
```

> ⚠️ 파이썬에는 `++count` 같은 증감 연산자가 **없다.**
> `count += 1` (또는 `count -= 1`) 처럼 써야 값이 실제로 바뀐다.

````python
count = 3

while count > 0:                 # count가 0보다 클 동안만 반복
    print(f"Hello World (남은 횟수 {count})")
    count -= 1                   # 반드시 값을 줄여야 반복이 끝난다

print("반복 종료, count =", count)
````

### 3-4. break — 반복 도중에 빠져나오기

`while True:` 는 조건이 항상 참이라 영원히 도는 반복문이다.
이때는 반드시 안쪽에 `break`를 넣어 빠져나갈 길을 만들어 줘야 한다.

````python
# 3부터 0까지 세다가 0이 되면 반복을 멈춘다
count = 3

while True:
    print(f"카운트 : {count}")

    if count == 0:               # 종료 조건
        print("발사!")
        break                    # 반복문 탈출

    count -= 1
````

> 사용자가 직접 숫자를 입력하게 하려면 아래처럼 `input()`을 쓴다.
> (노트북에서 실행하면 입력창이 뜰 때까지 멈추므로 필요할 때만 실행)
>
> ```python
> while True:
>     count = int(input("숫자를 입력하세요 (0 입력 시 종료) : "))
>     if count == 0:
>         print("종료합니다.")
>         break
> ```

## 4. 조건문 (if / elif / else)

### 4-1. if ~ else

```python
if 조건식:
    조건이 참일 때 실행
else:
    조건이 거짓일 때 실행
```

조건식의 결과는 항상 `True` 또는 `False` (bool 자료형)이다.

````python
money = 1500

# 조건식만 따로 확인해 보기 -> True / False
print("1000원 이상인가? :", money >= 1000)

if money > 1000:
    print("택시를 탄다!")
else:
    print("버스를 탄다!")
````

### 4-2. if ~ elif ~ else

조건이 3개 이상일 때 사용한다.
**위에서부터 순서대로 검사**해서 처음 참이 되는 곳만 실행하고 빠져나온다.
→ 그래서 **범위가 좁은 조건(큰 값)부터** 써야 한다.

````python
money = 100

if money >= 10000:
    print("택시를 탄다!")
elif money >= 1300:
    print("버스를 탄다!")
else:
    print("걸어간다!")
````

````python
# 점수로 학점 매기기 (90 이상 A, 70 이상 B, 나머지 C)
score = 90

if score >= 90:
    grade = 'A'
elif score >= 70:
    grade = 'B'
else:
    grade = 'C'

print("학점 :", grade)
````

### 5-2. 함수 + 자료형 다루기 (4 ~ 9번)

````python
# 4번 문제 - 10000원 이상인 상품에만 할인율을 적용하기
def apply_discount(price_dict, discount_rate):
    new_price_dict = {}

    for item_name, price in price_dict.items():
        if price >= 10000:
            # 할인 적용 후 소수점을 버려 정수로 만든다
            new_price_dict[item_name] = int(price * (1 - discount_rate))
        else:
            new_price_dict[item_name] = price      # 할인 대상이 아니면 그대로

    return new_price_dict

price_dict = {'커피': 20000, '치킨': 100, '필통': 1000}
print("4번 :", apply_discount(price_dict, 0.1))     # 10% 할인


# 5번 문제 - 이름의 공백을 없애고 첫 글자를 대문자로
def clean_and_title(raw_name):
    clean_name = raw_name.strip()
    return clean_name.title()

print("5번 :", clean_and_title(" moon "))


# 6번 문제 - 글자 수가 기준보다 긴 단어만 골라내기
def filter_long_words(word_list, min_length):
    long_words = []
    for word in word_list:
        if len(word) > min_length:
            long_words.append(word)
    return long_words

word_list = ["apple", "banana", "cat", "dog", "elephant"]
print("6번 :", filter_long_words(word_list, 5))


# 7번 문제 - 활성 계정인지 확인하기
user = {'name': 'Alice', 'role': 'admin', 'active': True}

# 값이 이미 True/False 이므로 == True 는 붙이지 않아도 된다
if user.get('active'):
    print("7번 : 접근 허용")
else:
    print("7번 : 접근 거부")


# 8번 문제 - 1~20 중 3의 배수만 더하기
total = 0
for number in range(1, 21):
    if number % 3 == 0:          # 3으로 나눈 나머지가 0 -> 3의 배수
        total += number
print("8번 :", total)


# 9번 문제 - 쉼표로 구분된 문자열을 리스트로 바꾸고 각 항목의 공백 제거
def parse_str_to_list(raw_text):
    item_list = raw_text.strip().split(",")     # 앞뒤 공백 제거 후 쉼표로 자르기

    clean_list = []
    for item in item_list:
        clean_list.append(item.strip())         # 항목별 공백도 제거

    return clean_list

print("9번 :", parse_str_to_list("  python, sql , r  "))
````

````python
# 10번 문제 - 로그인 3회 실패 시 계정 잠금
fail_count = 0
while True:
    if fail_count >= 3:
        print(f"10번 : 비밀번호 {fail_count}회 오류 - 계정이 잠깁니다")
        break

    fail_count += 1
    print(f"10번 : 로그인 시도 {fail_count}회")


# 11번 문제 - 학생별 평균을 구해 합격 여부 판정하기
def check_student_pass(student_list):
    for student in student_list:
        scores = student['scores']
        average = round(sum(scores) / len(scores), 1)   # 합계 / 과목 수 = 평균 (소수점 1자리)

        if average >= 80:
            print(f"11번 : {student['name']} 학생은 평균 {average}점으로 합격입니다.")
        else:
            print(f"11번 : {student['name']} 학생은 평균 {average}점으로 불합격입니다.")

# 학생 여러 명을 리스트에 담아 한 번에 전달
student_list = [
    {'name': '김철수', 'scores': [90, 85, 80]},
    {'name': '이영희', 'scores': [60, 75, 70]},
]
check_student_pass(student_list)


# 12번 문제 - 단어를 바꾸고 대문자로 만든 뒤 공백까지 제거하기
def sanitize_and_uppercase(text, old_word, new_word):
    # 메서드를 . 으로 이어서 한 번에 처리 (replace -> upper -> strip 순서로 실행)
    return text.replace(old_word, new_word).upper().strip()

sentence = "   python programming   "
print("12번 :", sanitize_and_uppercase(sentence, "python", "PYTHON"))


# 13번 문제 - 반복문으로 가장 큰 값 찾기
numbers = [15, 42, 88, 23, 71]
max_value = numbers[0]           # 첫 번째 값을 임시 최댓값으로 두고 시작

for number in numbers:
    if number > max_value:       # 더 큰 값을 만나면 교체
        max_value = number

print("13번 :", max_value)


# 14번 문제 - 장바구니 총액 구하기
def calculate_total_price(cart):
    total_price = 0              # sum 은 내장 함수 이름이라 변수명으로 쓰지 않는다
    for price in cart.values():
        total_price += price
    return total_price

cart = {'apple': 3000, 'banana': 2000, 'orange': 50000}
print("14번 :", calculate_total_price(cart))


# 15번 문제 - 이메일에서 도메인만 추출하기
def extract_domain(email):
    # '@'로 자른 뒤 뒷부분([1])을 소문자로
    return email.split('@')[1].lower()

email = "User_Admin@GMAIL.COM"
print("15번 :", extract_domain(email))
````
