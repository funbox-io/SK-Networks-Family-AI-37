# Encoding

```python
# 불필요한 컬럼 삭제
df2 = df1.drop(columns=['customerID'])
display(df2)
```

```python
# 정답 컬럼(Churn) -> label 인코딩 ('No':0, 'Yes':1) -> map 함수
df2['Churn'] = df2['Churn'].map({'No':0, 'Yes':1})
display(df2)
```

```python
# feature -> 범주형 컬럼 -> One-Hot Encoding 적용

## 범주형 컬럼 추출 -> df.select_dtypes(include=[]) 
df_categorical = df2.select_dtypes(include=['object'])
display(df_categorical)

print('-'*80)

## 범주형 컬럼으로 구성된 DataFrame -> 컬럼 이름 추출 -> df.columns
categorical_cols = df_categorical.columns
print(f'범주형 컬럼의 이름: \n{categorical_cols}')

print('-'*80)

## 범주형 컬럼의 이름(Index 자료형) -> 리스트 자료형으로 변환 -> tolist()
categorical_cols_list = categorical_cols.tolist()
print(f'범주형 컬럼 이름의 리스트: \n{categorical_cols_list}')

# 보충: 1줄의 코드로 작성 -> df2.select_dtypes(include=['object']).columns.tolist()

print('-'*80)

## pd.get_dummies() 호출, 원-핫 인코딩 적용
df_onehot = pd.get_dummies(data=df2, columns=categorical_cols_list, dtype=int)

## 결과 확인
display(df_onehot)
```

# 학습용 데이터와 평가용 데이터 생성

```python
# X_data 생성 -> 정답 컬럼 제거
X_data = df_onehot.drop(columns=['Churn'])
display(X_data)

print('-'*80)

# y_data 생성 -> 정답 컬럼 인덱싱
y_data = df_onehot['Churn']
print(y_data)

print('-'*80)

# 정답의 항목별 비율 확인
ratio = y_data.value_counts(normalize=True)
print(f'정답 컬럼의 항목별 비율: \n{ratio}')
```

```python
# 80:20의 비율로 학습용과 평가용 데이터 생성

## 필요한 함수 임포트
from sklearn.model_selection import train_test_split

## train_test_split() 함수 호출 -> 무작위성 통제 -> random_state, stratify
X_train, X_test, y_train, y_test = train_test_split(
    X_data,
    y_data,
    test_size=0.2,
    random_state=0,
    stratify=y_data
)
```

```python
# 학습용 데이터 확인 -> 행 인덱스
print(f'X_train의 행 인덱스: \n{X_train.index}')
print('-'*80)
print(f'y_train의 행 인덱스: \n{y_train.index}')

print('-'*80)

# 정답의 항목별 비율 확인
ratio_train = y_train.value_counts(normalize=True)
print(f'평가용 데이터의 정답의 항목별 비율: \n{ratio_train}')
```

# LogisticRegression(로지스틱 회귀) 모델

## 모델 생성

```python
# 필요한 모델 클래스 임포트
from sklearn.linear_model import LogisticRegression

# 클래스 호출, 모델 생성
lr = LogisticRegression()
```

## 모델 학습: fit()

```python
lr.fit(X_train, y_train)
print('학습 완료')
```

## 학습 완료 후 생성된 최적의 직선 확인: 기울기

```python
# 필요한 라이브러리 임포트
import numpy as np

# 기울기: model.coef_ -> 기울기(가중치)의 크기 추출
feature_importances = np.abs(lr.coef_[0])
print(f'각 컬럼별 직선의 기울기(가중치): \n{feature_importances}')
```

```python
# 기울기(가중치)의 크기와 컬럼 이름 -> DataFrame 생성

## 컬럼 이름(특성 이름)
feature_names = X_train.columns
print(f'특성(컬럼) 이름: \n{feature_names}')

print('-'*80)

## 컬럼 이름과 기울기(가중치)의 크기 -> dict 자료형 생성
dict_data = {
    'Feature': feature_names,
    'Importance': feature_importances
}

## DataFrame 생성 및 내림차순 정렬
df_lr_train_imp = pd.DataFrame(data=dict_data).sort_values(by='Importance', ascending=False)
display(df_lr_train_imp)
```

## 학습용 데이터에 대한 특성의 중요도 시각화

```python
# DataFrame -> 상위 5개 추출 -> df.head(5) 
df_lr_train_imp_top5 = df_lr_train_imp.head(5)
display(df_lr_train_imp_top5)

print('-'*80)

# 막대 그래프 -> top5 시각화 

## 필요한 모듈 임포트
import plotly.express as px

## 그래프 생성 출력
fig1 = px.bar(
    data_frame=df_lr_train_imp_top5,
    x='Feature',
    y='Importance',
    text_auto='.4f',
    title='[LogisticRegession] 학습용 데이터 특성 중요도 Top5'
)
fig1.show()
```

## 평가용 데이터를 이용한 예측: predict()

```python
pred_test_lr = lr.predict(X_test)
print(pred_test_lr)
```

## 모델 평가: accuracy_score() / classification_report()

```python
# 필요한 함수 임포트
from sklearn.metrics import accuracy_score

# 평가용 데이터에 대한 성능 측정
acc_lr = accuracy_score(y_test, pred_test_lr)

# 결과 확인
print(f'LogisticRegression 모델의 정확도: {acc_lr:.4f}')
```

```python
# 필요한 함수 임포트
from sklearn.metrics import classification_report

# 평가용 데이터에 대한 4가지 성능 측정
performance_lr = classification_report(y_test, pred_test_lr)

# 결과 확인
print(f'LogisticRegression 모델의 4가지 성능: \n{performance_lr}')
```

## LogisticRegression 분류 모델 고도화

### 탐색 도구 생성 및 탐색 실행

```python
# 필요한 클래스 임포트
from sklearn.linear_model import LogisticRegression 
from sklearn.model_selection import GridSearchCV

# 성능 고도화 대상 모델 생성
lr = LogisticRegression(random_state=0)

# 최적화 할 매개 변수와 탐색의 조건 설정 -> 5*4=20가지
params = {
    'C': [0.01, 0.1, 1, 10, 100],
    'max_iter': [100, 500, 1000, 5000]
}

# 탐색 도구 생성
grid_lr = GridSearchCV(
    estimator=lr,
    param_grid=params,
    scoring='f1',
    cv=5
)

# 학습 및 검증
grid_lr.fit(X_train, y_train)
print("검증 완료")
```

```python
# best 모델 추출
best_lr = grid_lr.best_estimator_
print(best_lr)
```

```python
# 검증용에 대한 최적의 성능 확인
print(grid_lr.best_score_)
```

#### best 조건을 이용하여 모델 생성

```python
# C=1(기본값), max_iter=500 조건으로 모델 생성
best_lr1 = LogisticRegression(max_iter=500,random_state=0)
print(best_lr1)
```

#### 모델 학습

```python
best_lr1.fit(X_train, y_train)
print("학습 완료")
```

### 학습용 데이터에 대한 특성의 중요도 시각화 

```python
# best 모델: 생성된 최적의 직선 기울기의 크기 확인
best_lr1_train_importances = np.abs(best_lr1.coef_[0])
print(best_lr1_train_importances)
```

```python
# 기울기(가중치)의 크기와 컬럼 이름 -> DataFrame 생성

## 컬럼 이름(특성 이름)
feature_names = X_train.columns
print(f'특성(컬럼) 이름: \n{feature_names}')

print('-'*80)

## 컬럼 이름과 기울기(가중치)의 크기 -> dict 자료형 생성
dict_data = {
    'Feature': feature_names,
    'Importance': best_lr1_train_importances
}

## DataFrame 생성 및 내림차순 정렬
df_best_lr_train_imp = pd.DataFrame(data=dict_data).sort_values(by='Importance', ascending=False)
display(df_best_lr_train_imp)
```

```python
# DataFrame -> 상위 5개 추출 -> df.head(5) 
df_best_lr_train_imp_top5 = df_best_lr_train_imp.head(5)
display(df_best_lr_train_imp_top5)

print('-'*80)

# 막대 그래프 -> top5 시각화 

## 필요한 모듈 임포트
import plotly.express as px

## 그래프 생성 출력
fig2 = px.bar(
    data_frame=df_best_lr_train_imp_top5,
    x='Feature',
    y='Importance',
    text_auto='.4f',
    title='[LogisticRegession] 학습용 데이터 특성 중요도 Top5'
)
fig2.show()
```

### 평가용 데이터를 이용한 예측

```python
pred_test_best_lr = best_lr.predict(X_test)
print(pred_test_best_lr)
```

```python
# 새로운 조건으로 학습시킨 모델
pred_test_best_lr1 = best_lr1.predict(X_test)
print(pred_test_best_lr1)
```
