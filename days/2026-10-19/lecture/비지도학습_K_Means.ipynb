# 비지도 학습_K-Means

```python
# 꽃을 이용한 군집화

import pandas as pd
import plotly.express as px
#from sklearn.cluster import k_means

#데이터 불러오기
path = r"C:\Users\Jinho\work\iris_dataset.csv"

#데이터프레임 생성
df = pd.read_csv(path)

display(df)
```

## 꽃잎 길이와 넓이 관계시각화

```python
fig = px.scatter(
    data_frame=df,
    x='petal width (cm)',
    y='petal length (cm)',
    color='label',
    title='꽃잎 관계 시각화'
)
fig.show()
```

## 군집화에 사용할 데이터 생성

```python
# 꽃잎의 길이, 꽃잎의 넓이
wl_data = df[['petal length (cm)','petal width (cm)']]
#w_data = df['petal width (cm)']
display(wl_data)
```

### K-Means 이용한 군집화

```python
# 필요한 모델 클래스 임포트

from sklearn.cluster import KMeans

#2차원 2개 군집
model1 = KMeans(n_clusters=2,random_state=0)
#예측과 평가 
output1 = model1.fit_predict(wl_data)
print(output1)
#display(output1)
```

```python
#거리 기반 평균으로 경계선(기준선) 만든다.
import numpy as np
print("="*50)
#1의 군집이(몇개로 되어있는지)
label1 = output1[0:50]
#빈도수(항목,빈도수)
counts1 = np.unique(label1,return_counts=True)
print(f"1.군집화:{counts1}")
print("="*50)

#군집이(몇개로 되어있는지)
label2 = output1[50:]
counts2 = np.unique(label2,return_counts=True)
print(f"2.군집화:{counts2}")
print("="*50)
```

## 최적의 클러스터 개수 설정

```python
# 1. SSE 측정
#############################################
# SSE보다 MSE(혹은 RMSE)를 더 자주 사용합니다
# SSE: 오차를 단순히 누적하여 합산한 값 (개수(n)가 많아질수록 무조건 값이 커진다)
# MSE (Mean Squared Error): SSE를 데이터 개수 n으로 나눈 평균 제곱 오차입니다. 
#############################################

## sse 저장용 빈 리스트 생성
sse = []

## x축(군집의 개수) 설정
clusters = range(1, 11)

## KMeans로 군집화 실행, 군집별 sse 추출 및 저장
for num in clusters:
    # num:1~10까지 
    model = KMeans(n_clusters=num, random_state=0)
    # (학습과 평가)
    model.fit_predict(wl_data)
    # SSE 값을 저장
    #최적의 군집 개수(K)를 눈으로 확인하기
    sse.append(model.inertia_)

print(f'sse 측정의 결과 : \n{sse}')
print('-' * 80)

# 2. plotly.express를 활용한 시각화
fig2 = px.line(
    x=clusters,
    y=sse,
    ## 꺾이는 지점(Elbow Point)을 명확하게 파악할 수 있도록 마커 추가
    markers=True,
    title='최적의 클러스터 개수 설정 : 엘보우(elbow) method',
    labels={'x': 'Number of Clusters (k)', 'y': 'SSE'}
)

fig2.show()
```
