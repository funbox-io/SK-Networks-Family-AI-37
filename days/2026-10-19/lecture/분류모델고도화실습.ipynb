# 당뇨병 건강 지표 데이터 분석

```python
import pandas as pd

# 데이터 불러오기
path = r"C:\Users\Jinho\work\diabetes_binary_5050split_health_indicators_BRFSS2015.csv"

# 데이터프레임 생성
df = pd.read_csv(path)

display(df)
```

## 학습용 데이터와 평가용 데이터 생성

```python
# 전체 데이터 학습용 80%: 평가용 20%
from sklearn.model_selection import train_test_split

#X_data 생성
X_data = df.drop(columns=['Diabetes_binary'])
#print(X_data)
print("="*50)

#
y_data = df['Diabetes_binary']
print(y_data)
print("="*50)

#
X_train,X_test,y_train,y_test = train_test_split(
    X_data,
    y_data,
    test_size=0.2,
    random_state=0,
    stratify=y_data
)
```

```python
#학습용 데이터 확인 - > 인덱스 확인
## X_train 확인

print("X_train:",X_train.index)
print("="*50)
print("y_train:",y_train.index)
print("="*50)
```

## DecisionTree 탐색도구를 이용한 고도화

```python
# GridSerchCV(탐색 모델)
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

# 성능 고도화 모델 생성
dt = DecisionTreeClassifier(random_state=0)

#검증용 데이터: 탐색 모델 타입: 딕셔너리
params = { 
    #최대
    'max_depth':[3,4,5,6,7,8,9,10],
    #최소
    'min_samples_leaf':[2,3,4,5]
}

#GridSearchCV 탐색 도구 호출,도구 생생
#하이퍼파라미터
grid_dt = GridSearchCV(
    #고도화 모델
    estimator=dt,
    #최적화 매개변수
    param_grid=params,
    #검증(우수한지 판정할 '평가 기준')
    scoring='accuracy',
    #교차검증(횟수:5회)
    cv=5
)

#학습용
grid_dt.fit(X_train,y_train)
print("탐색 완료!")
```

```python
#beat 모델 추출
best_dt = grid_dt.best_estimator_
#검증 완료 (출력!)
print("검증:",best_dt)
```

```python

# 검증 용,대한 최적의 성능 확인!
print(grid_dt.best_score_)
```

```python
#임포트
from sklearn.metrics import accuracy_score

#평가용 데이터 대한 예측
pred_test_dt = best_dt.predict(X_test)
print("best:",pred_test_dt)
print("="*50)

#평가용 데이터 대한 성능 평가
acc_dt = accuracy_score(y_test,pred_test_dt)
print("성능 측정:",acc_dt)
```
