# chain 실습

## 환경 설정

```python
# 필요한 함수 / 클래스 임포트
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser 
```

```python
# .env file -> API Key 가져오기
env_file_path='.env'
output = load_dotenv(env_file_path)
print(output)
```

## chain 생성

```python
# 첫 번째 구성 요소: 프롬프트

"""
1. 프롬프트 생성 -> ChatPromptTemplate 클래스 -> 전용함수: from_messages() 함수 호출
2. 변수 {topic}의 값 -> 가장 나중에 chain을 실행할 때 전달
"""

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "너는 친절하고 정확한 답변을 하는 AI 비서야. 사용자의 질문에 100 토큰 이내로 간결하게 답변해야 해"),
        ("user", "{topic}에 대해서 설명해 줘")
    ]
)

print(prompt)
```

```python
# 두 번째 구성 요소: llm 모델 생성
llm = ChatOpenAI(model='gpt-5-nano')
print(llm)
```

```python
# 세 번째 구성 요소: 출력 도구 생성 -> 텍스트(문자열)로 결과 전달 기능 -> 클래스 호출
output_parser = StrOutputParser()
print(output_parser)
```

```python
# 세 가지 구성 요소 -> 연결 -> chain 생성 -> '|' 기호 사용
chain = prompt | llm | output_parser
print(chain)
```

## chain 실행

```python
# invoke() 함수 호출 -> chain.invoke({topic}의 값)

"""
1. 사용자 프롬프트의 변수 {topic}의 값 -> invoke 함수의 입력 값 -> 값의 형식: 파이썬 딕셔너리 
2. 예시: chain.invoke({'topic':'양자 컴퓨터'})
"""

# chain 실행
output = chain.invoke({'topic':'양자 컴퓨터'})

# 결과 확인
print(f'chain 실행의 결과: \n{output}')
```
