# 실시간 외부 정보 검색 및 답변 완성 AI 에이전트 구현

## 환경 설정

```python
# 필요한 함수 / 클래스 임포트
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_tavily import TavilySearch
from langchain.agents import create_agent
```

```python
# .env file -> API Key 가져오기
env_file_path='.env'
output = load_dotenv(env_file_path)
print(output)
```

## 외부 검색 도구 생성

```python
# TavilySearch 클래스 호출, 검색 도구 생성
search_tool = TavilySearch(max_results=3, search_depth='advanced')
```

```python
# 에이전트가 사용할 도구 목록 작성
tools = [search_tool]
print(f'도구 목록: {tools}')
```

## AI 에이전트 생성

### LLM 생성

```python
# ChatOpenAI 클래스 호출, llm 생성
llm = ChatOpenAI(model='gpt-5-nano')
```

### 시스템 프롬프트 생성

```python
system_prompt = """
너는 웹 검색 도구를 적극 활용하여 사용자 질문에 정확하고 객관적인 답변을 제공하는 전문 AI 에이전트야.
전달받은 웹 검색 결과에 기반하여 환각(Hallucination) 없이 사실만을 정리해 답변해야 해. 
그리고 답변은 반드시 한국어로 작성할 것
"""
```

### create_agent 함수: 에이전트 생성

```python
# ReAct 에이전트 생성
agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=system_prompt
)
```

## AI 에이전트 실행

```python
# 에이전트: 도구 사용 판단, 실행, 관찰, 답변 생성까지 자동 반복 -> invoke() 함수 실행 -> 딕셔너리: 사용자 프롬프트 전달
response = agent.invoke({
    "messages": [{"role":"user", "content":"2026년 생성형 AI 시장 트렌드 및 최신 AI 에이전트 프레임워크 동향을 간단하게 정리해 줘"}]
})

# 결과 확인
print(f"[최종 에이전트 답변]: \n{response}")
```

```python
# LLM의 최종 답변 -> text 추출
text = response['messages'][-1].content
print(text)
```
