# 외부 검색 도구

## 환경 설정

```python
# 필요한 함수 / 클래스 임포트
from dotenv import load_dotenv
from langchain_tavily import TavilySearch
```

```python
# .env file -> API Key 가져오기
env_file_path='.env'
output = load_dotenv(env_file_path)
print(output)
```

## 외부 검색 도구 생성

```python
# TavilySearch 클래스 호출
search_tool = TavilySearch(max_results=3)
print(search_tool)
```

```python
# 도구 이름(name) 확인
tool_name = search_tool.name
print(tool_name)
```

```python
# 도구 설명(description) 확인
tool_description = search_tool.description
print(tool_description)
```

## 검색 실행

```python
# 사용자: 검색 쿼리 생성
query="Tesla의 2026년 2분기 실적"

# invoke() 함수 호출, 검색 실행, 결과 확인
query_results = search_tool.invoke(query)
print(query_results)
```
