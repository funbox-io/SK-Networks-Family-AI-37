# AutoML

```python
import pandas as pd

#데이터 불러오기
path = r"C:\Users\Jinho\work\Telco-Customer-Churn.csv"
df = pd.read_csv(path)

#데이터프레임 생성
display(df)
```

# 누락

```python
#기본정보 확인
#df.info()
# object 

#TotalCharges 컬럼: 숫자로 변환할 수 없는 문자열 행 ->누락(NaN) 변환 ->제거
#문자열에서 숫자로 변환(X) 에러(X) 누락(O)
#문자열 행 -> 누락(NaN)으로 변환 
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'],errors='coerce')

## 데이터프레임에 누락
df1 = df.dropna(ignore_index=True)
df2 = df1.drop(columns=['customerID'])
display(df1)
```

# 학습용 데이터와 평가용 데이터 생성

```python
#비율(분할 X)
from sklearn.model_selection import train_test_split
#X,Y 분할 하지 않는다.

train_df,test_df = train_test_split(
    df2,
    test_size=0.2,
    random_state=0,
    stratify=df2['Churn']
)
```

```python
#학습용 데이터, 결과 확인
print(f"학습용 데이터:",train_df.index)
print(f"평가용 데이터:",test_df.index)

#결과 비율
train_df_ratin = train_df['Churn'].value_counts(normalize=True)
```

# 학습(모델 생성)

```python
#필요한 클래스 임포트
from autogluon.tabular import TabularPredictor

predicter = TabularPredictor(label='Churn',eval_metric='accuracy',path='AutoGluon_Model')


predicter.fit(
    train_data=train_df,
    presets='best_quality',
    num_gpus=0,
    dynamic_stacking=False,
    num_stack_levels=1,
    time_limit=300
)
```

```python
#학습에 사용된 
leaderboard = predicter.leaderboard()
display(leaderboard)
```

```python
#best 모델
best_model = predicter.model_best
print(best_model)
```

```python
import warnings
## 평가용 데이터를 이용한 성능 평가
# evaluate() (예측과 평가 동시)
performance = predicter.evaluate(test_df)
print(f"고객 이탈 예측 데이터:{performance}")

#balanced accuracy: 정답별로 accuracy 평가 후 평균
#    accuracy(이탈 X) = 100%
#    accuracy(이탈 O) = 0%
#    평균 1 + 0 / 2


warnings.filterwarnings('ignore')
```

# 저장된 MD 이용하기

```python
#모델 지정 경로 설정
model_path = r"C:\Users\Jinho\work\AutoGluon_Model"

#저장된 모델(학습이 완료된 상태)  불러오기
loaded_pred = TabularPredictor.load(model_path)
```

```python
#성능 평가
performance = loaded_pred.evaluate(test_df)
print("저장된 모델을 이용한 성능평가:",performance)
```
