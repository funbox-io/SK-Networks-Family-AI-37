# 목표!: 64차원을 2차원으로 축소

```python
#PCA (선형 축소) 임포트
import numpy as np
import pandas as pd
import plotly.express as px
from sklearn.datasets import load_digits 
from sklearn.decomposition import PCA
```

# 64차원 손글씨 데이터 생성

```python
#MAX(검정):255 , MIN(흰색):0
#다운로드
digits = load_digits()

#print(digits)
X_data = digits.data
#정답 컬럼
y_data = digits.target

#이차원(.shape)
#print(f"(전):{X_data.shape}")
print("="*50)
output = np.unique(ar=y_data,return_counts=True)
#print(f"항목과 빈도수:{output}")

###################################################

#모델 생성
#n_components= 지정한만큼
pca_2d = PCA(n_components=2,random_state=0)

#데이터 변환 (64-2차원)
X_data_2d = pca_2d.fit_transform(X_data)
print(f"(후):{X_data_2d.shape}")

#display(type(X_data_2d))
print(X_data_2d)
```

## 64차원에서 95%를 유지하는 차원 축소

```python
#모델 생성
pca_95 = PCA(n_components=0.95,random_state=0)

#데이터 변환
X_data_95 = pca_95.fit_transform(X_data)

#데이터 출력!
#print(f"(95%):{X_data_95}")
df = pd.DataFrame(data=X_data_95)
display(df)
```
