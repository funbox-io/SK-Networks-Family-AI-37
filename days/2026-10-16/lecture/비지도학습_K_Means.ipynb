# 비지도 학습_K-Means

```python
# 꽃을 이용한 군집화

import pandas as pd
import plotly.express as px
#from sklearn.cluster import k_means

#데이터 불러오기
path = r"C:\Users\Jinho\work\iris_dataset.csv"

#데이터프레임 생성
df = pd.read_csv(path)

display(df)
```

## 꽃잎 길이와 넓이 관계시각화

```python
fig = px.scatter(
    data_frame=df,
    x='petal width (cm)',
    y='petal length (cm)',
    color='label',
    title='꽃잎 관계 시각화'
)
fig.show()
```

## 군집화에 사용할 데이터 생성

```python
# 꽃잎의 길이, 꽃잎의 넓이
wl_data = df[['petal length (cm)','petal width (cm)']]
#w_data = df['petal width (cm)']
display(wl_data)
```

### K-Means 이용한 군집화

```python
# 필요한 모델 클래스 임포트

from sklearn.cluster import KMeans

model1 = KMeans(n_clusters=2,random_state=0)

output1 = model1.fit_predict(wl_data)
print(output1)
#display(output1)
```
