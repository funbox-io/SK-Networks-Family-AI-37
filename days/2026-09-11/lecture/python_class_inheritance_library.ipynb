## 1. 클래스(Class)와 객체(Object)

클래스는 **나만의 커스텀 자료형**이다.
`int` `str` `list` 처럼 파이썬이 미리 만들어 둔 자료형이 있듯이,
내가 다루려는 대상(고객, 상품, 학생 …)을 직접 자료형으로 정의하는 것이다.

| 용어 | 뜻 |
|------|-----|
| **클래스(class)** | 설계도. 어떤 데이터와 기능을 가질지 정의한다 |
| **객체(object) · 인스턴스(instance)** | 설계도로 실제로 찍어낸 것 |
| **속성(attribute)** | 객체가 가지고 있는 **데이터** (`self.name`) |
| **메서드(method)** | 객체가 할 수 있는 **기능** (`purchase()`) |
| **생성자 `__init__()`** | 객체를 만들 때 **자동으로 한 번** 실행되는 초기화 함수 |
| **`self`** | 만들어진 **그 객체 자신**. 모든 메서드의 첫 번째 매개변수 |

이름 규칙 — **클래스명은 대문자로 시작**(`Customer`), **함수·메서드명은 소문자**(`purchase`).

```python
class 클래스이름:
    def __init__(self, 매개변수들):     # 생성자
        self.속성 = 매개변수

    def 메서드이름(self, 매개변수들):    # 인스턴스 메서드
        실행할 코드
```

> **📝 고친 부분**
> - `seif` → **`self`** (오타)
> - "속성: 여러 기능 종류" → 속성은 **데이터**, 기능은 **메서드**. 둘을 나눠 적음
> - "클래스 명(대문자)/함수 명(소문자)" 은 맞는 말이라 그대로 두고 예시를 붙임

### 1-1. 고객 정보 클래스 만들기

````python
class Customer:
    # 생성자 : 객체를 만들 때 자동으로 실행된다
    def __init__(self, name, age, sex):
        # 넘겨받은 값을 객체의 속성으로 저장
        self.name = name
        self.age = age
        self.sex = sex

    # 인스턴스 메서드 : 첫 번째 매개변수는 항상 self
    def purchase(self, price, product_name):
        print(f"[{self.name}] 고객님이 {price:,}원 상당의 {product_name}를 구매하셨습니다.")
````

### 1-2. 객체 생성과 메서드 호출

`객체 = 클래스이름(값1, 값2, ...)` 형태로 만든다.
이때 넘긴 값이 `__init__` 의 매개변수로 들어간다.

````python
# 객체 생성 (키워드 인수로 넘기면 순서를 바꿔도 된다)
obj1 = Customer(name='문진호', age=20, sex='남성')
obj2 = Customer('지현', 25, '여성')          # 순서대로 넘겨도 된다

# 속성 꺼내 쓰기
print("이름 :", obj1.name)
print("나이 :", obj1.age)

# 인스턴스 메서드 호출
obj1.purchase(20000, "아이패드")
obj2.purchase(1500000, "노트북")

# 자료형 확인 -> 내가 만든 Customer 가 하나의 자료형이 되었다
print("자료형 :", type(obj1))
````

> **📝 고친 부분**
> - `test = Customer()` 는 **TypeError**. `__init__` 이 `name, age, sex` 세 개를 요구하므로 값을 빠뜨릴 수 없다.
> - 주석 처리돼 있던 객체 생성·메서드 호출을 실제로 실행되게 풀어 둠

### 1-3. 사실 내장 자료형도 클래스다

필기에 적은 `[]` 와 `list()` 가 같다는 말이 바로 그 얘기다.
`list` 는 파이썬이 미리 만들어 둔 **클래스**이고, `list()` 는 그 클래스로 객체를 만드는 것이다.

> ⚠️ `list(10, 20, 30)` 은 **TypeError**. `list()` 는 값 여러 개가 아니라
> **하나의 반복 가능한 것**을 받는다. → `list([10, 20, 30])`

````python
# [] 와 list() 는 같은 결과
list_data1 = []
list_data2 = list()
print(list_data1, list_data2, list_data1 == list_data2)

# list() 에 값을 여러 개 넘기면 TypeError
try:
    wrong = list(10, 20, 30)
except TypeError as e:
    print("TypeError :", e)

# 리스트 하나를 통째로 넘겨야 한다
print("올바른 사용 :", list([10, 20, 30]))

print("list 의 자료형 :", type(list_data2))
````

## 2. 상속(Inheritance)

이미 만든 클래스를 **그대로 물려받아** 새 클래스를 만드는 것이다.

| 용어 | 뜻 |
|------|-----|
| **부모 클래스 (Super Class)** | 물려주는 쪽 |
| **자식 클래스 (Sub Class)** | 물려받는 쪽 |

- **코드 재사용성** — 부모 코드를 다시 쓰지 않아도 된다
- **유지 보수** — 공통 기능은 부모 **한 곳만** 고치면 된다
- **확장성** — 기존 코드를 두고 필요한 것만 추가한다
- 물려받는 방향은 **부모 → 자식** 한쪽뿐이다. 자식이 부모에게 줄 수는 없다

```python
class 자식클래스(부모클래스):      # 괄호 안에 부모 클래스를 적는다
    ...
```

- `super().__init__(...)` — 부모의 생성자를 불러 공통 속성을 그대로 초기화한다
- **메서드 오버라이딩** — 부모에게 물려받은 메서드를 **같은 이름으로 다시 정의**해 바꿔 쓰는 것

> **📝 고친 부분**
> - `화장성` → **확장성**, `부모 콛드` → **부모 코드** (오타)
> - 클래스명 `VIPCUStomer` → **`VIPCustomer`** (대소문자 오타)

````python
class VIPCustomer(Customer):        # 괄호 안에 부모 클래스
    def __init__(self, name, age, sex, discount_rate=0.1):
        super().__init__(name, age, sex)     # 부모의 __init__ 실행 (공통 속성)
        self.discount_rate = discount_rate   # 자식만 갖는 속성 추가

    # 메서드 오버라이딩 : 부모와 이름·매개변수를 맞춰야 바꿔치기가 된다
    def purchase(self, price, product_name):
        discount_price = int(price * (1 - self.discount_rate))
        print(f"[VIP] [{self.name}] 고객님이 "
              f"[{self.discount_rate * 100:.0f}]% 할인받아 "
              f"{discount_price:,}원에 {product_name}를 구매하셨습니다.")
````

````python
# 자식 클래스로 객체 만들기
vip_data = VIPCustomer("jinho", 30, "male", discount_rate=0.1)

# 물려받은 속성 (부모의 __init__ 이 만들어 준 것)
print("이름 :", vip_data.name, "/ 나이 :", vip_data.age)

# 오버라이딩한 메서드가 대신 실행된다
vip_data.purchase(20000, "아이패드")

# 부모 객체는 원래 메서드 그대로
obj1.purchase(20000, "아이패드")
````

> **📝 고친 부분**
> - `vip_data = VIPCUStomer(...)` 가 **주석 처리된 채** 아래에서 `vip_data.purchase(...)` 를 불러
>   `NameError` 가 나던 것 → 객체 생성을 살림
> - `"mele"` → `"male"` (오타)
> - `self.discount_price` → **지역 변수 `discount_price`**.
>   할인가는 그 순간 계산해 쓰고 마는 값이라 객체 속성으로 남겨 둘 이유가 없다.

## 3. 라이브러리 · 패키지 · 모듈

셋은 **크기 순서**로 이해하면 쉽다.

| 용어 | 단위 | 예 |
|------|------|-----|
| **모듈(module)** | 파이썬 파일 하나 (`.py`) | `random`, `os` |
| **패키지(package)** | 모듈 여러 개를 담은 폴더 | `urllib`, `matplotlib.pyplot` |
| **라이브러리(library)** | 패키지·모듈을 묶어 배포하는 단위 | `requests`, `pandas` |

- **내장(built-in)** — 파이썬을 깔면 이미 들어 있다 (`random`, `os`, `datetime`)
- **외부(external)** — `pip install` 로 따로 설치해야 한다 (`requests`, `pandas`)

**불러오는 두 가지 방법**

```python
import random                  # 모듈 전체 -> random.sample(...) 처럼 모듈명을 붙여 쓴다
from random import sample      # 함수 하나만 -> sample(...) 로 바로 쓴다
```

**대표적인 라이브러리**

| 분야 | 라이브러리 |
|------|-----------|
| 시각화 | `matplotlib` `seaborn` `plotly` |
| 데이터 분석 | `pandas` |
| 웹 요청 | `requests` |

### 3-1. 내장 모듈 `random` — 로또 번호 뽑기

`sample(대상, 개수)` 는 **중복 없이** 원하는 개수를 뽑아 리스트로 돌려준다.

````python
from random import sample

numbers = range(1, 46)          # 1 ~ 45 (46은 포함되지 않는다)

list_data = []
for loop in range(0, 5):        # 5줄짜리 로또 한 장
    info = sample(numbers, 6)   # 중복 없이 6개
    info.sort()                 # 보기 좋게 오름차순 정렬
    list_data.append(info)

for index, ticket in enumerate(list_data, start=1):
    print(f"{index}번째 줄 :", ticket)
````

> **📝 고친 부분**
> - 주석 `중복 없이 6까지 숫자 선택` → 실제로는 **1~45 중에서 6개**를 뽑는 것.
>   `sample()` 이 중복을 막아 주고, `for` 문은 **줄 수**(장당 몇 줄)를 정한다.
> - 결과를 한 줄에 몰아 찍던 것을 줄 단위로 나누고 `sort()` 로 정렬 (실제 로또 용지처럼)

## 4. 외부 라이브러리 `requests` — 웹 페이지 HTML 가져오기

`pip install requests` 로 설치한 뒤 쓴다.
`requests.get(주소)` 가 서버에 요청을 보내고, `.text` 에 응답 본문(HTML)이 들어 있다.

| 속성 | 뜻 |
|------|-----|
| `.status_code` | 응답 코드 (`200` 이면 정상) |
| `.text` | 본문을 문자열로 |
| `.content` | 본문을 바이트로 |

````python
import requests

url = "https://www.google.com"

try:
    output = requests.get(url, timeout=5)
    print("응답 코드 :", output.status_code)
    print("전체 길이 :", len(output.text), "자")
    print("앞부분 300자 :")
    print(output.text[:300])
except Exception as e:
    # 인터넷이 막혀 있는 환경에서도 노트북이 멈추지 않도록
    print("요청 실패 :", type(e).__name__, "-", e)
````

> **📝 고친 부분**
> - `print(output.text)` 는 HTML 전체(수만 자)를 쏟아내 노트북이 무거워진다 → **앞부분 300자만**
> - `status_code` 로 정상 응답인지 먼저 확인하도록 추가
> - 인터넷이 막힌 환경에서도 멈추지 않게 `try` ~ `except` 로 감쌈

## 5. 오늘 정리

- 클래스는 **나만의 자료형**이다. `__init__` 으로 속성을 초기화하고, 메서드로 기능을 만든다.
- `self` 는 **그 객체 자신**. 모든 인스턴스 메서드의 첫 번째 매개변수다.
- `[]` 와 `list()` 가 같은 이유 — 내장 자료형도 결국 **클래스**다.
- 상속은 `class 자식(부모):` 로 하고, 공통 속성은 `super().__init__(...)` 으로 물려받는다.
- **오버라이딩**은 부모 메서드를 **같은 이름·같은 매개변수**로 다시 정의해 바꿔 쓰는 것.
- 크기 순서는 **라이브러리 > 패키지 > 모듈**. `random` 은 내장, `requests` 는 `pip install` 이 필요한 외부 라이브러리.
