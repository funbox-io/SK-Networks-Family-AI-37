## 1. 예외 처리

### 1-1. 예외가 나면 그 줄에서 멈춘다

`10 / 0` 처럼 실행할 수 없는 연산을 만나면 `ZeroDivisionError` 가 나면서 프로그램이 멈춘다.
`try` ~ `except` 로 감싸 두면 멈추지 않고 `except` 쪽 코드로 넘어간다.

````python
def divide_ten(number):
    """10을 number로 나눈 결과를 돌려준다."""
    output = 10 / number
    return output


# 정상 호출
print("정상 호출 :", divide_ten(2))

# 0을 넘기면 ZeroDivisionError -> try 로 감싸서 멈추지 않게 한다
try:
    print("0 호출 :", divide_ten(0))
except ZeroDivisionError as e:
    print("0 호출 실패 :", e)

print("프로그램은 계속 실행됩니다.")
````

> **📝 고친 부분**
> - 함수 이름 `die_tem` → `divide_ten` (하는 일이 드러나는 이름으로)
> - 매개변수 `e` → `number`. `e` 는 보통 `except ... as e` 의 **에러 객체**에 쓰는 이름이라 헷갈린다.

### 1-2. 반복문 안에서 예외 처리하기

`range(0, 10)` 은 **0부터** 시작하므로 첫 번째 반복에서 `10 / 0` 이 되어 에러가 난다.
반복문 안에 `try` 를 두면 **그 회차만 건너뛰고 반복은 계속된다.**

````python
for loop in range(0, 10):
    try:
        result = 10 / loop          # loop 가 0일 때만 에러
    except ZeroDivisionError:
        print(f"{loop} : 0으로 나눌 수 없습니다.")
    else:
        print(f"{loop} : 10 / {loop} = {result}")
````

> **📝 고친 부분**
> - `loop = 10/loop` → `result = 10 / loop`.
>   반복 변수(`loop`)에 결과를 **덮어쓰면** 다음 줄부터 `loop` 가 원래 숫자가 아니게 된다.
> - 성공했을 때 `print("Error")` 가 찍히던 것을 바로잡음. 성공은 `else`, 실패는 `except`.
> - `except:` (전부 잡기) → `except ZeroDivisionError:` (무슨 에러인지 분명하게)

### 1-3. `if / else` 로도 같은 일을 할 수 있다

예외 처리는 조건문과 비슷한 구석이 있다.
**미리 걸러낼 수 있는 조건**(0인지 확인)은 `if` 로, **미리 알 수 없는 상황**은 `try` 로 처리한다.

````python
for num in range(0, 10):
    if num == 0:
        print(f"{num} : 0으로 나눌 수 없습니다.")
    else:
        print(f"{num} : 10 / {num} = {10 / num}")
````

> **📝 고친 부분**
> - `print("1으로 나눌 수 있습니다.")` 는 `num` 이 1이 아닐 때도 "1"이라고 나온다.
>   f-string 으로 실제 값을 넣어 출력하도록 고침.

## 2. 명시적 예외 처리

`except` 뒤에 에러 이름을 적으면 **그 에러만** 잡는다.
무슨 에러인지 정확히 알고 대응할 수 있는 대신, 적어 두지 않은 에러는 잡지 못한다.

### 2-1. `NameError` — 없는 변수를 부를 때

````python
name = "진호"

# 정의한 변수는 잘 나온다
print(f"나라를 구하는 이름 : {name}")

# 정의하지 않은 변수를 부르면 NameError
try:
    print(f"두 번째 이름 : {second_name}")
except NameError as e:
    print("NameError :", e)
````

>   `NameError` 를 보려면 **정의한 적 없는 이름**(`second_name`)을 불러야 한다.

### 2-2. 에러 여러 개를 한 번에 처리하기

에러를 여러 개 잡는 방법은 두 가지다.

```python
except (ZeroDivisionError, ValueError) as e:   # 1) 괄호로 묶어 튜플로
    ...

except ZeroDivisionError as e:                 # 2) except 절을 나눠서
    ...
except ValueError as e:
    ...
```

> ⚠️ `as` 는 **한 번만** 쓴다. `except A as B as :` 같은 문법은 없다.
> 그리고 `as e` 로 받은 `e` 는 **그 `except` 블록 안에서만** 살아 있다. `finally` 에서 쓰면 `NameError` 가 난다.

````python
for num in range(0, 3):
    try:
        output = 10 / num
    except (ZeroDivisionError, ValueError) as e:   # 괄호로 묶어 한 번에
        print(f"{num} : 에러 발생 -", e)
    else:
        print(f"{num} : 결과 {output}")
    finally:
        # finally 는 에러가 있든 없든 항상 실행된다 (여기서 e 를 쓰면 안 된다)
        print(f"{num} : 처리 완료")
````

> **📝 고친 부분**
> - `except ZeroDivisionError as FileNotFoundError as :` → `except (ZeroDivisionError, ValueError) as e:`
> - `finally` 안의 `print(e)` 를 지움. `e` 는 `except` 블록을 벗어나면 사라진다.
> - 오타 `ouput` → `output`

## 4. 파일 처리

### 4-1. 경로와 `with` 문

윈도우 경로의 역슬래시는 escape 문자로 해석되므로 아래 셋 중 하나로 쓴다.

```python
path = r"C:\Users\Jinho\miniconda3\work\테스트.txt"   # r 을 붙이거나
path = "C:/Users/Jinho/miniconda3/work/테스트.txt"        # / 를 쓰거나
path = "C:\\Users\\Jinho\\miniconda3\\work\\테스트.txt"    # \\ 로 두 번 쓰거나
```

`with open(...) as f:` 를 쓰면 블록이 끝날 때 **자동으로 닫힌다.**
한글이 깨지지 않도록 `encoding="utf-8"` 을 붙인다.

````python
# 실습은 이 노트북과 같은 폴더에 파일을 만든다
file_path = "테스트.txt"

# 쓰기 : mode='w' 는 파일이 있으면 기존 내용을 모두 지우고 새로 쓴다
with open(file=file_path, mode='w', encoding="utf-8") as f:
    for num in range(1, 11):
        data = f"{num}번째 줄입니다.\n"     # 줄바꿈은 \n 을 직접 붙인다
        f.write(data)

    print("마지막 번호 :", num)
# with 블록을 벗어나면 자동으로 close() 된다

print("파일이 닫혔는가? :", f.closed)
````

> **📝 고친 부분**
> - `with` 안에 있던 `f.close()` 를 지움. `with` 를 쓰면 **자동으로 닫히므로 필요 없다.**
> - `print("길이:", num)` → 여기서 `num` 은 길이가 아니라 **마지막 반복 번호**라서 문구를 바꿈.
>   실제 줄 수는 아래 4-2 에서 `len(readlines())` 로 센다.
> - `encoding="utf-8"` 추가 (한글이 깨지지 않도록)

### 4-2. 읽기 — `read()` / `readline()` / `readlines()`

| 메서드 | 결과 |
|--------|------|
| `read()` | 파일 전체를 **하나의 문자열**로 |
| `readline()` | **한 줄만** 문자열로 |
| `readlines()` | 전체를 **줄 단위 리스트**로 |

````python
# 전체를 하나의 문자열로
with open(file=file_path, mode='r', encoding="utf-8") as f:
    cont1 = f.read()
print("read() 앞부분 :", cont1[:20].replace("\n", " / "))

# 한 줄만
with open(file=file_path, mode='r', encoding="utf-8") as f:
    cont2 = f.readline()
print("readline() :", cont2.strip())

# 줄 단위 리스트로
with open(file=file_path, mode='r', encoding="utf-8") as f:
    cont3 = f.readlines()
print("readlines() 첫 두 줄 :", cont3[:2])
print("줄 수 :", len(cont3))
````

### 4-3. 파일 처리 + 예외 처리

없는 파일을 `'r'` 로 열면 `FileNotFoundError` 가 난다.
`try` 로 감싸 두면 프로그램이 멈추지 않는다.

````python
def read_file(path):
    try:
        with open(file=path, mode='r', encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        return f"{path} 파일이 없습니다."

print("첫 줄 :", read_file(file_path).splitlines()[0])
print(read_file("없는파일.txt"))
````

## 5. 오늘 정리

- 예외 처리는 **에러가 나도 프로그램이 멈추지 않게** 해 준다. `try` → `except` → (`else`) → (`finally`)
- 미리 걸러낼 수 있는 조건은 `if`, 미리 알 수 없는 상황은 `try` 로 처리한다.
- `as e` 로 받은 에러 객체는 **그 `except` 블록 안에서만** 쓸 수 있다.
- 에러 여러 개는 `except (A, B) as e:` 처럼 **괄호로 묶어** 한 번에 잡는다.
- 윈도우 경로는 `r"..."` 또는 `/` 로 쓴다. 한글 파일은 `encoding="utf-8"`.
- `with open(...) as f:` 를 쓰면 `close()` 를 따로 부르지 않아도 된다.
- `tuple`, `sum` 처럼 **내장 이름은 변수명으로 쓰지 않는다.**

````python
# 실습으로 만든 파일 정리
import os

try:
    os.remove(file_path)
    print("삭제 :", file_path)
except FileNotFoundError:
    print("이미 없음 :", file_path)
````
