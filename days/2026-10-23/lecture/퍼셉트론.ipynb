## 퍼셉트론 실습

```python
def AND(x1,x2):
    # 파이썬에 리스트는 딥러닝에 연산이 안된다?
    x = np.array([x1,x2])
    w = np.array([0.5,0.5])

    # 편향
    b = -0.7 # AND
    output = np.sum(x*w)+b

    step_func(output)
    """
    if output >= 0:
        return 1
    else:
        return 0
    """
```

```python
# 가중치와 편향을 활용한 퍼셉트론 표현-> NAND 구현

def NAND(x1,x2):
    x=np.array([x1,x2])
    w=np.array([-0.5,-0.5])

    # 편향
    b = 0.7
    output = np.sum(x*w) + b
    #print(output) #입력 신호의 총합

    step_func(output)
    """
    if output >= 0:
        return 1
    else: 
        return 0
    """
```

```python
# 가중치와 편향을 활용한 퍼셉트론 표현 -> OR 구현

def OR(x1,x2):
    x=np.array([x1,x2])
    w=np.array([0.5,0.5])

    b = -0.2 # OR 값
    output = np.sum(x*w)+b

    step_func(output)
    """
    if output >= 0:
        return 1
    else:
        return 0
    """
```

```python
# 가중치와 편향을 활용한 퍼셉트론 표현 -> XOR 구현

# MLP 구조
def XOR(x1,x2):
    # 1은닉층(h1) 서로 다른 게이트
    s1 = NAND(x1,x2)
    s2 = OR(x1,x2)
    # 2은닉층(h2) 출력층
    output = AND(s1,s2)

    # 출력 
    return output
```

```python
#계단함수 정의 입력 값이 0 미만 이면 '0', 0 이상 이면 1

def step_func(x):
    output = np.where(x>=0,1,0)
    print(output)
```

```python
def perceptron(x1,x2):
    x=np.array([x1,x2])
    w=np.array([0.5,0.5])

    b = 0.7

    output = np.sum(x*w)+b
    return output
```

```python
# 인공 뉴런() -> 활성화 함수 연결 -> AND 구현
inputs = [(0,0),(0,1),(1,0),(1,1)]
print("AND")
print("="*50)
for x in inputs:
    x1 = x[0]
    x2 = x[1]

    #퍼셉트론 함수 호출
    output = perceptron(x1,x2)
    print("1.OUTPUT:",output)
    #활성화 함수(계단 함수)

    output_2 = step_func(output)
    print("결과:",output_2)
    print("="*50)
```

## 시그모이드 함수 

```python
# e-z
# 0과 1사이의 연속적인 확률값 제공

def sigmold(x):
    return 1/(1 + np.exp(-x))
```

```python
# Z
inputs = [(0,0),(0,1),(1,0),(1,1)]

# for문
for x in inputs:
    x1 = x[0]
    x2 = x[1]

    output1 = perceptron(x1,x2)
    print("퍼셉트론 함수:",output1)
    output2 = sigmold(output1)
    print("시그모이드 함수:",output2)
```

## ReLu 함수

```python
# 은닉층: 퍼셉트로노의 입력 신호의 총합 -> 0이하는 0으로 차단, 양수는 그대로 통과!
def ReLu(x):
    return np.maximum(0,x)
```

```python
# 은닉충: 퍼셉트론 입력 신호의 총합 1 > 가상으로 설정(7개)
#0,1,0,0,0,2,5
perceptron_ouput = np.array([-0.5,1.0,-2.0,-0.5,0.0,2.0,5.0])

relu_output = ReLu(perceptron_ouput)
print("ReLu함수: ",relu_output)
```

## 소프트 맥스(SoftMax) 함수

```python
def softmax(arr):
    # 분자 부분 : 각 출력 뉴런(의 입력신호의 총합) 별로 가각 계산
    exp_arr = np.exp(arr)
    # 분모 부분: 각 출력 뉴런 별로 계산한것 -> 총합
    sum_exp_arr = np.sum(exp_arr)
    output = exp_arr / sum_exp_arr
    return output
    #print(output)
```

```python
#소프트 맥스 가상 정답(3개) 클래스 분류
#출력층-> 출력 뉴련 가상 3개 -> 가상 설정

output_pert_sum = np.array([0.3,2.9,4.0])
soft_num = softmax(output_pert_sum)
print(soft_num)
```
