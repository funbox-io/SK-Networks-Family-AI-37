# 당뇨병 건강 지표 데이터 분석

## 데이터 불러오기

```python
# 필요한 라이브러리 임포트
import pandas as pd

# 파일 경로 설정
file_path='C:/Users/playdata2/work/diabetes_binary_5050split_health_indicators_BRFSS2015.csv'

# DataFrame 생성
df = pd.read_csv(file_path)

# 결과 확인
display(df)
```

## 학습용과 평가용 데이터 생성

```python
# 전체 데이터 -> 학습용 : 평가용 = 80 : 20으로 분할

## 필요한 함수 임포트
from sklearn.model_selection import train_test_split

## X_data 생성 -> 정답 컬럼 제거
X_data = df.drop(columns=['Diabetes_binary'])
display(X_data)

print('-'*80)

## y_data 생성 -> 정답 인덱싱
y_data = df['Diabetes_binary']
print(y_data)

print('-'*80)

## train_test_split() 함수 호출 -> 무작위성 통제 -> random_state, stratify 적용
X_train, X_test, y_train, y_test = train_test_split(
    X_data,
    y_data,
    test_size=0.2,
    random_state=0,
    stratify=y_data
)
```

```python
# 학습용 데이터 확인 -> 인덱스 확인

## X_train 확인
print(f'X_train 행 인덱스: \n{X_train.index}')

print('-'*80)

## y_train 확인
print(f'y_train 행 인덱스: \n{y_train.index}')
```

## DecisionTree 분류 모델 고도화 

### GridSearchCV 탐색 도구를 이용한 고도화

#### 탐색 도구 생성 및 탐색 실행

```python
# 필요한 클래스 임포트
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

# 성능 고도화 대상 모델 생성
dt = DecisionTreeClassifier(random_state=0)

# 최적화 할 매개 변수와 탐색의 조건 설정 -> key:value -> dict 자료형
params = {
    'max_depth': [3,4,5,6,7,8,9,10],
    'min_samples_leaf': [2,3,4,5]
}

# GridSearchCV 탐색 도구 호출, 도구 생성
grid_dt = GridSearchCV(
    estimator=dt,
    param_grid=params,
    scoring='accuracy',
    cv=5
)

# 탐색 도구 학습 및 검증
grid_dt.fit(X_train, y_train)
print("탐색 완료")
```

```python
# best 모델 추출
best_dt = grid_dt.best_estimator_
print(best_dt)
```

```python
# 검증용에 대한 최적의 성능 확인
print(grid_dt.best_score_)
```

#### 평가용 데이터를 이용한 예측

```python
pred_test_dt = best_dt.predict(X_test)
print(pred_test_dt)
```

#### 모델 평가

```python
# 필요한 함수 임포트
from sklearn.metrics import accuracy_score

# 평가용 데이터에 대한 성능 측정
acc_dt = accuracy_score(y_test, pred_test_dt)

# 결과 확인
print(f'고도화된 DecisionTree 분류 모델의 평가용 데이터 대한 성능: {acc_dt:.4f}')
```

```python
# 평가 지표 변경: accuracy_score -> classification_report로 변견

## 필요한 함수 임포트
from sklearn.metrics import classification_report

## 평가용 데이터에 대한 4가지 성능 측정
performance_dt = classification_report(y_test, pred_test_dt) 

## 결과 확인
print(f"고도화된 DecisionTree 분류 모델의 평가용 데이터에 대한 4가지 성능: \n{performance_dt}")
```

## Stacking 분류 모델

### 1단계: 개별 기본 모델 생성

```python
# 필요한 클래스 임포트
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
# 앞에서 고도화를 완료한 개별 모델을 Stacking의 기본 모델로 지정
base_models = [
    ('dt', best_dt),
    ('rf', best_rf),
    ('xgb', best_xgb)
]
```

### 2단계: 최종 메타 모델 생성

```python
# 필요한 클래스 임포트
from sklearn.linear_model import LogisticRegression

# 1단계 모델들의 예측값을 받아 최종 판단을 내리는 모델 생성
final_meta = LogisticRegression(max_iter=1000)
```

### Stacking 모델 생성

```python
# 필요한 클래스 임포트
from sklearn.ensemble import StackingClassifier

# base_models, final_meta 결합
stack = StackingClassifier(
    estimators=base_models,
    final_estimator=final_meta
)
```

### 모델 학습

```python
stack.fit(X_train, y_train)
print("학습 완료")
```

### 평가용 데이터에 대한 예측

```python
pred_test_stack = stack.predict(X_test)
```

### 모델 평가

```python
# 필요한 함수 임포트
from sklearn.metrics import accuracy_score, classification_report

# 평가용 데이터에 대한 4가지 성능 측정
performance_stack = classification_report(y_test, pred_test_stack)

# 결과 확인
print(f'Stacking 분류 모델의 평가용 데이터에 대한 4가지 성능: \n{performance_stack}')
```
