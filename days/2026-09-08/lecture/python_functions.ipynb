# 파이썬 함수(Function) 정리

함수의 개념과 **사용자 정의 함수** 만드는 법, 그리고 자료형별로 딸려 있는
**리스트 · 딕셔너리 · 문자열 전용 함수(메서드)** 를 정리한 학습 노트입니다.
각 단원은 **개념 설명 → 예제 코드 → 연습 문제** 순서로 구성되어 있습니다.

> 변수 이름은 담긴 데이터가 무엇인지 바로 알 수 있도록 지었습니다.
> (`data1`, `output1` 처럼 뜻을 알 수 없는 이름 대신 `total_score`, `shopping_list` 처럼 사용)

## 1. 함수(function)란?

같은 코드를 여러 번 쓰지 않도록 **하나의 이름으로 묶어 둔 기능 덩어리**다.
재료(입력)를 넣으면 결과(출력)를 돌려주는 자판기라고 생각하면 쉽다.

| 종류 | 설명 | 예 |
|------|------|-----|
| **내장 함수** | 파이썬이 기본으로 갖고 있어 바로 쓸 수 있는 함수 | `print()` `len()` `type()` |
| **전용 함수(메서드)** | 특정 자료형에만 딸려 있는 함수. `변수.함수()` 형태로 부른다 | `리스트.append()` `문자열.split()` |
| **사용자 정의 함수** | 내가 직접 `def` 로 만들어 쓰는 함수 | `def add(x, y):` |

### 1-1. 함수 정의 — `def`

```
def 함수이름(매개변수1, 매개변수2):
    실행할 코드
    return 돌려줄 값
```

- **매개변수(parameter)** : 함수를 *정의*할 때 적는 이름 (`x`, `y`)
- **인수(argument)** : 함수를 *호출*할 때 실제로 넣는 값 (`3`, `5`)
- `return` : 결과를 밖으로 돌려준다. 없으면 `None` 을 돌려준다.

````python
# 두 수를 더해서 돌려주는 함수 정의
def add(first_number, second_number):
    total = first_number + second_number   # 연산
    return total                           # 결과를 돌려준다
````

### 1-2. 함수 호출

정의만 해서는 아무 일도 일어나지 않는다. **이름(인수)** 형태로 불러야 실행된다.

````python
# 호출 1 : 3과 5를 넣으면 8이 돌아온다
sum_result = add(3, 5)
print("3 + 5 =", sum_result)

print("-" * 80)

# 호출 2 : 같은 함수를 다른 값으로 몇 번이든 재사용할 수 있다
print("2 + 4 =", add(2, 4))
print("10 + 20 =", add(10, 20))
````

### 1-3. 기본값이 있는 매개변수

매개변수에 `= 값` 을 적어 두면, 호출할 때 그 자리를 **생략**할 수 있다.
생략하면 기본값이, 값을 넣으면 넣은 값이 쓰인다.

````python
# tax_rate 를 생략하면 0.1(10%)이 쓰인다
def price_with_tax(price, tax_rate=0.1):
    return price * (1 + tax_rate)

print("기본 세율 10% :", price_with_tax(10000))
print("세율 5% 지정  :", price_with_tax(10000, 0.05))
````

### 1-4. `return` 이 없는 함수

`return` 을 적지 않으면 함수는 **`None`** 을 돌려준다.
화면에 출력만 하는 함수와, 값을 돌려주는 함수는 다르다는 점을 기억하자.

````python
def print_greeting(name):
    print(f"안녕하세요, {name}님!")   # 출력만 하고 돌려주는 값은 없다

returned_value = print_greeting("김진호")
print("돌려받은 값 :", returned_value)   # None
````

### 1-5. 함수 연습 문제

````python
# 1번 - 두 수를 곱해서 돌려주는 함수
def multiply(first_number, second_number):
    return first_number * second_number

print("1번 문제 :", multiply(6, 7))

# 2번 - 점수 리스트를 받아 평균을 돌려주는 함수
def average(score_list):
    return sum(score_list) / len(score_list)

print("2번 문제 :", average([88, 95, 72]))

# 3번 - 짝수면 True, 홀수면 False 를 돌려주는 함수
def is_even(number):
    return number % 2 == 0

print("3번 문제 :", is_even(10), is_even(7))
````

## 2. 리스트 전용 함수

리스트 변수 뒤에 점을 찍고 부르는 함수들이다. (`리스트.함수()`)

| 함수 | 하는 일 |
|------|---------|
| `append(값)` | 맨 **뒤에 하나** 추가 |
| `insert(위치, 값)` | 원하는 **위치에 끼워 넣기** |
| `extend(리스트)` | 다른 리스트의 값들을 **모두 이어 붙이기** |
| `clear()` | 전체 비우기 |
| `len(리스트)` | 길이 세기 (내장 함수) |

### 2-1. `append()` — 맨 뒤에 하나 추가

⚠️ `append()` 는 **원본 리스트를 직접 바꾸고, 돌려주는 값은 `None`** 이다.
그래서 `새변수 = 리스트.append(값)` 처럼 쓰면 `None` 이 담긴다.

````python
number_list = [0, 1, 2, 3, 4]
print("추가 전 :", number_list)

number_list.append(10)          # 원본이 바뀐다
print("추가 후 :", number_list)

print("*" * 80)

# 흔한 실수 : 결과를 변수에 담으면 None 이 들어온다
wrong_result = number_list.append(20)
print("잘못된 사용 :", wrong_result)
print("원본은 바뀜 :", number_list)
````

````python
# 장바구니에 물건을 하나씩 담아 보기
shopping_list = []

shopping_list.append('빵')
print("담은 뒤 :", shopping_list)

shopping_list.append('우유')
print("담은 뒤 :", shopping_list)

shopping_list.append('사과')
print("담은 뒤 :", shopping_list)
````

### 2-2. `insert()` — 원하는 위치에 끼워 넣기

`리스트.insert(위치, 값)` — 그 위치부터 뒤의 값들은 한 칸씩 밀린다.

````python
waiting_line = ['김진호', '이지현', '박서준']
print("끼워 넣기 전 :", waiting_line)

waiting_line.insert(0, '최유리')       # 맨 앞에 새치기
print("맨 앞에 추가 :", waiting_line)

waiting_line.insert(2, '정민수')       # 2번 자리에 끼워 넣기
print("2번에 추가   :", waiting_line)
````

### 2-3. `extend()` 와 `+` — 리스트 합치기

| 방법 | 원본 | 결과 |
|------|------|------|
| `A.extend(B)` | **A가 직접 바뀐다** | 돌려주는 값 없음(`None`) |
| `A + B` | 원본은 그대로 | **합쳐진 새 리스트**를 돌려준다 |

````python
basket = ['계란']
extra_items = ['사과', '두부']

basket.extend(extra_items)         # basket 자체가 늘어난다
print("extend 결과 :", basket)

print("-" * 80)

# + 연산자 : 원본은 그대로 두고 새 리스트를 만든다
first_list = ['계란', '우유']
second_list = ['사과']
merged_list = first_list + second_list

print("first_list  :", first_list)     # 그대로
print("merged_list :", merged_list)    # 새로 합쳐진 리스트
````

### 2-4. `clear()` 와 `len()`

````python
number_list = [1, 2, 3, 4, 5, 6]
print("길이 :", len(number_list))       # 값이 몇 개인지 세어 준다

number_list.clear()                     # 전부 비우기
print("비운 뒤 :", number_list, "/ 길이 :", len(number_list))
````

### 2-5. 리스트 함수 연습 문제

````python
# 1번 - 빈 리스트에 좋아하는 과일 3개를 하나씩 추가하기
fruit_list = []
fruit_list.append('딸기')
fruit_list.append('포도')
fruit_list.append('수박')
print("1번 문제 :", fruit_list)

# 2번 - 맨 앞에 '망고' 끼워 넣기
fruit_list.insert(0, '망고')
print("2번 문제 :", fruit_list)

# 3번 - 다른 리스트를 이어 붙이기
tropical_fruits = ['파인애플', '바나나']
fruit_list.extend(tropical_fruits)
print("3번 문제 :", fruit_list)

# 4번 - 전체 개수 세기
print("4번 문제 :", len(fruit_list), "개")
````

## 3. 딕셔너리 전용 함수

| 함수 | 하는 일 | 돌려주는 값 |
|------|---------|------------|
| `keys()` | key 전체 | `dict_keys([...])` |
| `values()` | value 전체 | `dict_values([...])` |
| `items()` | key와 value 쌍 전체 | `dict_items([(k, v), ...])` |
| `get(key)` | 값 하나 꺼내기 | 없으면 **`None`** (에러 없음) |
| `clear()` | 전체 비우기 | 없음 |

### 3-1. `keys()` — key만 꺼내기

돌려주는 값은 리스트가 아니라 `dict_keys` 라는 전용 자료형이다.
리스트처럼 쓰고 싶으면 `list()` 로 바꿔 준다.

````python
member = {'name': '이지현', 'age': 20, 'phone': '010-2000-0000'}

key_data = member.keys()
print("전체 key :", key_data)

print("*" * 80)

print("자료형   :", type(key_data))
print("리스트로 :", list(key_data))     # list() 로 바꾸면 다루기 쉽다
````

### 3-2. `values()` — value만 꺼내기

````python
actor = {'name': '박보검', 'age': 31, 'phone': '010-0000-0000'}

value_data = actor.values()
print("전체 value :", value_data)
print("리스트로   :", list(value_data))
````

### 3-3. `items()` — key와 value를 한 번에

`(key, value)` 튜플들의 묶음을 돌려준다. 반복문에서 특히 많이 쓴다.

````python
actor = {'name': '박보검', 'age': 31, 'phone': '010-0000-0000'}

item_data = actor.items()
print("전체 key, value :", item_data)

print("-" * 80)

# key와 value를 한 쌍씩 꺼내 쓰기
for key, value in actor.items():
    print(f"{key} : {value}")
````

### 3-4. `get()` — 에러 없이 값 꺼내기

| 방법 | 없는 key를 찾으면 |
|------|------------------|
| `딕셔너리[key]` | **KeyError** 발생 |
| `딕셔너리.get(key)` | `None` 반환 (에러 없음) |

`get()` 은 값을 *읽기만* 한다. 없는 key를 만들어 주지는 않는다.

````python
actor = {'name': '박보검', 'age': 31, 'phone': '010-0000-0000'}

print("이름 :", actor.get('name'))
print("주소 :", actor.get('address'))          # 없는 key -> None
print("주소(기본값) :", actor.get('address', '정보 없음'))   # 기본값 지정도 가능

print("-" * 80)

# 대괄호로 없는 key를 찾으면 에러가 난다
try:
    print(actor['address'])
except KeyError as error:
    print("KeyError 발생 :", error)
````

### 3-5. 딕셔너리 함수 연습 문제

````python
product = {'name': '커피', 'price': 5000, 'stock': 100}

# 1번 - key 목록을 리스트로 출력하기
print("1번 문제 :", list(product.keys()))

# 2번 - value 목록을 리스트로 출력하기
print("2번 문제 :", list(product.values()))

# 3번 - key와 value를 한 줄씩 출력하기
print("3번 문제 :")
for key, value in product.items():
    print(f"  {key} : {value}")

# 4번 - 없는 key를 get() 으로 안전하게 꺼내기
print("4번 문제 :", product.get('color', '색상 정보 없음'))
````

## 4. 문자열 전용 함수

| 함수 | 하는 일 |
|------|---------|
| `lower()` / `upper()` | 소문자 / 대문자로 바꾸기 |
| `split(기준)` | 기준으로 **잘라서 리스트**로 만들기 |
| `기준.join(리스트)` | 리스트를 기준으로 **이어 붙여 문자열**로 만들기 |
| `replace(찾을값, 바꿀값)` | 바꿔치기 |
| `strip()` | 앞뒤 공백·줄바꿈 지우기 |

⚠️ 문자열은 **immutable(변경 불가)** 이라서, 이 함수들은 원본을 바꾸지 않고
**새 문자열을 돌려준다.** 그래서 결과를 꼭 변수에 담아야 한다.

### 4-1. `lower()` / `upper()` — 대소문자 바꾸기

````python
language_name = "PYTHON"

lower_name = language_name.lower()
upper_name = "python".upper()

print("소문자로 :", lower_name)
print("대문자로 :", upper_name)
print("원본은 그대로 :", language_name)   # immutable
````

### 4-2. `split()` — 문자열을 잘라 리스트로

- 기준을 적지 않으면 **공백**을 기준으로 자른다.
- 결과는 **리스트**다.

````python
# 쉼표를 기준으로 자르기
error_names = "Arithmetic,Value,Type"
error_list = error_names.split(',')
print("쉼표 기준 :", error_list)

print("-" * 80)

# 기준 없이 : 공백을 기준으로 자른다
sentence = "Life is too short"
word_list = sentence.split()
print("공백 기준 :", word_list)
print("단어 개수 :", len(word_list))
````

### 4-3. `join()` — 리스트를 이어 붙여 문자열로

`split()` 의 반대 기능이다. **이어 붙일 기준 문자열**에 점을 찍고 부른다.

````python
colors = ['red', 'blue', 'yellow']

comma_joined = ','.join(colors)      # 쉼표로 연결
space_joined = ' '.join(colors)      # 공백으로 연결

print("쉼표 연결 :", comma_joined)
print("공백 연결 :", space_joined)
````

### 4-4. `replace()` — 바꿔치기

````python
sentence = "Life is too short"
print(f"함수 사용 전 :\n{sentence}")

print("*" * 80)

replaced = sentence.replace("Life", "My leg")
print(f"함수 사용 후 :\n{replaced}")
````

### 4-5. `strip()` — 앞뒤 공백·줄바꿈 지우기

사용자가 입력한 값이나 파일에서 읽은 값을 정리할 때 자주 쓴다.

````python
messy_text = "\n\n  BaseExceptionGroup  \n"

print(f"정리 전 : [{messy_text}]")
print(f"정리 후 : [{messy_text.strip()}]")
````

### 4-6. 문자열 함수 연습 문제

````python
# 1번 - 이메일에서 아이디와 도메인 분리하기
email = "funbox@example.com"
email_parts = email.split('@')
print("1번 문제 :", email_parts[0], "/", email_parts[1])

# 2번 - 대문자로 바꿔 출력하기
print("2번 문제 :", email.upper())

# 3번 - 리스트를 ' - ' 로 이어 붙이기
menu = ['아메리카노', '라떼', '녹차']
print("3번 문제 :", ' - '.join(menu))

# 4번 - 문장에서 특정 단어 바꾸기
notice = "오늘 수업은 오전에 진행됩니다."
print("4번 문제 :", notice.replace("오전", "오후"))

# 5번 - 지저분한 입력값 정리 후 길이 재기
user_input = "   김진호  \n"
clean_input = user_input.strip()
print("5번 문제 :", f"[{clean_input}]", "/ 길이 :", len(clean_input))
````

## 5. 오늘 정리

- 함수는 **기능을 이름으로 묶어 재사용**하는 도구다. `def` 로 만들고 `return` 으로 결과를 돌려준다.
- `append()`, `extend()`, `clear()` 처럼 **원본을 직접 바꾸는 함수**는 돌려주는 값이 `None` 이다.
  → `새변수 = 리스트.append(값)` 은 흔한 실수.
- `split()`, `join()`, `replace()`, `strip()` 처럼 **새 값을 돌려주는 함수**는 결과를 변수에 담아야 한다.
  → 문자열은 immutable 이라 원본이 바뀌지 않는다.
- 없는 key 가 있을 수 있으면 `딕셔너리[key]` 대신 `딕셔너리.get(key)` 를 쓴다.
