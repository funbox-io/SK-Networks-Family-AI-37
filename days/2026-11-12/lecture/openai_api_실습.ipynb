#Chatbot API 실습

#환경 설정

```python
from dotenv import load_dotenv
from openai import OpenAI
```

```python
# KEY 설정
env_path = ".env"
output = load_dotenv(dotenv_path=env_path)
print(output)
```

```python
#OpenAI 서버 연결
client = OpenAI()
print(client)
```

#Chat API(completions) 호출

```python
#Chat API(completions) 호출
response = client.chat.completions.create(
    model="gpt-5-nano",
    messages=[
        #시스템(역할) 프롬프트(role) - 모델에게 지시하는 프롬프트
        {"role": "system", "content": "너는 친절하고 정확한 날씨 안내를 해주는 AI야"},

        #사용자() 프롬프트(role) - 모델에게 질문하는 프롬프트
        {"role": "user", "content": "최대 200 토큰 내에서 대규모 언어 모델에 대해서 알려줘"}
    ]
)
print(f"응답 확인:\n {response}")
```

```python
#생성된 답변(텍스트 추출)
response_text = response.choices[0].message.content
print(f"AI 답변:\n {response_text}")
```

#Web Search API 호출

```python
#Web Search 검색 기능 추가
#일반적인 웹 검색
response = client.responses.create(
    model="gpt-5-nano",
    tools=[{"type": "web_search_preview"}],
    #시스템 프롬프트
    instructions="너는 리서치 전문가야, 사용자 요청에 대해서 친절한 답변해",
    #사용자 프롬프트
    inputs="hot 뉴스 요약해서 알려줘"
)
response_text = response.output_text
print(f"웹 응답 확인:\n {response_text}")
```
