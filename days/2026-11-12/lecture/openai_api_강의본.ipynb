# 챗봇(챗팅) API

## 환경 설정

```python
# 필요한 함수 / 클래스 임포트
from dotenv import load_dotenv
from openai import OpenAI
```

```python
# .env file에서 API Key 가져오기
env_file_path='.env'
output = load_dotenv(env_file_path)
print(output)
```

```python
# OpenAI Server와의 연결 통로 생성
client = OpenAI()
print(client)
```

## Chat Completions API
- 채팅(QA) 기능 제공

```python
# Chat Completions API(채팅 기능 완성 API) 호출 -> 함수
response = client.chat.completions.create(
    model='gpt-5-nano',
    # 웹 버전: 사용자의 요청 사항 작성 -> 텍스트 -> 처리
    # 사용자 요쳥 -> 시스템 프롬프트, 사용자 프롬프트 각각 작성
    messages=[
        {'role':'system', 'content':'너는 친절하고 정확한 답변을 하는 AI 비서야'},
        {'role':'user', 'content':'최대 200 토큰 내에서 대규모 언어 모델에 대해서 알려줘'}
    ]
)

print(f'응답 확인: \n{response}')
```

```python
# 새성된 답변(텍스트 추출)
response_text = response.choices[0].message.content
print(f'AI 답변: \n{response_text}')
```

## Responses API
- 웹 검색 기능 추가

```python
# Response API 호출 - 일반적인 웹 검색 실행
response = client.responses.create(
    model='gpt-5-nano',
    # 웹 검색 도구 활성화
    tools=[{"type": "web_search_preview"}],
    # 시스템 프롬프트: instructions 매개 변수로 전달
    instructions="너는 리서치 전문가이며, 사용자 요청에 대해서 친절하고 정확한 답변을 해야 해",
    # 사용자 프롬프트: input 매개 변수로 전달
    input="오늘 주요 글로벌 IT 및 AI 관련 핵심 뉴스를 3가지로 요약해 줘" 
)

print(f'응답 확인: \n{response}')
```

```python
# API 응답 -> 텍스트 추출
text = response.output_text
print(text)
```

```python
#Response API 호출

#사용자 프롬프트
prompt = """
https://platform.openai.com/docs/api-reference/responses/create
를 읽어서 Responses API에 대해 300 토큰 이내에서 요약 정리해주세요.
"""
response = client.responses.create(
    model = "gpt-5-nano",
    tools = [{"type": "web_search_preview"}],
    #시스템 프롬프트
    instructions = "리서치 전문가이며, 사용자 요청에 대해서 친절하고 정확한 답변을 해야 해",
    #사용자 프롬프트
    input = prompt
)

text = response.output_text
print(f"응답 확인\n {text}")
```
