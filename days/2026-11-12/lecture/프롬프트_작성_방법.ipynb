```python
#openai API 클라이언트 생성
client = OpenAI()
```

## chat comletion API 호출


```python
"""
python API 실행 호출 예시
"""
def get_completion(prompt,model="gpt-5-nano"):
    #채팅 완성 기능 API 호출 -> 응답 생성
    response = client.chat.completions.create(
        model=model,
        #system prompt: 모델의 역할과 행동을 정의하는 메시지
        messages=[
            #시스템 프롬프트
            {"role": "system", "content": "너는 사용자의 답변에 말해주는 도우미야."},
            #사용자 프롬프트 
            {"role": "user", "content": prompt}
        ],
        # 응답 형식 생성
        response_format={'type': 'json_object'}
    )
    return response.choices[0].message.content

```

```python
#사용자 프롬프트 작성:  JSON 형식으로 응답 생성
user_prompt = """
3개의 가상의 책 제목, 저자, 장르를 생성해 줘.
아래 키를 사용하여 json 형식으로 제공해야되
book_id,title,author,genre
"""
```

```python
#Chat Completion API 실행 함수 호출
output = get_completion(prompt = user_prompt)
print(f"LLM 응답:{output}")
```
