````python
#모델 연결
client = OpenAI()
````

````python
#chat completions API 
def get_completion(prompt,model="gpt-5-nano"):
    #API 호출,응답 생성
    response = client.chat.completions.create(
        model = model,
        messages = [
            {'role':'system','content':'너는 한글로 된 고객 리뷰에 대한 경력 10년 이상의 우수한 감성 분석 전문가야. JSON 형식으로 응답해.'},
            {'role':'user','content':prompt}
        ],
        response_format={'type':'json_object'}
    )
    return response.choices[0].message.content
````

````python
#고객 리뷰 텍스트 생성 및 사용자 프롬프트 생성

#고객 리뷰 텍스트 생성
customer_review = f"""
바지 밑위가 길고 제 허벅지는 터져나가서 상의를 넣어 입고 가만있을 때 정면모습이 쪼금 애매한 주름이 지네요
그래도 착용감은 편안하고 밝은 바지인데도 부한 느낌도 없어서 마음에 들어요
크림진이 하나도 없어서 고민고민하다 구매했는데 이정도면 만족입니다
다만 허리에 있는 가죽패치가 검정이라 세탁시 이염될까봐 무서운거랑 바지밑단 주름이 좀 자글자글하게 지는 단점은 있어요!
"""

# 정답 예시를 생성
example = {"감성 분석 결과":"긍정"}

# 사용자 프롬프트
user_prompt = f"""
triple backticks로 처리된 한글 text를 분석해서 <정답 예시> 처럼 긍정인지 부정인지
json 형식으로 제시해줘
한글(text): ```{customer_review}```

<정답 예시> 
{example} 
"""
print(user_prompt)
````

````python
#chat completions API 실행 함수 호출
output1 = get_completion(prompt=user_prompt,model="gpt-5-nano")
print(f"LLM 출력:\n {output1}")
````
